### Dataset Quality Overview

A function is created to provide an initial data-quality summary for each dataset.

For each dataset, the function will reports:

- No. of rows and columns;
- The data type of each column;
- The number of missing values;
- The percentage of values that are missing;
- The number of unique values; and
- The number of exact duplicate rows.

This provides a consistent initial assessment before any cleaning decisions are made.

```python
def dataset_overview(df, name):
    print(f"DATASET: {name}")
    print("=" * 60)

    print(f"Rows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")

    overview = pd.DataFrame({
        "data_type": df.dtypes,
        "missing_values": df.isna().sum(),
        "missing_percentage": (df.isna().mean() * 100).round(2),
        "unique_values": df.nunique(dropna=True)
    })

    display(overview)

    print("Exact duplicate rows:", df.duplicated().sum())
```

#### 2026 Fixtures

```python
dataset_overview(fixtures_clean, "2026 Fixtures")
```

#### 2026 teams

```python
dataset_overview(teams_clean, "2026 Teams")
```

#### World Cup Editions 

```python
dataset_overview(editions_clean, "World Cup Editions")
```

#### World Cup Matches

```python
dataset_overview(matches_clean, "World Cup Matches")
```

#### World Cup top scorers

```python
dataset_overview(top_scorers_clean, "World Cup Top Scorers")
```

## Investigation into the Missing Values in 2026 Fixtures

Several fixture-related columns contain exactly 32 missing values. This will first be inspected to determine the appropriate decision.

```python
fixtures_clean[
    fixtures_clean["team1_fifa_rank"].isna()
][
    [
        "group",
        "stage",
        "team1",
        "team2",
        "date",
        "team1_confederation",
        "team1_fifa_rank",
        "team1_coach",
        "team2_confederation",
        "team2_fifa_rank",
        "team2_coach"
    ]
]
```

#### Conclusion 
Inspection of the 32 records containing missing team information showed that the missing values are structural rather than data-quality errors.

All affected records correspond to knockout-stage fixtures. The `team1` and `team2` columns contain tournament progression placeholders such as `1A`, `2B`, `Best 3rd #1`, `R32 W1`, `QF1`, `SF1` and `Finalist 1`, rather than confirmed team names.

Because the actual participating teams depend on the results of earlier tournament matches, attributes such as confederation, FIFA ranking and coach cannot be assigned to these placeholders.

The `group` field is also missing for these records because knockout-stage matches do not belong to a World Cup group.

Therefore:

- the 32 fixture records will be retained;
- the missing `group` values will be retained as not applicable to knockout-stage fixtures;
- missing confederation, FIFA ranking and coach values will not be imputed;
- tournament progression placeholders in `team1` and `team2` will be preserved;
- no rows will be removed because of these missing values.

This prevents artificial information from being introduced into the dataset and preserves the original tournament structure.

### Counting missing information by tournament stage

```python
fixtures_clean.groupby("stage").agg(
    total_fixtures=("stage", "size"),
    missing_group=("group", lambda x: x.isna().sum()),
    missing_team1_rank=("team1_fifa_rank", lambda x: x.isna().sum()),
    missing_team2_rank=("team2_fifa_rank", lambda x: x.isna().sum())
)
```

#### Conclusion 

Grouping the fixtures by tournament stage confirmed that the missing values follow a clear structural pattern.

All 72 Group Stage fixtures contain complete `group`, `team1_fifa_rank` and `team2_fifa_rank` information.

In contrast, all 32 knockout-stage fixtures contain missing values in these fields:

- 16 Round of 32 fixtures;
- 8 Round of 16 fixtures;
- 4 Quarter-final fixtures;
- 2 Semi-final fixtures;
- 1 Third Place Match; and
- 1 Final.

These 32 fixtures account for all of the missing values identified during the initial data-quality assessment.

This confirms that the missing values are associated with knockout-stage tournament placeholders rather than being randomly missing data. The values will therefore be retained as missing and will not be imputed or used as a reason to remove the fixture records.


### Checking the 'group' missing values separately

#### Validate Missing Group Values

The `group` column contains 32 missing values.

Although the previous checks suggest that these values are associated with knockout-stage fixtures, the `group` column is examined separately to confirm that the missing values are expected rather than data-quality errors.

First, all records with a missing `group` value are displayed to identify the tournament stages and fixture placeholders associated with them.

The number of missing `group` values is then calculated for each tournament stage. This provides a concise numerical confirmation of whether missing group information occurs exclusively outside the Group Stage.

```python
fixtures_clean[
    fixtures_clean["group"].isna()
][["group", "stage", "team1", "team2"]]
```

```python
fixtures_clean.groupby("stage")["group"].apply(
    lambda x: x.isna().sum()
)
```

#### Conclusion

Inspection of records with missing `group` values confirmed that all 32 occur in knockout-stage fixtures.

The missing values begin with the Round of 32 and continue through the Round of 16, Quarter-finals, Semi-finals, Third Place Match and Final.

This is expected because the `group` variable only applies during the Group Stage. Once teams progress to the knockout stages, fixtures are no longer associated with a World Cup group.

Therefore, the missing `group` values represent **not applicable** rather than unknown or incorrectly recorded information.

The values will remain missing rather than being imputed or replaced with an artificial group category.

### Inspecting team1 and team2 - Validating Fixture Team Labels Against the Teams Dataset

The `team1` and `team2` columns contain both named national teams and tournament progression placeholders.

The unique values in both columns are first inspected to understand how teams and future knockout-stage participants are represented.

The combined fixture labels are then compared with the 48 teams contained in `wc_2026_teams.csv`.

This cross-dataset validation helps identify whether unmatched values represent inconsistent team names, unexpected records or legitimate tournament progression placeholders.

```python
sorted(fixtures_clean["team1"].unique())
```

```python
sorted(fixtures_clean["team2"].unique())
```

```python
fixture_teams = set(fixtures_clean["team1"]).union(
    set(fixtures_clean["team2"])
)

teams_dataset_names = set(teams_clean["team"])

unmatched_fixture_labels = sorted(
    fixture_teams - teams_dataset_names
)

print("Number of fixture labels not found in the teams dataset:")
print(len(unmatched_fixture_labels))

print("\nLabels:")
print(unmatched_fixture_labels)
```

#### Conclusion

The cross-dataset comparison identified 63 fixture labels that are not present among the 48 named teams currently recorded in the 2026 teams dataset.

Inspection showed that all 63 unmatched values represent tournament progression placeholders rather than named national teams.

The placeholders include:

- group-position labels such as `1A` to `1L` and `2A` to `2L`;
- qualifying third-place placeholders such as `Best 3rd #1` to `Best 3rd #8`;
- Round of 32 progression labels such as `R32 W1` to `R32 W16`;
- later knockout-stage placeholders such as `QF1` to `QF8` and `SF1` to `SF4`; and
- final-stage placeholders including `3rd Place`, `Finalist 1` and `Finalist 2`.

These values encode the tournament pathway for fixtures whose participating teams depend on earlier results. They are therefore meaningful structural information and should not be removed, renamed as countries or replaced with missing values.

No additional named national teams were identified among the 63 unmatched fixture labels. However, because the `wc_2026_teams.csv` dataset has not yet been independently cleaned and validated, this comparison is treated as a preliminary cross-dataset consistency check rather than a definitive validation of team-name consistency.

No fixture team names will be changed on the basis of this preliminary comparison. The tournament progression placeholders will be retained as supplied.

A full cross-dataset comparison of team names and shared team attributes will be performed after both the fixtures and teams datasets have been independently cleaned and validated.

### Checking text columns for Leading and Trailing Whitespace

Text columns are checked for leading or trailing whitespace.

Unnecessary whitespace can cause identical values to be treated as different categories and may create problems when datasets are joined using team names or other categorical variables.

The check is performed before modifying the values so that any required cleaning can be justified.

```python
fixture_text_columns = [
    "group",
    "stage",
    "team1",
    "team2",
    "venue",
    "city",
    "country",
    "kickoff_et",
    "team1_confederation",
    "team1_coach",
    "team2_confederation",
    "team2_coach"
]

for col in fixture_text_columns:
    non_missing = fixtures_clean[col].dropna()

    whitespace_count = (
        non_missing.astype(str)
        != non_missing.astype(str).str.strip()
    ).sum()

    print(f"{col}: {whitespace_count}")
```

#### Conclusion

No leading or trailing whitespace was identified in any of the inspected text columns.

This includes the group, stage, team, venue, city, country, kick-off time, confederation and coach fields.

Therefore, no whitespace cleaning is required for these columns and the existing text values will be retained.

### Validate Fixture Dates

The `date` column is currently stored as an object rather than a datetime value.

Before converting the column, the dates are parsed into a temporary series using `errors="coerce"`.

Any value that cannot be interpreted as a valid date will become a missing value (`NaT`) in the temporary result. This allows invalid dates to be identified before the original column is changed.

```python
fixture_dates_test = pd.to_datetime(
    fixtures_clean["date"],
    errors="coerce"
)

print("Invalid dates:", fixture_dates_test.isna().sum())
print("Earliest date:", fixture_dates_test.min())
print("Latest date:", fixture_dates_test.max())
```

#### Conclusion

All fixture dates were successfully parsed as valid datetime values, with zero invalid dates identified.

The fixture dates range from 11 June 2026 to 19 July 2026.

No invalid or unparseable date values were identified. Therefore, the `date` column can safely be converted from an object data type to a datetime data type.

### Validating Coach Values

The `team1_coach` and `team2_coach` columns will be investigated. 

The available coach values are now inspected within the fixtures dataset to identify potentially inconsistent or unusual values.

```python
# Checking how many unique coaches appear
print(
    "Unique Team 1 coaches:",
    fixtures_clean["team1_coach"].nunique(dropna=True)
)

print(
    "Unique Team 2 coaches:",
    fixtures_clean["team2_coach"].nunique(dropna=True)
)
```

```python
print("Team 1 coaches:")
print(
    fixtures_clean["team1_coach"]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nTeam 2 coaches:")
print(
    fixtures_clean["team2_coach"]
    .value_counts(dropna=False)
    .sort_index()
)
```

```python
# Checking whether the same named team has multiple coach values 
team1_coach_check = (
    fixtures_clean
    .dropna(subset=["team1_coach"])
    .groupby("team1")["team1_coach"]
    .nunique()
)

team2_coach_check = (
    fixtures_clean
    .dropna(subset=["team2_coach"])
    .groupby("team2")["team2_coach"]
    .nunique()
)

print("Team 1 teams with more than one coach value:")
print(team1_coach_check[team1_coach_check > 1])

print("\nTeam 2 teams with more than one coach value:")
print(team2_coach_check[team2_coach_check > 1])
```

#### Conclusion

Both `team1_coach` and `team2_coach` contain 48 unique non-missing coach names. The remaining 32 values in each column are missing and were previously confirmed to correspond to knockout-stage fixtures where the participating teams have not yet been determined.

The frequency distributions show that coach names are repeated across Group Stage fixtures as expected because each named team appears in multiple fixtures.

A team-level consistency check was also performed to determine whether the same named team was associated with more than one coach value. No teams with multiple coach values were identified in either the `team1` or `team2` position.

Therefore, no internal inconsistencies requiring correction were identified in the coach columns at this stage. The existing coach values will be retained, and the structurally missing values will not be imputed.

### Validating Venue, City and Country Combinations

The `venue`, `city` and `country` columns contain no missing values.

These location fields are now examined to identify potentially inconsistent venue-to-city or city-to-country relationships within the fixtures dataset.

A consistent relationship is important because the same venue or city should not be associated with conflicting locations unless there is a valid reason.

```python
fixture_locations = (
    fixtures_clean[
        ["venue", "city", "country"]
    ]
    .drop_duplicates()
    .sort_values(
        ["country", "city", "venue"]
    )
)

print(
    "Number of unique venue-city-country combinations:",
    len(fixture_locations)
)

display(fixture_locations)
```

```python
# Checking for conflicted locations i.e. a venue being associated with multiple cities or the same city label associated with multiple countries. 

venue_city_check = (
    fixtures_clean
    .groupby("venue")["city"]
    .nunique()
)

city_country_check = (
    fixtures_clean
    .groupby("city")["country"]
    .nunique()
)

print("Venues associated with more than one city:")
print(venue_city_check[venue_city_check > 1])

print("\nCities associated with more than one country:")
print(city_country_check[city_country_check > 1])
```

Output consists of empty series which means no such conflicts found within this dataset.

#### Conclusion

A total of 16 unique venue-city-country combinations were identified. This corresponds with the 16 unique venues and 16 unique cities identified during the initial dataset inspection.

No venue was associated with more than one city, and no city was associated with more than one country.

The location information is therefore internally consistent within the fixtures dataset, and no conflicting venue-to-city or city-to-country relationships were identified.

No cleaning or standardisation of these location fields is required on the basis of this check.

This validation establishes consistency within the dataset but does not independently verify the factual accuracy of every venue, city and country assignment.

Most consecutive World Cup editions in the dataset are separated by four years, but there is a 12-year interval between the 1938 and 1950 tournaments.

Because this differs from the normal four-year tournament cycle, this interval was investigated rather than automatically being treated as missing data.

This discovery highlights why prior research/domain knowledge into FIFA and football matches is so important before diving into any data work. 
Historically, we know that World Cup matches take place every 4 years so the 12-year interval betwen 1938 and 1950 is important to investigate into which wouldnt be possible by just going through all the datasets given. 

After research we found historical information from FIFA confirms that the World Cup experienced a 12-year hiatus following the 1938 tournament because of the Second World War. The competition subsequently resumed in 1950.

The absence of tournament editions between 1938 and 1950 therefore represents a genuine historical interruption rather than missing records in the dataset.

No additional tournament rows should be created or imputed for this period.

#### Conclusion

The edition and tournament-year variables are internally consistent.

The dataset contains 22 unique edition numbers and 22 unique tournament years, with no duplicate values in either field.

Edition numbers form a continuous sequence from 1 to 22, while tournament years range from 1930 to 2022.

The interval between consecutive editions is four years throughout the dataset except between the 1938 and 1950 tournaments, where a 12-year interval is present.

This unusual interval was investigated rather than automatically being treated as a missing-data problem. FIFA historical information confirms that the World Cup experienced a 12-year hiatus following the 1938 tournament because of the Second World War, with the competition resuming in 1950.

The gap is therefore a genuine feature of World Cup history rather than evidence of missing tournament records.

No rows will be added or imputed for the years between 1938 and 1950, and no changes are required to the `edition` or `year` columns.

### Validating Tournament Start and End Dates

The `start_date` and `end_date` columns are currently stored as object values containing the month and day of each tournament date, while the corresponding year is stored separately in the `year` column.

Before modifying the original columns, complete date strings are temporarily constructed by combining each start and end date with its corresponding tournament year.

These temporary values are then parsed as datetime values using `errors="coerce"`. Any date that cannot be successfully interpreted will therefore become `NaT`, allowing invalid dates to be identified before a transformation is applied.

The relationship between the parsed start and end dates is also checked to confirm that no tournament is recorded as ending before it begins.

```python
# Constructing and parsing the dates temporarily.
edition_start_dates_test = pd.to_datetime(
    editions_clean["start_date"]
    + " "
    + editions_clean["year"].astype(str),
    errors="coerce"
)

edition_end_dates_test = pd.to_datetime(
    editions_clean["end_date"]
    + " "
    + editions_clean["year"].astype(str),
    errors="coerce"
)

print(
    "Invalid start dates:",
    edition_start_dates_test.isna().sum()
)

print(
    "Invalid end dates:",
    edition_end_dates_test.isna().sum()
)
```

```python
# Checking the relationship between start and end.
date_order_check = pd.DataFrame({
    "year": editions_clean["year"],
    "original_start_date": editions_clean["start_date"],
    "original_end_date": editions_clean["end_date"],
    "parsed_start_date": edition_start_dates_test,
    "parsed_end_date": edition_end_dates_test
})

date_order_check["end_before_start"] = (
    date_order_check["parsed_end_date"]
    < date_order_check["parsed_start_date"]
)

display(date_order_check)

print(
    "\nTournaments ending before they start:",
    date_order_check["end_before_start"].sum()
)
```

```python
# Checking tournament duration 
tournament_duration_test = (
    edition_end_dates_test - edition_start_dates_test
).dt.days

print(tournament_duration_test.describe())

display(
    pd.DataFrame({
        "year": editions_clean["year"],
        "start_date": edition_start_dates_test,
        "end_date": edition_end_dates_test,
        "duration_days": tournament_duration_test
    })
)
```

#### Conclusion

All 22 tournament start dates and all 22 tournament end dates were successfully parsed after combining the existing month-and-day values with the corresponding tournament year.

No invalid start or end dates were identified.

The chronological relationship between the two date fields was also validated. No tournament has an end date occurring before its start date.

The temporary duration calculation produced tournament durations ranging from 14 to 32 days. No impossible or internally contradictory date relationships were identified.

These checks confirm that the date values are structurally valid and can be represented as complete datetime values.

However, this validation establishes internal consistency rather than independently confirming the historical accuracy of every tournament date. Historical information will be considered again during the later cross-dataset validation stage.

Because the year is currently stored separately from the month-and-day values in `start_date` and `end_date`, converting these columns to complete datetime values will provide a more useful and unambiguous representation for later analysis.

### Validating Tournament Finishing Positions

The four recorded teams within an individual edition are therefore checked for duplicate values. A repeated team within the same edition would indicate an internal inconsistency because one team should not occupy more than one of these finishing positions in the same tournament.

This check evaluates the internal relationship between the finishing-position fields.

```python
finishing_position_columns = [
    "champion",
    "runner_up",
    "third_place",
    "fourth_place"
]

finishing_position_check = editions_clean[
    ["year"] + finishing_position_columns
].copy()

finishing_position_check["unique_finishing_teams"] = (
    finishing_position_check[finishing_position_columns]
    .nunique(axis=1)
)

finishing_position_check["duplicate_finishing_team"] = (
    finishing_position_check["unique_finishing_teams"] < 4
)

display(finishing_position_check)

print(
    "\nEditions with a team appearing in more than one finishing position:",
    finishing_position_check["duplicate_finishing_team"].sum()
)
```

#### Conclusion

### Validating the Host-Winner Indicator

The `host_won` variable records whether the host nation won each World Cup edition.

Its internal consistency is checked against the `host` and `champion` fields rather than assuming that the existing `Yes` and `No` values are correct.

Because the dataset includes a jointly hosted tournament, host values containing `/` are separated into their individual host countries before comparison. This ensures that the validation method can correctly handle both single-host and multi-host tournaments.

The calculated result is used only for validation at this stage. The existing `host_won` values will not be changed unless a discrepancy is identified and investigated.

```python
def calculate_host_won(row):
    hosts = [
        host.strip()
        for host in row["host"].split("/")
    ]

    return "Yes" if row["champion"] in hosts else "No"


calculated_host_won = editions_clean.apply(
    calculate_host_won,
    axis=1
)

host_won_check = editions_clean[
    ["year", "host", "champion", "host_won"]
].copy()

host_won_check["calculated_host_won"] = calculated_host_won

host_won_check["matches_recorded_value"] = (
    host_won_check["host_won"]
    == host_won_check["calculated_host_won"]
)

display(host_won_check)

print(
    "\nHost-won discrepancies:",
    (~host_won_check["matches_recorded_value"]).sum()
)
```

#### Investigation of Host-Winner Discrepancies

The internal validation identified two editions where the recorded `host_won` value disagreed with the relationship between the `host` and `champion` fields:

- 1930: Uruguay is recorded as both host and champion, but `host_won` is recorded as `No`.
- 1938: France is recorded as host and Italy as champion, but `host_won` is recorded as `Yes`.

These inconsistencies could not be resolved through these datasets alone, therefore the two cases were investigated using FIFA historical information by researching/domain knowledge.

FIFA confirms that Uruguay hosted and won the 1930 World Cup. The 1930 `host_won` value should therefore be `Yes`.

FIFA also confirms that France hosted the 1938 World Cup but Italy won the tournament. The 1938 `host_won` value should therefore be `No`.

These two records are therefore confirmed data-quality errors in the `host_won` field rather than legitimate exceptions.

The two values will be corrected while leaving the associated `host` and `champion` values unchanged.

```python
editions_clean.loc[
    editions_clean["year"] == 1930,
    "host_won"
] = "Yes"

editions_clean.loc[
    editions_clean["year"] == 1938,
    "host_won"
] = "No"
```

```python
# Revalidating after correction 
calculated_host_won_after_correction = editions_clean.apply(
    calculate_host_won,
    axis=1
)

host_won_recheck = editions_clean[
    ["year", "host", "champion", "host_won"]
].copy()

host_won_recheck["calculated_host_won"] = (
    calculated_host_won_after_correction
)

host_won_recheck["matches_recorded_value"] = (
    host_won_recheck["host_won"]
    == host_won_recheck["calculated_host_won"]
)

display(host_won_recheck)

print(
    "\nRemaining host-won discrepancies:",
    (~host_won_recheck["matches_recorded_value"]).sum()
)

print("\nUpdated host_won counts:")
print(editions_clean["host_won"].value_counts())
```

#### Conclusion

The `host_won` field was validated by independently comparing the recorded host nation or nations with the recorded tournament champion.

The initial validation identified two inconsistent records:

- 1930 recorded Uruguay as both host and champion but incorrectly recorded `host_won` as `No`;
- 1938 recorded France as host and Italy as champion but incorrectly recorded `host_won` as `Yes`.

Both discrepancies were investigated using FIFA historical information and confirmed to be data-quality errors in the `host_won` field.

The 1930 value was corrected from `No` to `Yes`, while the 1938 value was corrected from `Yes` to `No`.

Following these corrections, the validation was repeated across all 22 editions. No remaining discrepancies were identified between the recorded and independently calculated host-winner indicators.

The final distribution remains 16 `No` values and 6 `Yes` values because the two corrections occurred in opposite directions.

The `host_won` field is now internally consistent with the `host` and `champion` fields.

The `goals`, `goals_per_match` and `attendance` variables were examined across all 22 World Cup editions to identify unusual values and large edition-to-edition changes.

Considerable historical variation is present. However, this is expected because the number of participating teams, number of matches, tournament structure, host countries and scale of the competition have changed over time.

The most notable values were investigated rather than automatically being treated as outliers or data-quality errors.

The 1954 tournament has the highest recorded goals-per-match value in the dataset at 5.38. Historical FIFA information confirms that the 1954 World Cup remains the highest-scoring edition in terms of average goals per match.

The 1994 tournament records the highest attendance in the dataset at 3,587,538. FIFA historical information confirms this cumulative attendance figure and identifies the 1994 World Cup as having established a tournament attendance record.

The 2022 tournament records the highest total number of goals in the dataset at 172. FIFA confirms that 172 goals were scored across its 64 matches, which established a new World Cup goalscoring record at that time.

The edition-to-edition changes therefore do not provide evidence of erroneous values. They reflect genuine historical variation in tournament size, scoring patterns and attendance.

No changes are required to `goals`, `goals_per_match` or `attendance` on the basis of this validation.

```python
# Inspecting before changing it. 
west_germany_bulgaria_1970 = matches_clean.loc[
    (matches_clean["year"] == 1970)
    & (matches_clean["team1"] == "West Germany")
    & (matches_clean["team2"] == "Bulgaria")
]

display(west_germany_bulgaria_1970)
```

#### Correction of the 1970 West Germany vs Bulgaria Record

Further investigation confirmed that the unusual West Germany vs Bulgaria record contains errors in both the score and match date.

The dataset records the match as West Germany 3–10 Bulgaria on 2 June 1970.

Historical validation confirms that the World Cup group-stage match was played on 7 June 1970 in León, Mexico, with West Germany defeating Bulgaria 5–2.

The teams, tournament stage, venue, city and country are consistent with the historical match, so only the incorrect score and date fields require correction.

The record will therefore be corrected as follows:

- `score1`: 3 → 5
- `score2`: 10 → 2
- `date`: 2 June 1970 → 7 June 1970

```python
match_1970_mask = (
    (matches_clean["year"] == 1970)
    & (matches_clean["team1"] == "West Germany")
    & (matches_clean["team2"] == "Bulgaria")
)

matches_clean.loc[
    match_1970_mask,
    "score1"
] = 5

matches_clean.loc[
    match_1970_mask,
    "score2"
] = 2

matches_clean.loc[
    match_1970_mask,
    "date"
] = pd.Timestamp("1970-06-07")
```

```python
# Verifying correction
display(
    matches_clean.loc[
        match_1970_mask,
        [
            "year",
            "stage",
            "team1",
            "score1",
            "score2",
            "team2",
            "venue",
            "city",
            "country",
            "date",
            "notes"
        ]
    ]
)
```

Note: Further invesigation needed but a few notes on these venue-location conflicts and why they may be there: 

- Estadio Nacional — Brasília/Santiago, Brazil/Chile: this can legitimately occur because Estadio Nacional is a generic stadium name used for different stadiums in different countries. So the fact that the name occurs in both Chile and Brazil does not itself mean either record is wrong.

- Estadio Sausalito — Vina del Mar / Viña del Mar: this looks like a text-standardisation issue.

- Olympiastadion — Berlin/Munich and Germany/West Germany: there are genuinely Olympiastadion venues in both Berlin and Munich, and the country difference may reflect historical naming because 1974 records use West Germany whereas 2006 records use Germany. So this isn't automatically an error.

- Stanford Stadium — Palo Alto/Stanford: as discussed, this is likely a locality-label variation rather than an incorrect stadium. We should preserve it unless our later standardisation policy gives us a reason to choose one representation.

  
- Volksparkstadion — Hamburg, Germany/West Germany: same city and stadium; the country difference is historical because your 1974 records use West Germany and your 2006 records use Germany. We should not replace historical West Germany with Germany merely to make this test pass.

#### Creating Direction-Independent Team-Pair Keys

The two datasets use different team-position fields:

- `team1` and `team2` in the World Cup match dataset
- `home_team` and `away_team` in the international results dataset

The same historical match may therefore have the teams recorded in a different order.

To avoid treating reversed team order as a different match, temporary direction-independent team-pair keys are created by alphabetically sorting the two team names.

These keys are used only for validation and do not modify the original team fields.

```python
matches_validation = matches_clean.copy()
reference_validation = world_cup_reference.copy()

matches_validation["team_pair"] = matches_validation.apply(
    lambda row: " | ".join(
        sorted([row["team1"], row["team2"]])
    ),
    axis=1
)

reference_validation["team_pair"] = reference_validation.apply(
    lambda row: " | ".join(
        sorted([row["home_team"], row["away_team"]])
    ),
    axis=1
)

display(
    matches_validation[
        ["year", "team1", "team2", "team_pair"]
    ].head()
)

display(
    reference_validation[
        ["year", "home_team", "away_team", "team_pair"]
    ].head()
)
```

#### Testing Match-Key Uniqueness

Before using tournament year and team pair as a match identifier, the combination is tested for uniqueness in both datasets.

If the same two teams played each other more than once during the same World Cup, year and team pair alone would not safely identify an individual match.

Any repeated combinations are therefore inspected before the datasets are merged.

```python
selected_key_duplicates = (
    matches_validation
    .groupby(["year", "team_pair"])
    .size()
    .reset_index(name="record_count")
)

selected_key_duplicates = selected_key_duplicates.loc[
    selected_key_duplicates["record_count"] > 1
]

reference_key_duplicates = (
    reference_validation
    .groupby(["year", "team_pair"])
    .size()
    .reset_index(name="record_count")
)

reference_key_duplicates = reference_key_duplicates.loc[
    reference_key_duplicates["record_count"] > 1
]

print(
    "Repeated year + team-pair combinations "
    "in wc_all_matches:",
    len(selected_key_duplicates)
)

display(selected_key_duplicates)

print(
    "\nRepeated year + team-pair combinations "
    "in results_clean FIFA World Cup records:",
    len(reference_key_duplicates)
)

display(reference_key_duplicates)
```

```python
# Inspecting all repeated pairs 
reference_repeated_records = (
    reference_validation
    .merge(
        reference_key_duplicates[
            ["year", "team_pair"]
        ],
        on=["year", "team_pair"],
        how="inner"
    )
    [
        [
            "year",
            "home_team",
            "home_score",
            "away_score",
            "away_team",
            "date",
            "city",
            "country",
            "team_pair"
        ]
    ]
    .sort_values(
        ["year", "team_pair", "date"]
    )
)

display(reference_repeated_records)
```

#### Correction Identified During Repeated Match-Key Investigation

Inspection of repeated tournament-year and team-pair combinations identified an incorrect 1930 Group Stage record.

The dataset recorded:

`Argentina 6–3 USA` on `1930-07-26`

Historical verification showed that Argentina's 6–3 Group Stage victory was against Mexico on 19 July 1930.

Argentina did play the USA on 26 July 1930, but this was the separate semi-final, which Argentina won 6–1.

The Group Stage record is therefore corrected from:

- opponent: USA → Mexico
- date: 1930-07-26 → 1930-07-19

The 6–3 score is retained because it correctly represents the Argentina–Mexico match.

The separate Argentina 6–1 USA semi-final record is retained unchanged as those are correct. 

```python
argentina_1930_group_mask = (
    (matches_clean["year"] == 1930)
    & (matches_clean["stage"] == "Group Stage")
    & (matches_clean["team1"] == "Argentina")
    & (matches_clean["score1"] == 6)
    & (matches_clean["score2"] == 3)
    & (matches_clean["team2"] == "USA")
)

print(
    "Records identified for correction:",
    argentina_1930_group_mask.sum()
)

display(
    matches_clean.loc[
        argentina_1930_group_mask
    ]
)
```

```python
# Applying correction
matches_clean.loc[
    argentina_1930_group_mask,
    "team2"
] = "Mexico"

matches_clean.loc[
    argentina_1930_group_mask,
    "date"
] = pd.Timestamp("1930-07-19")
```

```python
# Verifying changes 
argentina_1930_check = matches_clean.loc[
    (matches_clean["year"] == 1930)
    & (matches_clean["team1"] == "Argentina"),
    [
        "year",
        "stage",
        "team1",
        "score1",
        "score2",
        "team2",
        "date",
        "venue",
        "city",
        "country"
    ]
].sort_values("date")

display(argentina_1930_check)

print(
    "\nIncorrect Argentina 6-3 USA records remaining:",
    (
        (matches_clean["year"] == 1930)
        & (matches_clean["team1"] == "Argentina")
        & (matches_clean["score1"] == 6)
        & (matches_clean["score2"] == 3)
        & (matches_clean["team2"] == "USA")
    ).sum()
)

print("Date dtype:", matches_clean["date"].dtype)
```

```python
# Rebuilding validation copy 
matches_validation = matches_clean.copy()
```

```python
matches_validation = matches_clean.copy()

matches_validation["team_pair"] = matches_validation.apply(
    lambda row: " | ".join(
        sorted([row["team1"], row["team2"]])
    ),
    axis=1
)

print(
    "Repeated year + team-pair combinations "
    "after correction:"
)

updated_key_counts = (
    matches_validation
    .groupby(["year", "team_pair"])
    .size()
    .reset_index(name="record_count")
)

updated_key_duplicates = updated_key_counts.loc[
    updated_key_counts["record_count"] > 1
]

display(updated_key_duplicates)
```

#### Identifying Team-Name Differences Before Historical Matching

Before the historical match records are compared, team labels are inspected across the two datasets.

Equivalent national teams may be represented using different names, particularly because the datasets cover a long historical period. These naming differences could cause genuine matches to appear unmatched even when the underlying records agree.

The original team fields are not modified at this stage.

Instead, differences are identified first so that temporary validation-only mappings can be created where equivalence can be justified. Permanent naming standardisation will be performed later as a project-wide process across all relevant datasets.

#### Creating Validation-Only Team-Name Equivalences

Inspection of historical team naming showed that the two datasets use different conventions for several national teams.

For the purpose of historical match validation only, equivalent labels are temporarily mapped to a common comparison name:

- `USA` and `United States`
- `Ireland` and `Republic of Ireland`
- `East Germany` and `German DR`
- `West Germany` and the reference dataset's retrospective `Germany` label
- `Soviet Union` and the reference dataset's retrospective `Russia` label

These mappings are not applied to the original team columns.

In particular, historically meaningful names such as `West Germany`, `East Germany` and `Soviet Union` are retained in the cleaned World Cup dataset. The temporary mappings are used only to allow equivalent historical match records to be identified across datasets.

Permanent naming standardisation will be considered separately at the project-wide standardisation stage.

```python
# Create fresh validation copies
matches_validation = matches_clean.copy()
reference_validation = reference_historical.copy()


def standardise_team_for_validation(team, year):
    """
    Standardise equivalent team labels for historical
    cross-dataset validation only.
    
    This does not alter the original cleaned datasets.
    """

    # USA
    if team in ["USA", "United States"]:
        return "United States"

    # Republic of Ireland
    if team in ["Ireland", "Republic of Ireland"]:
        return "Republic of Ireland"

    # East Germany
    if team in ["East Germany", "German DR"]:
        return "East Germany"

    # West Germany / retrospective Germany label
    if team in ["West Germany", "Germany"] and year <= 1990:
        return "West Germany"

    # Soviet Union / retrospective Russia label
    if team in ["Soviet Union", "Russia"] and year <= 1990:
        return "Soviet Union"

    return team
```

```python
matches_validation["team1_validation"] = (
    matches_validation.apply(
        lambda row: standardise_team_for_validation(
            row["team1"],
            row["year"]
        ),
        axis=1
    )
)

matches_validation["team2_validation"] = (
    matches_validation.apply(
        lambda row: standardise_team_for_validation(
            row["team2"],
            row["year"]
        ),
        axis=1
    )
)

reference_validation["home_team_validation"] = (
    reference_validation.apply(
        lambda row: standardise_team_for_validation(
            row["home_team"],
            row["year"]
        ),
        axis=1
    )
)

reference_validation["away_team_validation"] = (
    reference_validation.apply(
        lambda row: standardise_team_for_validation(
            row["away_team"],
            row["year"]
        ),
        axis=1
    )
)
```

```python
# Verifying what the temporary mapping changed
selected_mapping_changes = pd.concat(
    [
        matches_validation.loc[
            matches_validation["team1"]
            != matches_validation["team1_validation"],
            ["year", "team1", "team1_validation"]
        ].rename(
            columns={
                "team1": "original_name",
                "team1_validation": "validation_name"
            }
        ),

        matches_validation.loc[
            matches_validation["team2"]
            != matches_validation["team2_validation"],
            ["year", "team2", "team2_validation"]
        ].rename(
            columns={
                "team2": "original_name",
                "team2_validation": "validation_name"
            }
        )
    ]
).drop_duplicates().sort_values(
    ["original_name", "year"]
)

display(selected_mapping_changes)
```

```python
reference_mapping_changes = pd.concat(
    [
        reference_validation.loc[
            reference_validation["home_team"]
            != reference_validation["home_team_validation"],
            ["year", "home_team", "home_team_validation"]
        ].rename(
            columns={
                "home_team": "original_name",
                "home_team_validation": "validation_name"
            }
        ),

        reference_validation.loc[
            reference_validation["away_team"]
            != reference_validation["away_team_validation"],
            ["year", "away_team", "away_team_validation"]
        ].rename(
            columns={
                "away_team": "original_name",
                "away_team_validation": "validation_name"
            }
        )
    ]
).drop_duplicates().sort_values(
    ["original_name", "year"]
)

display(reference_mapping_changes)
```

#### Refining the Historical Team-Name Validation Rules

Inspection of Germany-related records showed that `wc_all_matches` uses historically meaningful naming conventions:

- `Germany` in 1934
- `West Germany` from 1954 to 1990
- `East Germany` for the German Democratic Republic in 1974
- `Germany` again from 1994 onwards

The initial validation rule was therefore too broad because it temporarily converted the 1934 German team to `West Germany`.

The validation-only mapping is refined so that `Germany` is treated as equivalent to `West Germany` only for World Cups from 1954 through 1990.

This refinement affects only temporary comparison fields and does not modify either cleaned source dataset.

```python
def standardise_team_for_validation(team, year):
    """
    Standardise equivalent team labels for historical
    cross-dataset validation only.

    This does not alter the original cleaned datasets.
    """

    # USA
    if team in ["USA", "United States"]:
        return "United States"

    # Republic of Ireland
    if team in ["Ireland", "Republic of Ireland"]:
        return "Republic of Ireland"

    # East Germany
    if team in ["East Germany", "German DR"]:
        return "East Germany"

    # West Germany period
    if team in ["West Germany", "Germany"] and 1954 <= year <= 1990:
        return "West Germany"

    # Soviet Union period represented retrospectively
    # as Russia in the reference dataset
    if team in ["Soviet Union", "Russia"] and 1958 <= year <= 1990:
        return "Soviet Union"

    return team
```

```python
# Recreating the validation columns 
matches_validation = matches_clean.copy()
reference_validation = reference_historical.copy()

matches_validation["team1_validation"] = (
    matches_validation.apply(
        lambda row: standardise_team_for_validation(
            row["team1"],
            row["year"]
        ),
        axis=1
    )
)

matches_validation["team2_validation"] = (
    matches_validation.apply(
        lambda row: standardise_team_for_validation(
            row["team2"],
            row["year"]
        ),
        axis=1
    )
)

reference_validation["home_team_validation"] = (
    reference_validation.apply(
        lambda row: standardise_team_for_validation(
            row["home_team"],
            row["year"]
        ),
        axis=1
    )
)

reference_validation["away_team_validation"] = (
    reference_validation.apply(
        lambda row: standardise_team_for_validation(
            row["away_team"],
            row["year"]
        ),
        axis=1
    )
)
```

```python
# Verifying the corrected historical mapping 
validation_mapping_check = pd.DataFrame({
    "input_name": [
        "Germany",
        "Germany",
        "West Germany",
        "Germany",
        "German DR",
        "Russia",
        "Russia",
        "USA",
        "Ireland"
    ],
    "year": [
        1934,
        1954,
        1974,
        1994,
        1974,
        1986,
        1994,
        1930,
        1990
    ]
})

validation_mapping_check["validation_name"] = (
    validation_mapping_check.apply(
        lambda row: standardise_team_for_validation(
            row["input_name"],
            row["year"]
        ),
        axis=1
    )
)

display(validation_mapping_check)
```

```python
# Recreating the keys from validation names 

matches_validation["validation_team_pair"] = (
    matches_validation.apply(
        lambda row: " | ".join(
            sorted([
                row["team1_validation"],
                row["team2_validation"]
            ])
        ),
        axis=1
    )
)

reference_validation["validation_team_pair"] = (
    reference_validation.apply(
        lambda row: " | ".join(
            sorted([
                row["home_team_validation"],
                row["away_team_validation"]
            ])
        ),
        axis=1
    )
)

print("Validation keys recreated successfully.")

display(
    matches_validation[
        [
            "year",
            "team1",
            "team2",
            "validation_team_pair"
        ]
    ].head(10)
)
```

#### Matching Historical Records by Tournament Year and Team Pair

After accounting for justified historical naming differences, the selected World Cup match records are compared with the historical FIFA World Cup reference data.

The initial comparison uses only:

- tournament year
- validation-standardised team pair

Dates and scores are deliberately excluded from the initial match criteria because these fields are themselves being validated.

For each selected World Cup record, the number of possible reference records with the same tournament year and team pair is calculated.

This separates records into:

- no candidate historical match;

Records with multiple candidates require additional matching logic because the same teams genuinely played each other more than once during some World Cups.

```python
candidate_counts = (
    reference_validation
    .groupby(
        ["year", "validation_team_pair"]
    )
    .size()
    .reset_index(name="reference_candidate_count")
)

matches_validation = matches_validation.merge(
    candidate_counts,
    on=["year", "validation_team_pair"],
    how="left"
)

matches_validation["reference_candidate_count"] = (
    matches_validation["reference_candidate_count"]
    .fillna(0)
    .astype(int)
)

print("Total selected records:", len(matches_validation))

print(
    "\nCandidate match counts:"
)

display(
    matches_validation[
        "reference_candidate_count"
    ]
    .value_counts()
    .sort_index()
    .rename_axis("number_of_reference_candidates")
    .reset_index(name="selected_records")
)
```

```python
# Inspecting records with no candidate 
no_candidate_matches = matches_validation.loc[
    matches_validation["reference_candidate_count"] == 0,
    [
        "year",
        "stage",
        "team1",
        "score1",
        "score2",
        "team2",
        "date",
        "venue",
        "city",
        "country",
        "validation_team_pair"
    ]
].sort_values(["year", "date"])

print(
    "Records with no historical team-pair candidate:",
    len(no_candidate_matches)
)

display(no_candidate_matches)
```

#### Comparing Match Dates With Historical Candidates

All 184 selected World Cup records have at least one historical reference candidate with the same tournament year and team pair.

The next validation stage compares the recorded match dates.

Each selected record is joined to every historical reference record with the same tournament year and validation-standardised team pair. A date-agreement flag is then calculated.

For team pairs that met only once during a tournament, this provides a direct date comparison.

For team pairs that met more than once, all possible historical candidates are retained initially. This prevents an arbitrary candidate from being selected and allows the recorded date to identify the corresponding historical fixture where possible.

A date disagreement is treated as a validation flag rather than an automatic correction.

```python
candidate_comparison = (
    matches_validation
    .reset_index()
    .rename(columns={"index": "selected_record_id"})
    .merge(
        reference_validation[
            [
                "year",
                "validation_team_pair",
                "date",
                "home_team",
                "home_score",
                "away_score",
                "away_team",
                "city",
                "country"
            ]
        ],
        on=["year", "validation_team_pair"],
        how="left",
        suffixes=("_selected", "_reference")
    )
)

candidate_comparison["date_matches"] = (
    candidate_comparison["date_selected"]
    == candidate_comparison["date_reference"]
)

print(
    "Candidate comparison rows:",
    len(candidate_comparison)
)

print("\nDate agreement across candidate rows:")

display(
    candidate_comparison["date_matches"]
    .value_counts(dropna=False)
    .rename_axis("date_matches")
    .reset_index(name="candidate_rows")
)
```

```python
# Determining whether each selected record has a date match 
date_validation_summary = (
    candidate_comparison
    .groupby("selected_record_id")
    .agg(
        reference_candidates=(
            "date_reference",
            "size"
        ),
        matching_date_candidates=(
            "date_matches",
            "sum"
        )
    )
    .reset_index()
)

date_validation_summary["has_matching_date"] = (
    date_validation_summary[
        "matching_date_candidates"
    ] > 0
)

print("Selected records checked:", len(date_validation_summary))

display(
    date_validation_summary[
        "has_matching_date"
    ]
    .value_counts()
    .rename_axis("has_matching_date")
    .reset_index(name="selected_records")
)
```

```python
# Finding records whose dates do not agree 
date_mismatches = (
    matches_validation
    .reset_index()
    .rename(columns={"index": "selected_record_id"})
    .merge(
        date_validation_summary.loc[
            ~date_validation_summary["has_matching_date"],
            ["selected_record_id"]
        ],
        on="selected_record_id",
        how="inner"
    )
)

print(
    "Selected records with no matching historical date:",
    len(date_mismatches)
)

display(
    date_mismatches[
        [
            "selected_record_id",
            "year",
            "stage",
            "team1",
            "score1",
            "score2",
            "team2",
            "date",
            "venue",
            "city",
            "country"
        ]
    ].sort_values(["year", "date"])
)
```

#### Validating Match Scores Against Date-Identified Historical Fixtures

After resolving the verified date discrepancies, all 184 selected World Cup match records can be associated with a historical reference fixture using:

- tournament year;
- validation-standardised team pair; and
- match date.

The next stage validates the recorded scores.

Because the two datasets do not always list the teams in the same order, reference scores cannot be compared directly using home and away columns. The reference score is therefore aligned to the `team1` and `team2` orientation used in the selected World Cup dataset before comparison.

This allows genuine score discrepancies to be identified without incorrectly flagging records simply because the teams are listed in reverse order.

Any score disagreement will be investigated before a correction is made.

```python
matched_fixtures = candidate_comparison.loc[
    candidate_comparison["date_matches"]
].copy()

print("Date-identified reference matches:", len(matched_fixtures))

print(
    "Selected records represented:",
    matched_fixtures["selected_record_id"].nunique()
)

print(
    "Selected records appearing more than once:",
    matched_fixtures["selected_record_id"].duplicated().sum()
)
```

```python
# Aligning reference scores to team1 and team2 - accounting for reversed team order. 
def align_reference_scores(row):
    """
    Align the historical reference score to the team1/team2
    orientation used in wc_all_matches.
    """

    if (
        row["team1_validation"] == row["home_team_validation"]
        and
        row["team2_validation"] == row["away_team_validation"]
    ):
        return pd.Series({
            "reference_score1": row["home_score"],
            "reference_score2": row["away_score"],
            "reference_orientation": "same"
        })

    elif (
        row["team1_validation"] == row["away_team_validation"]
        and
        row["team2_validation"] == row["home_team_validation"]
    ):
        return pd.Series({
            "reference_score1": row["away_score"],
            "reference_score2": row["home_score"],
            "reference_orientation": "reversed"
        })

    else:
        return pd.Series({
            "reference_score1": np.nan,
            "reference_score2": np.nan,
            "reference_orientation": "unresolved"
        })
```

```python
score_comparison = (
    matches_validation
    .reset_index()
    .rename(columns={"index": "selected_record_id"})
    .merge(
        reference_validation[
            [
                "year",
                "validation_team_pair",
                "date",
                "home_team",
                "away_team",
                "home_team_validation",
                "away_team_validation",
                "home_score",
                "away_score"
            ]
        ],
        left_on=[
            "year",
            "validation_team_pair",
            "date"
        ],
        right_on=[
            "year",
            "validation_team_pair",
            "date"
        ],
        how="left"
    )
)

aligned_scores = score_comparison.apply(
    align_reference_scores,
    axis=1
)

score_comparison = pd.concat(
    [
        score_comparison,
        aligned_scores
    ],
    axis=1
)
```

```python
# Checking every reference score was aligned 
print(
    "Score comparison records:",
    len(score_comparison)
)

print(
    "\nReference orientation:"
)

display(
    score_comparison[
        "reference_orientation"
    ]
    .value_counts(dropna=False)
    .rename_axis("orientation")
    .reset_index(name="records")
)

print(
    "\nUnresolved orientations:",
    (
        score_comparison["reference_orientation"]
        == "unresolved"
    ).sum()
)
```

```python
# Comparing all 184 scores 
score_comparison["score_matches"] = (
    (score_comparison["score1"] == score_comparison["reference_score1"])
    &
    (score_comparison["score2"] == score_comparison["reference_score2"])
)

display(
    score_comparison[
        "score_matches"
    ]
    .value_counts()
    .rename_axis("score_matches")
    .reset_index(name="selected_records")
)
```

```python
# Isolating any discrepancies 
score_mismatches = score_comparison.loc[
    ~score_comparison["score_matches"],
    [
        "selected_record_id",
        "year",
        "stage",
        "team1",
        "score1",
        "score2",
        "team2",
        "date",
        "reference_score1",
        "reference_score2",
        "reference_orientation",
        "home_team",
        "home_score",
        "away_score",
        "away_team"
    ]
].sort_values(["year", "date"])

print(
    "Records with score discrepancies:",
    len(score_mismatches)
)

display(score_mismatches)
```

#### Correcting Verified Historical Score Discrepancies

Systematic score validation identified six selected World Cup match records whose recorded scores disagreed with their date-identified historical reference fixtures.

Each discrepancy was independently checked against FIFA historical information rather than automatically treating the comparison dataset as authoritative.

The following corrections were verified:

- 1962: Italy 0–0 Chile → Italy 0–2 Chile
- 1982: West Germany 0–0 France → West Germany 3–3 France
- 1986: France 0–0 Brazil → France 1–1 Brazil
- 2002: South Korea 2–0 Portugal → South Korea 1–0 Portugal
- 2014: Argentina 0–0 Belgium → Argentina 1–0 Belgium
- 2018: Argentina 1–3 France → Argentina 3–4 France

For matches decided by penalty shoot-outs, `score1` and `score2` continue to represent the match score before the shoot-out. Shoot-out outcomes, where available, remain supplementary information in the `notes` field.

Only the verified score fields are corrected. The original raw dataset remains unchanged.





```python
score_corrections = [
    {
        "year": 1962,
        "team1": "Italy",
        "team2": "Chile",
        "score1": 0,
        "score2": 2
    },
    {
        "year": 1982,
        "team1": "West Germany",
        "team2": "France",
        "score1": 3,
        "score2": 3
    },
    {
        "year": 1986,
        "team1": "France",
        "team2": "Brazil",
        "score1": 1,
        "score2": 1
    },
    {
        "year": 2002,
        "team1": "South Korea",
        "team2": "Portugal",
        "score1": 1,
        "score2": 0
    },
    {
        "year": 2014,
        "team1": "Argentina",
        "team2": "Belgium",
        "score1": 1,
        "score2": 0
    },
    {
        "year": 2018,
        "team1": "Argentina",
        "team2": "France",
        "score1": 3,
        "score2": 4
    }
]
```

```python
# Confirming each correction identifies exactly one row. Thiw was done so that corrections are in a controlled way. 
for correction in score_corrections:

    mask = (
        (matches_clean["year"] == correction["year"])
        & (matches_clean["team1"] == correction["team1"])
        & (matches_clean["team2"] == correction["team2"])
    )

    print(
        correction["year"],
        correction["team1"],
        "vs",
        correction["team2"],
        "- records identified:",
        mask.sum()
    )
```

```python
# All 6 have returned 1 so will run the correction. 
for correction in score_corrections:

    mask = (
        (matches_clean["year"] == correction["year"])
        & (matches_clean["team1"] == correction["team1"])
        & (matches_clean["team2"] == correction["team2"])
    )

    matches_clean.loc[
        mask,
        ["score1", "score2"]
    ] = [
        correction["score1"],
        correction["score2"]
    ]
```

```python
# Verifying the 6 corrected records. 
corrected_score_records = matches_clean.loc[
    (
        (matches_clean["year"] == 1962)
        & (matches_clean["team1"] == "Italy")
        & (matches_clean["team2"] == "Chile")
    )
    |
    (
        (matches_clean["year"] == 1982)
        & (matches_clean["team1"] == "West Germany")
        & (matches_clean["team2"] == "France")
    )
    |
    (
        (matches_clean["year"] == 1986)
        & (matches_clean["team1"] == "France")
        & (matches_clean["team2"] == "Brazil")
    )
    |
    (
        (matches_clean["year"] == 2002)
        & (matches_clean["team1"] == "South Korea")
        & (matches_clean["team2"] == "Portugal")
    )
    |
    (
        (matches_clean["year"] == 2014)
        & (matches_clean["team1"] == "Argentina")
        & (matches_clean["team2"] == "Belgium")
    )
    |
    (
        (matches_clean["year"] == 2018)
        & (matches_clean["team1"] == "Argentina")
        & (matches_clean["team2"] == "France")
    ),
    [
        "year",
        "stage",
        "team1",
        "score1",
        "score2",
        "team2",
        "date",
        "notes"
    ]
].sort_values("year")

display(corrected_score_records)
```

#### Revalidating Match Scores After Verified Corrections

The six independently verified score discrepancies have now been corrected in the cleaned World Cup match dataset.

The historical validation is rerun from the corrected `matches_clean` dataset rather than manually altering the previous comparison results.

This confirms whether all 184 selected match records now agree with their date-identified historical reference fixtures after accounting for differences in team ordering and historical naming conventions.

```python
# Rebuilding the validation copy. 
matches_validation = matches_clean.copy()

matches_validation["team1_validation"] = (
    matches_validation.apply(
        lambda row: standardise_team_for_validation(
            row["team1"],
            row["year"]
        ),
        axis=1
    )
)

matches_validation["team2_validation"] = (
    matches_validation.apply(
        lambda row: standardise_team_for_validation(
            row["team2"],
            row["year"]
        ),
        axis=1
    )
)

matches_validation["validation_team_pair"] = (
    matches_validation.apply(
        lambda row: " | ".join(
            sorted([
                row["team1_validation"],
                row["team2_validation"]
            ])
        ),
        axis=1
    )
)
```

```python
# Recreating the score comparison 
score_comparison = (
    matches_validation
    .reset_index()
    .rename(columns={"index": "selected_record_id"})
    .merge(
        reference_validation[
            [
                "year",
                "validation_team_pair",
                "date",
                "home_team",
                "away_team",
                "home_team_validation",
                "away_team_validation",
                "home_score",
                "away_score"
            ]
        ],
        on=[
            "year",
            "validation_team_pair",
            "date"
        ],
        how="left"
    )
)

aligned_scores = score_comparison.apply(
    align_reference_scores,
    axis=1
)

score_comparison = pd.concat(
    [
        score_comparison,
        aligned_scores
    ],
    axis=1
)
```

```python
# Checking the matching structure again 
print("Selected records:", len(matches_clean))
print("Score comparison records:", len(score_comparison))

print(
    "Unresolved orientations:",
    (
        score_comparison["reference_orientation"]
        == "unresolved"
    ).sum()
)

print(
    "Missing reference scores:",
    score_comparison[
        ["reference_score1", "reference_score2"]
    ].isna().any(axis=1).sum()
)
```

```python
# Rerunning the actual score check
score_comparison["score_matches"] = (
    (score_comparison["score1"] == score_comparison["reference_score1"])
    &
    (score_comparison["score2"] == score_comparison["reference_score2"])
)

display(
    score_comparison[
        "score_matches"
    ]
    .value_counts(dropna=False)
    .rename_axis("score_matches")
    .reset_index(name="selected_records")
)

print(
    "Remaining score discrepancies:",
    (~score_comparison["score_matches"]).sum()
)
```

Following correction and revalidation of the participating teams, match dates and scores, the location information is compared with the corresponding historical reference records.

The comparison focuses first on `city` and `country`.

Venue names are not used as a direct cross-dataset comparison field because stadium naming conventions can vary considerably between datasets and across time. Venue information has already undergone a separate historical audit.

City and country disagreements are therefore treated as validation flags rather than automatic errors. Any disagreement will be investigated before changes are made.

```python
# Rebuilding the validation copy 

matches_validation = matches_clean.copy()

matches_validation["team1_validation"] = (
    matches_validation.apply(
        lambda row: standardise_team_for_validation(
            row["team1"],
            row["year"]
        ),
        axis=1
    )
)

matches_validation["team2_validation"] = (
    matches_validation.apply(
        lambda row: standardise_team_for_validation(
            row["team2"],
            row["year"]
        ),
        axis=1
    )
)

matches_validation["validation_team_pair"] = (
    matches_validation.apply(
        lambda row: " | ".join(
            sorted([
                row["team1_validation"],
                row["team2_validation"]
            ])
        ),
        axis=1
    )
)

```

### Final Correction Audit

Before completing the independent cleaning of the World Cup matches dataset, the original raw working dataframe is compared with `matches_clean`.

This provides a complete audit trail of all values changed during cleaning and historical validation.

The purpose of this check is to confirm:

- which records were modified;
- which columns were modified;
- the original and cleaned values;
- that no unintended fields were changed; and
- that the number of changes is consistent with the corrections identified during the cleaning process.

This comparison is performed before the final dataset quality checks.

```python
changes = []

for column in matches.columns:

    raw_values = matches[column]
    clean_values = matches_clean[column]

    # Treat two missing values as equal
    different = ~(
        raw_values.eq(clean_values)
        | (raw_values.isna() & clean_values.isna())
    )

    for idx in matches.index[different]:

        changes.append({
            "record_id": idx,
            "year": matches.loc[idx, "year"],
            "team1": matches.loc[idx, "team1"],
            "team2": matches.loc[idx, "team2"],
            "column": column,
            "original_value": matches.loc[idx, column],
            "cleaned_value": matches_clean.loc[idx, column]
        })

changes_audit = pd.DataFrame(changes)

display(
    changes_audit.sort_values(
        ["record_id", "column"]
    )
)

print("Total changed cells:", len(changes_audit))
print("Records containing at least one change:",
      changes_audit["record_id"].nunique())
```

```python
# Summarising changes by column
change_summary = (
    changes_audit["column"]
    .value_counts()
    .rename_axis("column")
    .reset_index(name="changed_cells")
)

display(change_summary)

print(
    "Check total:",
    change_summary["changed_cells"].sum()
)
```

```python
expected_changed_columns = {
    "team1",
    "team2",
    "score1",
    "score2",
    "date",
    "venue",
    "city"
}

unexpected_changed_columns = sorted(
    set(changes_audit["column"])
    - expected_changed_columns
)

print("Unexpected changed columns:", unexpected_changed_columns)
```

```python
# Checking which records changed 
changed_records = (
    changes_audit[
        [
            "record_id",
            "year",
            "team1",
            "team2"
        ]
    ]
    .drop_duplicates()
    .sort_values(["year", "record_id"])
)

display(changed_records)

print(
    "Number of changed match records:",
    len(changed_records)
)
```

### Assessing the Reliability of Assists and Penalties

Initial inspection showed that the `assists` and `penalties` fields have very limited variation.

Nineteen of the 22 assist values are recorded as zero, while all 22 penalty values are zero. This raises the possibility that some zero values may represent unavailable or unpopulated information rather than confirmed statistical zeroes.

Before these fields are used analytically, their reliability must therefore be assessed.

The values are investigated separately from the `goals` field, which has already been successfully validated across all 22 editions.

```python
# Identifying the pattern

assist_penalty_check = top_scorers_clean[
    [
        "year",
        "player",
        "goals",
        "assists",
        "penalties"
    ]
].copy()

assist_penalty_check["has_assist_data"] = (
    assist_penalty_check["assists"] > 0
)

assist_penalty_check["has_penalty_data"] = (
    assist_penalty_check["penalties"] > 0
)

display(assist_penalty_check)

print(
    "Editions with non-zero assists:",
    assist_penalty_check["has_assist_data"].sum()
)

print(
    "Editions with non-zero penalties:",
    assist_penalty_check["has_penalty_data"].sum()
)
```

```python
# Checking if zero penalities can represent genuine values. 
penalty_validation_examples = top_scorers_clean[
    top_scorers_clean["year"].isin([1966, 2018])
][
    [
        "year",
        "player",
        "goals",
        "penalties"
    ]
]

display(penalty_validation_examples)
```

```python
# Checking assist pattern by period 
assist_coverage_by_period = top_scorers_clean[
    [
        "year",
        "player",
        "assists"
    ]
].copy()

assist_coverage_by_period["period"] = np.where(
    assist_coverage_by_period["year"] < 2010,
    "Before 2010",
    "2010 onwards"
)

display(
    assist_coverage_by_period.groupby("period").agg(
        editions=("year", "count"),
        zero_assists=("assists", lambda x: (x == 0).sum()),
        non_zero_assists=("assists", lambda x: (x > 0).sum())
    )
)
```

#### Assessing Historical Assist Coverage

The dataset contains tournament-level top-scorer records rather than individual match records.

Seven of the 22 records relate to World Cups before 1966 (1930–1962). FIFA's historical assist statistics are recorded from 1966 onwards, meaning the pre-1966 zero values cannot automatically be interpreted in the same way as later recorded assist statistics.

Pre-1966 assist figures may be available through retrospective historical reconstruction, but these do not have the same provenance as FIFA's recorded assist statistics.

The assist field is therefore assessed according to its historical coverage before any cleaning decision is made.

```python
assist_penalty_validation = top_scorers_clean[
    [
        "year",
        "player",
        "assists",
        "penalties"
    ]
].copy()

assist_penalty_validation["assist_status"] = np.where(
    assist_penalty_validation["year"] < 1966,
    "Pre-1966 - requires retrospective verification",
    "1966 onwards - requires verification"
)

display(assist_penalty_validation)

print(
    "\nPre-1966 records:",
    (assist_penalty_validation["year"] < 1966).sum()
)

print(
    "1966 onwards records:",
    (assist_penalty_validation["year"] >= 1966).sum()
)
```

```python
# Demonstrating penalty data issue (Every penalty is 0)
penalty_verified_examples = pd.DataFrame({
    "year": [1966, 2018],
    "player": ["Eusébio", "Harry Kane"],
    "dataset_penalties": [0, 0],
    "fifa_verified_penalties": [4, 3]
})

penalty_verified_examples["values_match"] = (
    penalty_verified_examples["dataset_penalties"]
    == penalty_verified_examples["fifa_verified_penalties"]
)

display(penalty_verified_examples)
```

#### Conclusion

The `assists` and `penalties` fields do not currently provide consistently reliable statistical coverage across all 22 tournament records.

For assists, seven records relate to World Cups before 1966. These precede FIFA's recorded historical assist series and would require retrospective historical verification if assist statistics are required for the project.

From 1966 onwards, the dataset still shows an unusual coverage pattern. Non-zero assist values occur only in 2010, 2014 and 2022. Therefore, the remaining zero values cannot automatically be assumed to represent confirmed zero assists without further verification.

The `penalties` field is demonstrably unreliable as currently populated. All 22 records contain zero, despite verified examples in which top scorers scored penalties. For example, the dataset records zero for both Eusébio in 1966 and Harry Kane in 2018, despite verified penalty-goal totals of four and three respectively.

The existing `assists` and `penalties` values are retained temporarily rather than partially replacing individual records and creating an inconsistent mixture of original and reconstructed statistics.

These fields will not be used as reliable analytical variables unless a complete and consistently sourced reconstruction is performed during the later feature-engineering stage.

No values are changed during this step.

### Validating Team Result

The `team_result` field describes how the listed top scorer's national team performed in the corresponding World Cup.

This field is potentially relevant to the project because it links individual scoring performance with tournament-level team success.

The values are first checked for completeness and consistency. Results that can be directly validated against the independently cleaned `wc_all_editions` dataset are then compared using the recorded champion, runner-up, third-place and fourth-place teams.

Broader results such as `Quarter-final` and `Semi-final` require separate consideration because `wc_all_editions` does not contain a complete finishing position for every participating team.

```python
# Inspecting the categories 
print(
    "Missing team_result values:",
    top_scorers_clean["team_result"].isna().sum()
)

print(
    "Unique team_result values:",
    top_scorers_clean["team_result"].nunique()
)

print("\nTeam result counts:")

print(
    top_scorers_clean["team_result"]
    .value_counts(dropna=False)
)

print("\nUnique values:")

print(
    sorted(top_scorers_clean["team_result"].dropna().unique())
)
```

```python
# Inspecting each record
display(
    top_scorers_clean[
        [
            "year",
            "player",
            "country",
            "goals",
            "team_result"
        ]
    ].sort_values("year")
)
```

```python
# Validating top-four results
top_four_check = top_scorers_clean[
    top_scorers_clean["team_result"].isin(
        ["Champions", "Runner-up", "Third place", "Fourth place"]
    )
][
    [
        "edition",
        "year",
        "player",
        "country",
        "team_result"
    ]
].merge(
    editions_clean[
        [
            "edition",
            "year",
            "champion",
            "runner_up",
            "third_place",
            "fourth_place"
        ]
    ],
    on=["edition", "year"],
    how="left"
)

def expected_team_result(row):
    if row["country"] == row["champion"]:
        return "Champions"
    elif row["country"] == row["runner_up"]:
        return "Runner-up"
    elif row["country"] == row["third_place"]:
        return "Third place"
    elif row["country"] == row["fourth_place"]:
        return "Fourth place"
    else:
        return "No top-four match"

top_four_check["expected_result"] = top_four_check.apply(
    expected_team_result,
    axis=1
)

top_four_check["result_matches"] = (
    top_four_check["team_result"]
    == top_four_check["expected_result"]
)

display(top_four_check)

print(
    "\nMatching top-four results:",
    top_four_check["result_matches"].sum()
)

print(
    "Different top-four results:",
    (~top_four_check["result_matches"]).sum()
)
```

```python
# Re-running validation 
top_four_check = top_scorers_clean[
    top_scorers_clean["team_result"].isin(
        ["Champions", "Runner-up", "Third place", "Fourth place"]
    )
][
    [
        "edition",
        "year",
        "player",
        "country",
        "team_result"
    ]
].merge(
    editions_clean[
        [
            "edition",
            "year",
            "champion",
            "runner_up",
            "third_place",
            "fourth_place"
        ]
    ],
    on=["edition", "year"],
    how="left"
)

def expected_team_result(row):
    if row["country"] == row["champion"]:
        return "Champions"
    elif row["country"] == row["runner_up"]:
        return "Runner-up"
    elif row["country"] == row["third_place"]:
        return "Third place"
    elif row["country"] == row["fourth_place"]:
        return "Fourth place"
    else:
        return "No top-four match"

top_four_check["expected_result"] = top_four_check.apply(
    expected_team_result,
    axis=1
)

top_four_check["result_matches"] = (
    top_four_check["team_result"]
    == top_four_check["expected_result"]
)

display(top_four_check)

print(
    "\nMatching top-four results:",
    top_four_check["result_matches"].sum()
)

print(
    "Different top-four results:",
    (~top_four_check["result_matches"]).sum()
)
```

#### Conclusion

The `team_result` field contains no missing values and initially contained seven result categories.

### Validating Joint Top-Scorer Representation

The dataset contains exactly one record for each World Cup edition. However, some tournaments had multiple players sharing the highest goal total.

Earlier inspection identified different approaches to representing these situations. The 1962 tournament is represented using an aggregated `Multiple Winners` record, while the 1994 and 2010 tournaments contain a single named player.

The representation of joint top scorers is therefore compared with the independently cleaned `wc_all_editions` dataset.

The purpose of this step is to determine whether the one-row-per-edition structure accurately represents tied top-scorer situations and whether any changes are required before the dataset is used analytically.

```python
# Comparing scorer fields 
top_scorer_comparison = top_scorers_clean[
    [
        "edition",
        "year",
        "player",
        "country",
        "goals"
    ]
].merge(
    editions_clean[
        [
            "edition",
            "year",
            "top_scorer",
            "top_scorer_country",
            "top_scorer_goals"
        ]
    ],
    on=["edition", "year"],
    how="left"
)

display(top_scorer_comparison)
```

```python
joint_top_scorer_comparison = top_scorer_comparison[
    top_scorer_comparison["year"].isin(
        [1962, 1994, 2010]
    )
]

display(joint_top_scorer_comparison)
```

```python
# Checking whether the goal totals still agree 
joint_top_scorer_comparison = (
    joint_top_scorer_comparison.copy()
)

joint_top_scorer_comparison["goal_total_matches"] = (
    joint_top_scorer_comparison["goals"]
    == joint_top_scorer_comparison["top_scorer_goals"]
)

display(
    joint_top_scorer_comparison[
        [
            "year",
            "player",
            "top_scorer",
            "goals",
            "top_scorer_goals",
            "goal_total_matches"
        ]
    ]
)
```

Some things we have picked up on during this analysis: 

 - 1962: wc_top_scorers deliberately aggregates the six joint scorers into "Multiple Winners", while editions_clean lists the individual names.
- 1994: wc_top_scorers records only Hristo Stoichkov, while editions_clean records Hristo Stoichkov / Oleg Salenko.
- 2010: wc_top_scorers records only Thomas Müller, while editions_clean records all the tied top scorers.
- The goal totals themselves agree: 4, 6 and 5 respectively.

```python
# Quantifying the representation difference 
top_scorer_comparison["player_representation_matches"] = (
    top_scorer_comparison["player"]
    == top_scorer_comparison["top_scorer"]
)

print(
    "Exact player representation matches:",
    top_scorer_comparison["player_representation_matches"].sum()
)

print(
    "Different player representations:",
    (~top_scorer_comparison["player_representation_matches"]).sum()
)

display(
    top_scorer_comparison[
        ~top_scorer_comparison["player_representation_matches"]
    ][
        [
            "year",
            "player",
            "country",
            "top_scorer",
            "top_scorer_country",
            "goals",
            "top_scorer_goals"
        ]
    ]
)
```

#### Conclusion

Comparison with `wc_all_editions` confirms that the top-scorer goal total is consistent across all three editions in which multiple players shared the highest goal total.

However, the two datasets represent joint top scorers differently:

- In 1962, `wc_top_scorers` uses the aggregated label `Multiple Winners`, while `wc_all_editions` lists the individual joint top scorers.
- In 1994, `wc_top_scorers` records Hristo Stoichkov only, while `wc_all_editions` records Hristo Stoichkov and Oleg Salenko as joint top scorers.
- In 2010, `wc_top_scorers` records Thomas Müller only, while `wc_all_editions` records all players who shared the highest goal total.

The corresponding goal totals remain consistent between the datasets: 4 goals in 1962, 6 in 1994 and 5 in 2010.

No additional rows are created during cleaning. `wc_top_scorers` has a consistent one-row-per-edition structure across all 22 tournaments, and expanding only the tied editions would change the grain of the dataset and require additional player-level information that is not consistently available in the existing fields.

The existing records are therefore retained. The joint-scorer representation is documented as a limitation of the dataset.

If player-level top-scorer analysis is required later, a separate derived player-level dataset can be constructed from validated historical information without altering the original edition-level structure.

No values are changed during this step.

This provides an explicit audit trail and confirms that only verified corrections have been applied.

Validation-only investigations, such as the assessment of assists, penalties, joint top scorers and formatting differences, should not appear as changes because the original values were retained.

```python
# Finding every changed cell. We haven't changed any data types for this datasets so we can compare the raw and cleaned values directly. 
changed_cells = []

for column in top_scorers.columns:
    changed_mask = (
        top_scorers[column].fillna("<MISSING>").astype(str)
        !=
        top_scorers_clean[column].fillna("<MISSING>").astype(str)
    )

    for index in top_scorers.index[changed_mask]:
        changed_cells.append({
            "row_index": index,
            "year": top_scorers.loc[index, "year"],
            "player": top_scorers.loc[index, "player"],
            "column": column,
            "raw_value": top_scorers.loc[index, column],
            "clean_value": top_scorers_clean.loc[index, column]
        })

changes_audit = pd.DataFrame(changed_cells)

display(changes_audit)

print(
    "\nTotal changed cells:",
    len(changes_audit)
)
```

```python
# Counting changes by columns 
print("Changes by column:")

print(
    changes_audit["column"]
    .value_counts()
)
```

```python
# Counting changed records 
changed_records = (
    changes_audit[
        ["row_index", "year", "player"]
    ]
    .drop_duplicates()
    .sort_values(["year", "player"])
)

display(changed_records)

print(
    "\nTotal changed records:",
    len(changed_records)
)
```

```python
# Ensuring nothing else has changed for a controlled cleaning process 
expected_changed_columns = {
    "goals",
    "team_result"
}

actual_changed_columns = set(
    changes_audit["column"].unique()
)

print(
    "Expected changed columns:",
    expected_changed_columns
)

print(
    "Actual changed columns:",
    actual_changed_columns
)

print(
    "Only expected columns changed:",
    actual_changed_columns == expected_changed_columns
)
```

### `wc_top_scorers` Independent Cleaning Summary

Independent cleaning and validation of `wc_top_scorers` is now complete.

#### Dataset structure

The dataset contains 22 rows and 11 columns, representing one summary record for each FIFA World Cup edition from 1930 to 2022.

The final cleaned working copy retains the original 22 × 11 structure.

Final quality checks confirmed:

- 22 unique editions.
- 22 unique tournament years.
- No duplicate editions or years.
- No exact duplicate rows.
- No structurally missing values.
- No negative goal values.
- No zero or negative `matches_played` values.

#### Verified corrections

Three factual values were corrected following validation against independently cleaned tournament information and historical evidence:

- 1930 — Guillermo Stábile's `team_result` was corrected from `Champions` to `Runner-up`.
- 1934 — Oldřich Nejedlý's `team_result` was corrected from `Semi-final` to `Runner-up`.
- 1950 — Ademir's `goals` value was corrected from 8 to 9.

The correction audit confirmed exactly three changed cells across three records:

- `team_result`: 2 changes.
- `goals`: 1 change.

No other columns were modified.

#### Assists and penalties

The `assists` and `penalties` fields were investigated because their distributions suggested incomplete or unreliable historical coverage.

All 22 penalty values are recorded as zero, despite verified examples showing that some listed top scorers scored penalty goals during their tournaments. The existing `penalties` field therefore cannot currently be treated as a reliable analytical variable.

Assist coverage is also inconsistent. Seven records relate to tournaments before 1966, preceding FIFA's recorded historical assist series, while non-zero values in this dataset occur only in 2010, 2014 and 2022.

The original values are retained rather than partially reconstructing these fields from different sources. `assists` and `penalties` should not be used as reliable analytical variables unless complete and consistently sourced values are reconstructed later.

#### Matches played and position

`matches_played` contains no missing, zero or negative values and ranges from 4 to 7.

Individual-player values were found to be structurally plausible. However, the 1962 `Multiple Winners` record is an aggregated record, so its `matches_played` value should not be interpreted as the appearance total of a single player.

All 22 `position` values are recorded as `FW`. The field is retained, but because it contains no variation it is unlikely to provide useful discriminatory information for statistical analysis or modelling in its current form.

#### Host validation

Host information was compared with `wc_all_editions`.

Twenty-one of the 22 values matched exactly. The only difference was the representation of the joint hosts in 2002:

- `wc_top_scorers`: `South Korea/Japan`
- `wc_all_editions`: `South Korea / Japan`

This is a formatting difference rather than a factual error and is therefore deferred to the project-wide standardisation stage.

#### Team-result validation

Following the two verified corrections, all 18 records explicitly classified as champion, runner-up, third place or fourth place are consistent with `wc_all_editions`.

The remaining records use `Quarter-final`, `Semi-final` or `Varies`. These values are retained, although the field currently mixes final-placement terminology with stage-based terminology. This will be considered during later categorical standardisation and feature engineering.

#### Joint top scorers

The dataset consistently contains one row per World Cup edition, but joint top scorers are not represented consistently within that structure.

Three editions require particular consideration:

- 1962 uses the aggregated label `Multiple Winners`.
- 1994 records Hristo Stoichkov, although Hristo Stoichkov and Oleg Salenko shared the highest goal total.
- 2010 records Thomas Müller, although multiple players shared the highest goal total.

Comparison with `wc_all_editions` confirmed that the corresponding top goal totals remain consistent across these editions.

No additional rows were created because expanding the tied editions would change the dataset from one row per edition to one row per player and would require additional consistently validated player-level information.

If player-level top-scorer analysis is required later, a separate derived dataset can be constructed without changing the grain of this dataset.

#### Cleaning decision

`wc_top_scorers` is now considered independently cleaned and validated for the purposes of this stage.

The following issues remain intentionally deferred:

- project-wide country, host and text standardisation;
- reconstruction of reliable assists or penalty-goal statistics if required;
- potential standardisation of `team_result`;
- treatment of joint top scorers for any later player-level analysis;
- formal cross-dataset validation across the complete project.

The cleaned working dataframe remains `top_scorers_clean`.

The dataset is not exported yet. Final processed files will be saved only after project-wide standardisation, formal cross-dataset validation and final quality checks have been completed.

```python
# Comparing city names 
city_sets = {
    "results": set(
        results_clean["city"].dropna().astype(str)
    ),
    "2026_fixtures": set(
        fixtures_clean["city"].dropna().astype(str)
    ),
    "wc_matches": set(
        matches_clean["city"].dropna().astype(str)
    )
}

for dataset_name, cities in city_sets.items():
    print(
        f"{dataset_name}: {len(cities)} unique cities"
    )
```

```python
# Checking WC match cities against results_clean (as wc_matches contains only 184 selected WC records)
wc_cities = set(
    matches_clean["city"].dropna().astype(str)
)

results_cities = set(
    results_clean["city"].dropna().astype(str)
)

wc_cities_not_exactly_in_results = sorted(
    wc_cities - results_cities
)

print(
    "World Cup match cities without an exact match in results:",
    len(wc_cities_not_exactly_in_results)
)

print("\nCities:")
for city in wc_cities_not_exactly_in_results:
    print(city)
```

```python
# Normalised city comparison 
import unicodedata
import re

def normalise_text_for_comparison(value):

    if pd.isna(value):
        return None

    value = str(value)

    value = unicodedata.normalize(
        "NFKD",
        value
    )

    value = "".join(
        character
        for character in value
        if not unicodedata.combining(character)
    )

    value = value.lower()

    value = re.sub(
        r"[^a-z0-9]",
        "",
        value
    )

    return value
```

```python
results_city_lookup = {}

for city in sorted(results_cities):

    key = normalise_text_for_comparison(city)

    results_city_lookup.setdefault(
        key,
        []
    ).append(city)


potential_city_matches = []

for wc_city in wc_cities_not_exactly_in_results:

    key = normalise_text_for_comparison(
        wc_city
    )

    possible_matches = results_city_lookup.get(
        key,
        []
    )

    if possible_matches:

        potential_city_matches.append({
            "wc_matches_city": wc_city,
            "possible_results_city": possible_matches
        })


potential_city_matches_df = pd.DataFrame(
    potential_city_matches
)

display(potential_city_matches_df)
```

```python
# Verifying our known city differences 
known_city_variants = [
    "Vina del Mar",
    "Viña del Mar",
    "St Petersburg",
    "Saint Petersburg"
]

for city in known_city_variants:

    print(f"\n{city}")

    for dataset_name, df in {
        "results": results_clean,
        "2026_fixtures": fixtures_clean,
        "wc_matches": matches_clean
    }.items():

        if "city" in df.columns:

            count = (
                df["city"]
                .astype("string")
                .eq(city)
                .sum()
            )

            print(
                f"  {dataset_name:<15}: {count}"
            )
```

### Applying City Name Standardisation

City naming has now been compared across all datasets containing city information.

Two equivalent city representations in `wc_matches` were identified and validated:

- `Vina del Mar` → `Viña del Mar`
- `St Petersburg` → `Saint Petersburg`

These changes standardise spelling and naming conventions without changing the underlying geographic meaning.

The `Stanford` / `Palo Alto` difference is not changed because it represents a locality convention difference rather than a simple formatting inconsistency.

No city changes are required in `results`, `2026_fixtures` or `wc_editions`.

```python
# Counting values before change 
city_standardisation = {
    "Vina del Mar": "Viña del Mar",
    "St Petersburg": "Saint Petersburg"
}

print("Values before standardisation:")

for old_value, new_value in city_standardisation.items():

    old_count = (
        matches_clean["city"]
        .eq(old_value)
        .sum()
    )

    new_count = (
        matches_clean["city"]
        .eq(new_value)
        .sum()
    )

    print(
        f"{old_value} -> {new_value}: "
        f"{old_count} value(s) to change; "
        f"{new_count} already standardised"
    )
```

```python
# Standarisation that will only change the city column of matches clean
matches_clean["city"] = (
    matches_clean["city"]
    .replace(city_standardisation)
)
```

```python
# Verifying transformation 
print("Values after standardisation:")

for old_value, new_value in city_standardisation.items():

    print(
        f"\n{old_value}:",
        matches_clean["city"].eq(old_value).sum()
    )

    print(
        f"{new_value}:",
        matches_clean["city"].eq(new_value).sum()
    )
```

Location-country values were compared across all relevant datasets.

2. **Joint-host formatting**

3. **Historical country naming**
   - `West Germany` is retained as a historically meaningful location label rather than being globally replaced with `Germany`.
   - Historical entities will be handled separately from simple formatting and naming differences.

### Team-Name Standardisation Decisions

Cross-dataset investigation identified several different types of team-name differences.

The following project-wide naming conventions are selected for equivalent modern national-team labels:

- `USA` → `United States`
- `Czechia` → `Czech Republic`
- `Türkiye` → `Turkey`

These changes provide consistent analytical labels across the project while retaining the existing naming convention used by the larger historical datasets.

The 1990 `Ireland` record in `wc_matches` was separately validated against `results`. The corresponding FIFA World Cup fixture is recorded as `Republic of Ireland` versus Romania on 25 June 1990, with the same score and location. The `Ireland` value in this record will therefore be changed to `Republic of Ireland` (as former ends in 1956 so does not apply to the 1990 record).

Historically distinct national-team labels are preserved rather than globally mapped to modern successor states. These include:

- `East Germany`
- `West Germany`
- `Soviet Union`
- `Czechoslovakia`

The World Cup match data also contains `Germany` records from different historical periods. These records will not be globally rewritten during standardisation. If historical national-team lineage is required for later analysis, it should be represented using a separate derived analytical field rather than by overwriting the historical source labels.

```python
# Finding number of affected cells before performing transformation
team_standardisation_rules = {
    "USA": "United States",
    "Czechia": "Czech Republic",
    "Türkiye": "Turkey",
    "Ireland": "Republic of Ireland"
}

team_changes_before = []

for dataset_name, info in team_name_fields.items():

    df = info["dataframe"]

    for column in info["columns"]:

        for old_value, new_value in team_standardisation_rules.items():

            count = (
                df[column]
                .astype("string")
                .eq(old_value)
                .sum()
            )

            if count > 0:

                team_changes_before.append({
                    "dataset": dataset_name,
                    "column": column,
                    "old_value": old_value,
                    "new_value": new_value,
                    "cells_to_change": int(count)
                })

team_changes_before_df = pd.DataFrame(
    team_changes_before
)

display(team_changes_before_df)

print(
    "\nTotal team-name cells to change:",
    team_changes_before_df["cells_to_change"].sum()
)
```

```python
# Applying replacements only to team fields. 
for dataset_name, info in team_name_fields.items():

    df = info["dataframe"]

    for column in info["columns"]:

        df[column] = (
            df[column]
            .replace(team_standardisation_rules)
        )
```

```python
# Verifying old labels are gone 
for old_value, new_value in team_standardisation_rules.items():

    old_total = 0
    new_total = 0

    for dataset_name, info in team_name_fields.items():

        for column in info["columns"]:

            old_total += (
                info["dataframe"][column]
                .astype("string")
                .eq(old_value)
                .sum()
            )

            new_total += (
                info["dataframe"][column]
                .astype("string")
                .eq(new_value)
                .sum()
            )

    print(f"\n{old_value} -> {new_value}")
    print("Old label remaining:", old_total)
    print("New label total     :", new_total)
```

```python
# Verifying the historical labels mentioned are preserved 
historical_labels = [
    "East Germany",
    "West Germany",
    "Soviet Union",
    "Czechoslovakia"
]

for team in historical_labels:

    print(f"\n{team}")
    print("-" * 40)

    for dataset_name, info in team_name_fields.items():

        total_count = 0

        for column in info["columns"]:

            total_count += (
                info["dataframe"][column]
                .astype("string")
                .eq(team)
                .sum()
            )

        if total_count > 0:
            print(
                f"{dataset_name:<20}: {total_count}"
            )
```

### Inspecting Venue Naming and Location Consistency

Venue information is available in `2026_fixtures` and `wc_matches`.

Because these datasets primarily represent different World Cup editions, their venue lists are not expected to match. A direct requirement for identical venue names across the two datasets would therefore not be meaningful.

Venue validation instead focuses on:

- missing values;
- leading or trailing whitespace;
- duplicate naming representations;
- consistency between venues and their associated city/country values.

```python
# Venue quality checks 
venue_fields = {
    "2026_fixtures": fixtures_clean["venue"],
    "wc_matches": matches_clean["venue"]
}

for dataset_name, series in venue_fields.items():

    print(f"\n{dataset_name}")
    print("-" * 50)

    print(
        "Rows:",
        len(series)
    )

    print(
        "Missing venues:",
        series.isna().sum()
    )

    print(
        "Unique venues:",
        series.nunique(dropna=True)
    )

    whitespace_count = (
        series.dropna().astype(str)
        !=
        series.dropna().astype(str).str.strip()
    ).sum()

    print(
        "Whitespace issues:",
        whitespace_count
    )
```

```python
def normalise_venue_for_comparison(value):

    if pd.isna(value):
        return None

    return normalise_text_for_comparison(value)
```

```python
for dataset_name, df in {
    "2026_fixtures": fixtures_clean,
    "wc_matches": matches_clean
}.items():

    venue_variants = (
        df[["venue"]]
        .dropna()
        .drop_duplicates()
        .copy()
    )

    venue_variants["comparison_key"] = (
        venue_variants["venue"]
        .apply(normalise_venue_for_comparison)
    )

    possible_duplicates = (
        venue_variants
        .groupby("comparison_key")["venue"]
        .agg(list)
        .reset_index()
    )

    possible_duplicates = possible_duplicates[
        possible_duplicates["venue"]
        .apply(len) > 1
    ]

    print(f"\n{dataset_name}")
    print("-" * 50)

    if len(possible_duplicates) == 0:
        print(
            "No differently written venue names "
            "share the same normalised representation."
        )
    else:
        display(possible_duplicates)
```

```python
# Checking whether the same venue is associated with multiple cities. 
for dataset_name, df in {
    "2026_fixtures": fixtures_clean,
    "wc_matches": matches_clean
}.items():

    venue_city_counts = (
        df.groupby("venue")["city"]
        .nunique(dropna=True)
        .sort_values(ascending=False)
    )

    conflicting_venue_cities = (
        venue_city_counts[
            venue_city_counts > 1
        ]
    )

    print(f"\n{dataset_name}")
    print("-" * 50)

    print(
        "Venues associated with more than one city:",
        len(conflicting_venue_cities)
    )

    if len(conflicting_venue_cities) > 0:
        display(conflicting_venue_cities)
```

```python
# Doing the same for countries 
for dataset_name, df in {
    "2026_fixtures": fixtures_clean,
    "wc_matches": matches_clean
}.items():

    venue_country_counts = (
        df.groupby("venue")["country"]
        .nunique(dropna=True)
        .sort_values(ascending=False)
    )

    conflicting_venue_countries = (
        venue_country_counts[
            venue_country_counts > 1
        ]
    )

    print(f"\n{dataset_name}")
    print("-" * 50)

    print(
        "Venues associated with more than one country:",
        len(conflicting_venue_countries)
    )

    if len(conflicting_venue_countries) > 0:
        display(conflicting_venue_countries)
```

### Investigating Venue-Location Conflicts

Initial venue validation found no missing values, whitespace issues or obvious alternative spellings.

The 2026 fixture data contains no venue-to-city or venue-to-country conflicts.

In `wc_matches`, however, several venue names are associated with more than one city or country:

- `Estadio Nacional`
- `Stanford Stadium`
- `Olympiastadion`
- `Volksparkstadion`

This does not necessarily indicate inconsistent data. Stadium names may be reused in different countries or may refer to the same venue using different locality conventions.

The individual records are therefore inspected.

```python
# Inspecting all records for the four venues
venues_to_investigate = [
    "Estadio Nacional",
    "Stanford Stadium",
    "Olympiastadion",
    "Volksparkstadion"
]

venue_conflict_records = matches_clean[
    matches_clean["venue"].isin(
        venues_to_investigate
    )
][
    [
        "year",
        "date",
        "stage",
        "team1",
        "team2",
        "venue",
        "city",
        "country"
    ]
].sort_values(
    ["venue", "year", "date"]
)

display(venue_conflict_records)
```

```python
# Summarising the venue/location combinations 
venue_location_summary = (
    matches_clean[
        matches_clean["venue"].isin(
            venues_to_investigate
        )
    ]
    .groupby(
        ["venue", "city", "country"],
        dropna=False
    )
    .size()
    .reset_index(name="match_count")
    .sort_values(
        ["venue", "country", "city"]
    )
)

display(venue_location_summary)
```

```python
# Adding year to the venue identity check 
venue_identity_check = (
    matches_clean[
        matches_clean["venue"].isin(
            venues_to_investigate
        )
    ]
    .groupby(
        ["year", "venue"]
    )
    .agg(
        cities=("city", lambda x: sorted(set(x.dropna()))),
        countries=("country", lambda x: sorted(set(x.dropna()))),
        matches=("venue", "size")
    )
    .reset_index()
)

display(venue_identity_check)
```

#### Venue Investigation Conclusion

The apparent venue-location conflicts were investigated individually and do not indicate venue-name errors requiring correction.

`Estadio Nacional` occurs in two different tournament locations: Santiago, Chile in 1962 and Brasília, Brazil in 2014. The shared venue name therefore refers to different physical stadiums.

`Olympiastadion` similarly occurs in Munich, West Germany in 1974 and Berlin, Germany in 2006. These are different stadium locations sharing the same venue name.

`Volksparkstadion` is consistently associated with Hamburg. Its country is recorded as `West Germany` in 1974 and `Germany` in 2006, reflecting the historically appropriate country labels for the respective tournaments.

`Stanford Stadium` is recorded as both `Stanford` and `Palo Alto` within the 1994 tournament. This represents the previously identified locality convention rather than a venue-name inconsistency. The source city values are retained rather than creating an artificial combined location or forcing one locality representation.

No venue-name transformations are required.

The investigation also demonstrates that venue name alone should not be treated as a globally unique stadium identifier. Where venue identity is required in later analysis, venue should be interpreted together with location and, where relevant, tournament year.

## Formal Cross-Dataset Validation

### Validation Strategy

The next stage validates relationships between datasets that describe the same matches, tournaments, teams or events.

Cross-dataset validation is performed only where a meaningful relationship exists. Datasets are not expected to contain identical records when they represent different levels of detail or different analytical grains.

The validation stage will examine relationships including:

- 2026 fixtures against 2026 team
- World Cup editions against historical World Cup matches
- World Cup editions against World Cup top-scorer records
- selected World Cup matches against the wider international results dataset
- goalscorer records against corresponding match results
- penalty-shootout records against corresponding match results
- historical naming relationships where the former-name reference is relevant

```python
# Indentifying named teams in fixtures 
official_2026_teams = set(
    teams_clean["team"]
    .dropna()
    .astype(str)
)

fixture_team_labels = set(
    fixtures_clean["team1"]
    .dropna()
    .astype(str)
).union(
    set(
        fixtures_clean["team2"]
        .dropna()
        .astype(str)
    )
)

named_fixture_teams = (
    fixture_team_labels
    &
    official_2026_teams
)

fixture_placeholders = sorted(
    fixture_team_labels
    -
    official_2026_teams
)

teams_missing_from_fixtures = sorted(
    official_2026_teams
    -
    fixture_team_labels
)

print(
    "Official teams:",
    len(official_2026_teams)
)

print(
    "Official teams found in fixtures:",
    len(named_fixture_teams)
)

print(
    "Official teams missing from fixtures:",
    len(teams_missing_from_fixtures)
)

print(
    "Non-team / progression fixture labels:",
    len(fixture_placeholders)
)
```

```python
# Displaying anything missing 
print("Official teams missing from fixtures:")
print("-" * 40)

if len(teams_missing_from_fixtures) == 0:
    print("None")
else:
    for team in teams_missing_from_fixtures:
        print(team)
```

```python
# Confirming remaining labels are structural placeholders 
print("Fixture labels not present in the official team list:")
print("-" * 60)

for label in fixture_placeholders:
    print(label)
```

#### 2026 Fixture Team-Coverage Conclusion

Formal cross-dataset validation between `2026_fixtures` and `2026_teams` confirmed that all 48 official teams are represented in the fixture dataset.

No official teams are missing from the fixtures.

The remaining 63 fixture labels that do not appear in the official team list are structural progression placeholders used for knockout-stage scheduling. These include group-position labels, best third-place qualifiers, Round of 32 winners, quarter-final and semi-final placeholders, and final/third-place placeholders.

These values are intentional and should be retained because they describe the tournament structure before the relevant teams are known.

Team coverage between the two 2026 datasets is therefore complete, with no unexplained mismatches.

### Validating 2026 Team data

After confirming complete team coverage, the team metadata recorded in `2026_fixtures` is validated against `2026_teams`.

The shared attributes are:

- FIFA rank;
- confederation;
- coach.

Only fixtures containing named official teams are included in this comparison. Knockout-stage progression placeholders are excluded because their team identities, and therefore their associated metadata, are not yet known.

The validation is performed separately for `team1` and `team2` before the results are combined.

```python
# Building lookup 
team_metadata_lookup = (
    teams_clean[
        [
            "team",
            "fifa_rank",
            "confederation",
            "coach"
        ]
    ]
    .copy()
    .rename(
        columns={
            "fifa_rank": "expected_fifa_rank",
            "confederation": "expected_confederation",
            "coach": "expected_coach"
        }
    )
)

display(team_metadata_lookup.head())
```

```python
# Validating team1
team1_validation = (
    fixtures_clean[
        fixtures_clean["team1"].isin(
            official_2026_teams
        )
    ][
        [
            "date",
            "stage",
            "team1",
            "team1_fifa_rank",
            "team1_confederation",
            "team1_coach"
        ]
    ]
    .merge(
        team_metadata_lookup,
        left_on="team1",
        right_on="team",
        how="left",
        validate="many_to_one"
    )
)

team1_validation["rank_match"] = (
    team1_validation["team1_fifa_rank"]
    ==
    team1_validation["expected_fifa_rank"]
)

team1_validation["confederation_match"] = (
    team1_validation["team1_confederation"]
    ==
    team1_validation["expected_confederation"]
)

team1_validation["coach_match"] = (
    team1_validation["team1_coach"]
    ==
    team1_validation["expected_coach"]
)

print("Named team1 fixture records:", len(team1_validation))
print(
    "FIFA rank mismatches:",
    (~team1_validation["rank_match"]).sum()
)
print(
    "Confederation mismatches:",
    (~team1_validation["confederation_match"]).sum()
)
print(
    "Coach mismatches:",
    (~team1_validation["coach_match"]).sum()
)
```

```python
# Validating team2
team2_validation = (
    fixtures_clean[
        fixtures_clean["team2"].isin(
            official_2026_teams
        )
    ][
        [
            "date",
            "stage",
            "team2",
            "team2_fifa_rank",
            "team2_confederation",
            "team2_coach"
        ]
    ]
    .merge(
        team_metadata_lookup,
        left_on="team2",
        right_on="team",
        how="left",
        validate="many_to_one"
    )
)

team2_validation["rank_match"] = (
    team2_validation["team2_fifa_rank"]
    ==
    team2_validation["expected_fifa_rank"]
)

team2_validation["confederation_match"] = (
    team2_validation["team2_confederation"]
    ==
    team2_validation["expected_confederation"]
)

team2_validation["coach_match"] = (
    team2_validation["team2_coach"]
    ==
    team2_validation["expected_coach"]
)

print("Named team2 fixture records:", len(team2_validation))
print(
    "FIFA rank mismatches:",
    (~team2_validation["rank_match"]).sum()
)
print(
    "Confederation mismatches:",
    (~team2_validation["confederation_match"]).sum()
)
print(
    "Coach mismatches:",
    (~team2_validation["coach_match"]).sum()
)
```

### Validating World Cup Editions Against Historical Matches

The cleaned World Cup edition data is now validated against the selected historical World Cup match data.

`wc_editions` contains one row per tournament edition, while `wc_matches` contains a selected subset of historical World Cup matches rather than the complete match history.

The number of records in `wc_matches` is therefore not expected to equal the official match totals recorded in `wc_editions`.

Validation instead focuses on relationships that can meaningfully be compared between the datasets, including:

- tournament-year coverage;
- whether selected match dates fall within the corresponding tournament date range;
- host/location relationships where appropriate;
- final and placement information where the selected match records provide sufficient evidence.

```python
# Comparing tournament-year coverage 
edition_years = set(
    editions_clean["year"]
    .dropna()
    .astype(int)
)

match_years = set(
    matches_clean["year"]
    .dropna()
    .astype(int)
)

print(
    "World Cup edition years:",
    len(edition_years)
)

print(
    "World Cup match years:",
    len(match_years)
)

print(
    "\nEdition years with no selected match records:"
)

missing_match_years = sorted(
    edition_years - match_years
)

if len(missing_match_years) == 0:
    print("None")
else:
    for year in missing_match_years:
        print(year)

print(
    "\nMatch years without a corresponding edition:"
)

unexpected_match_years = sorted(
    match_years - edition_years
)

if len(unexpected_match_years) == 0:
    print("None")
else:
    for year in unexpected_match_years:
        print(year)
```

```python
# Merging each selected math with its tournamnets official start/end dates
match_date_validation = (
    matches_clean[
        [
            "year",
            "date",
            "stage",
            "team1",
            "team2"
        ]
    ]
    .merge(
        editions_clean[
            [
                "year",
                "start_date",
                "end_date"
            ]
        ],
        on="year",
        how="left",
        validate="many_to_one"
    )
)

match_date_validation["within_tournament_dates"] = (
    (match_date_validation["date"]
     >= match_date_validation["start_date"])
    &
    (match_date_validation["date"]
     <= match_date_validation["end_date"])
)

print(
    "Selected match records:",
    len(match_date_validation)
)

print(
    "Matches within tournament date range:",
    match_date_validation[
        "within_tournament_dates"
    ].sum()
)

print(
    "Matches outside tournament date range:",
    (
        ~match_date_validation[
            "within_tournament_dates"
        ]
    ).sum()
)
```

```python
# Displaying any date discrepancies 
date_discrepancies = (
    match_date_validation[
        ~match_date_validation[
            "within_tournament_dates"
        ]
    ]
)

display(date_discrepancies)
```

```python
# For each tournmanet, checking to see the earliest and latest selected match in matches_clean dataset alongside the official tournament start and end dates. 
selected_match_ranges = (
    matches_clean
    .groupby("year")
    .agg(
        first_selected_match=("date", "min"),
        last_selected_match=("date", "max"),
        selected_matches=("date", "size")
    )
    .reset_index()
    .merge(
        editions_clean[
            [
                "year",
                "start_date",
                "end_date"
            ]
        ],
        on="year",
        how="left",
        validate="one_to_one"
    )
)

display(selected_match_ranges)
```

#### World Cup Edition and Match-Date Validation Conclusion

Formal validation between `wc_editions` and `wc_matches` confirmed complete tournament-year coverage.

All 22 World Cup edition years represented in `wc_editions` are present in `wc_matches`, and no match records contain a year without a corresponding tournament edition.

All 184 selected historical match records fall within the official start and end dates recorded for their corresponding World Cup edition.

The earliest and latest selected matches do not always equal the official tournament boundaries. This is expected because `wc_matches` contains a selected subset of historical matches rather than the complete match history.

No date or edition-year discrepancies were identified, and no transformations are required.

### Validating Tournament Outcomes Against Selected Matches

Tournament finishing positions recorded in `wc_editions` are now compared with the corresponding placement matches available in `wc_matches`.

This validation focuses on:

- champion and runner-up information against final-stage records;
- third- and fourth-place information where a corresponding placement match is represented.

Historical World Cup formats are taken into account. Not every edition used the same knockout structure, so the existence of a conventional `Final` or `3rd Place` match is not assumed for every tournament.

The available stage records are inspected before outcome comparisons are made.

```python
# Inspecting final-related stage labels by year 
final_related_stages = (
    matches_clean[
        matches_clean["stage"]
        .astype(str)
        .str.contains(
            "Final|3rd Place|Third",
            case=False,
            na=False
        )
    ][
        [
            "year",
            "date",
            "stage",
            "team1",
            "score1",
            "score2",
            "team2"
        ]
    ]
    .sort_values(["year", "date"])
)

display(final_related_stages)
```

```python
final_counts = (
    matches_clean[
        matches_clean["stage"] == "Final"
    ]
    .groupby("year")
    .size()
    .reset_index(name="final_records")
)

final_coverage = (
    editions_clean[
        [
            "year",
            "champion",
            "runner_up"
        ]
    ]
    .merge(
        final_counts,
        on="year",
        how="left"
    )
)

final_coverage["final_records"] = (
    final_coverage["final_records"]
    .fillna(0)
    .astype(int)
)

display(final_coverage)
```

```python
# Comparing final participants with champion and runner-up 
final_matches = (
    matches_clean[
        matches_clean["stage"] == "Final"
    ][
        [
            "year",
            "date",
            "team1",
            "score1",
            "score2",
            "team2"
        ]
    ]
    .merge(
        editions_clean[
            [
                "year",
                "champion",
                "runner_up"
            ]
        ],
        on="year",
        how="left",
        validate="one_to_one"
    )
)

final_matches["participants_match"] = (
    final_matches.apply(
        lambda row:
        {
            row["team1"],
            row["team2"]
        }
        ==
        {
            row["champion"],
            row["runner_up"]
        },
        axis=1
    )
)

print(
    "Conventional final records:",
    len(final_matches)
)

print(
    "Finals where participants match champion/runner-up:",
    final_matches["participants_match"].sum()
)

print(
    "Finals with participant discrepancy:",
    (~final_matches["participants_match"]).sum()
)
```

```python
# Displaying any participant discrepancies 
final_participant_discrepancies = (
    final_matches[
        ~final_matches["participants_match"]
    ]
)

display(final_participant_discrepancies)
```

#### Final-Participant Validation

Twenty-one of the 22 World Cup editions contain a conventional `Final` record in `wc_matches`.

The 1950 edition does not contain a match labelled `Final`, reflecting its different tournament structure rather than an unexplained missing record.

For all 21 editions containing a conventional final, the two final participants exactly match the champion and runner-up recorded in `wc_editions`.

No final-participant discrepancies were identified.

```python
# Validating champions and runner-ups - inscpecting final scores and notes first as this contained extra information that may be useful. 
final_outcome_inspection = (
    matches_clean[
        matches_clean["stage"] == "Final"
    ][
        [
            "year",
            "date",
            "team1",
            "score1",
            "score2",
            "team2",
            "notes"
        ]
    ]
    .sort_values("year")
)

display(final_outcome_inspection)

```

```python
# Identifying finls levels on the recorded score 
level_finals = final_outcome_inspection[
    final_outcome_inspection["score1"]
    ==
    final_outcome_inspection["score2"]
]

print(
    "Finals level on the recorded match score:",
    len(level_finals)
)

display(level_finals)
```

```python
# Checking these finals against shootouts_clean
level_final_shootouts = (
    level_finals[
        [
            "year",
            "date",
            "team1",
            "team2"
        ]
    ]
    .merge(
        shootouts_clean[
            [
                "date",
                "home_team",
                "away_team",
                "winner"
            ]
        ],
        left_on=[
            "date",
            "team1",
            "team2"
        ],
        right_on=[
            "date",
            "home_team",
            "away_team"
        ],
        how="left"
    )
)

display(level_final_shootouts)
```

```python
# Validating non-level final winners - where the recorded score itself determines the winner 
non_level_finals = final_matches[
    final_matches["score1"]
    != final_matches["score2"]
].copy()

non_level_finals["match_winner"] = np.where(
    non_level_finals["score1"]
    >
    non_level_finals["score2"],
    non_level_finals["team1"],
    non_level_finals["team2"]
)

non_level_finals["winner_matches_champion"] = (
    non_level_finals["match_winner"]
    ==
    non_level_finals["champion"]
)

print(
    "Non-level conventional finals:",
    len(non_level_finals)
)

print(
    "Winners matching edition champion:",
    non_level_finals[
        "winner_matches_champion"
    ].sum()
)

print(
    "Winner discrepancies:",
    (
        ~non_level_finals[
            "winner_matches_champion"
        ]
    ).sum()
)

display(
    non_level_finals[
        ~non_level_finals[
            "winner_matches_champion"
        ]
    ]
)
```

The 21 conventional finals split neatly into:

- 18 finals decided by the recorded match score → all 18 winners match editions_clean[champion].
- 3 finals level after the recorded match score → 1994, 2006 and 2022. All three have corresponding records in shootouts_clean, and the shoot-out winner is Brazil, Italy and Argentina respectively, exactly the champions recorded in editions_clean.
- 1950 remains the one edition without a conventional "Final" record and needs to be treated according to its different tournament structure rather than forced into the normal final logic.

So there are 0 unexplained champion/runner-up discrepancies.

However we will validate the shoot out final winners against champions as a extra measure instead of relying on visual inspection. 


```python
level_final_validation = (
    level_final_shootouts
    .merge(
        editions_clean[
            ["year", "champion"]
        ],
        on="year",
        how="left",
        validate="one_to_one"
    )
)

level_final_validation["winner_matches_champion"] = (
    level_final_validation["winner"]
    ==
    level_final_validation["champion"]
)

print(
    "Finals requiring shoot-out validation:",
    len(level_final_validation)
)

print(
    "Shoot-out winners matching edition champion:",
    level_final_validation[
        "winner_matches_champion"
    ].sum()
)

print(
    "Shoot-out winner discrepancies:",
    (
        ~level_final_validation[
            "winner_matches_champion"
        ]
    ).sum()
)

display(
    level_final_validation[
        ~level_final_validation[
            "winner_matches_champion"
        ]
    ]
)
```

```python
# Inspecting the 1950 exception by first inspecting the selected 1950 match records
matches_1950 = (
    matches_clean[
        matches_clean["year"] == 1950
    ][
        [
            "date",
            "stage",
            "team1",
            "score1",
            "score2",
            "team2",
            "notes"
        ]
    ]
    .sort_values("date")
)

display(matches_1950)
```

```python
# with Edition record: 

edition_1950 = editions_clean[
    editions_clean["year"] == 1950
][
    [
        "year",
        "host",
        "champion",
        "runner_up",
        "third_place",
        "fourth_place",
        "format"
    ]
]

display(edition_1950)
```

#### Champion and Runner-Up Validation Conclusion

Champion and runner-up information was successfully validated across all 22 World Cup editions while accounting for differences in historical tournament format.

Of the 22 editions, 21 contain a conventional `Final` record in `wc_matches`.

For these 21 editions:

- 18 finals were decided by the recorded match score, and all 18 match winners correspond to the champion recorded in `wc_editions`;
- 3 finals (1994, 2006 and 2022) were level on the recorded match score and required shoot-out validation using `shootouts_clean`;
- all 3 shoot-out winners correspond to the champion recorded in `wc_editions`.

The 1950 World Cup is represented differently. `wc_editions` records its format as `Round Robin Final`, while the selected historical match data uses the stage label `Final Round` rather than `Final`. The selected final-round records include Uruguay's 2–1 victory over Brazil on 16 July 1950, and the edition data records Uruguay as champion and Brazil as runner-up.

The absence of a conventional `Final` record for 1950 is therefore explained by the tournament structure and should not be treated as missing or erroneous data.

No champion or runner-up discrepancies were identified, and no transformations are required.

### Validating Third- and Fourth-Place Outcomes

Third- and fourth-place information in `wc_editions` is now compared with the corresponding placement matches available in `wc_matches`.

As with the final validation, historical tournament formats are taken into account. A conventional third-place match is not assumed to exist for every edition.

For editions containing a `3rd Place` match, validation will examine:

- whether the recorded participants correspond to the edition's third- and fourth-place teams;
- whether the match result identifies the recorded third-place team as the winner;
- any matches requiring additional interpretation because of historical format or score representation.

```python
# Checking third-place match coverage 
third_place_counts = (
    matches_clean[
        matches_clean["stage"] == "3rd Place"
    ]
    .groupby("year")
    .size()
    .reset_index(name="third_place_records")
)

third_place_coverage = (
    editions_clean[
        [
            "year",
            "third_place",
            "fourth_place"
        ]
    ]
    .merge(
        third_place_counts,
        on="year",
        how="left"
    )
)

third_place_coverage["third_place_records"] = (
    third_place_coverage["third_place_records"]
    .fillna(0)
    .astype(int)
)

display(third_place_coverage)
```

```python
# Comparing placement-match participants 
third_place_matches = (
    matches_clean[
        matches_clean["stage"] == "3rd Place"
    ][
        [
            "year",
            "date",
            "team1",
            "score1",
            "score2",
            "team2",
            "notes"
        ]
    ]
    .merge(
        editions_clean[
            [
                "year",
                "third_place",
                "fourth_place"
            ]
        ],
        on="year",
        how="left",
        validate="one_to_one"
    )
)

third_place_matches["participants_match"] = (
    third_place_matches.apply(
        lambda row:
        {
            row["team1"],
            row["team2"]
        }
        ==
        {
            row["third_place"],
            row["fourth_place"]
        },
        axis=1
    )
)

print(
    "Third-place match records:",
    len(third_place_matches)
)

print(
    "Matches where participants agree with "
    "third/fourth place:",
    third_place_matches[
        "participants_match"
    ].sum()
)

print(
    "Participant discrepancies:",
    (
        ~third_place_matches[
            "participants_match"
        ]
    ).sum()
)
```

```python
display(
    third_place_matches[
        ~third_place_matches[
            "participants_match"
        ]
    ]
)
```

```python
# Validate third place match winners 
level_third_place_matches = (
    third_place_matches[
        third_place_matches["score1"]
        ==
        third_place_matches["score2"]
    ]
)

print(
    "Third-place matches level on recorded score:",
    len(level_third_place_matches)
)

display(level_third_place_matches)
```

```python
# Comparing match winner with recorded third place 
third_place_outcome_validation = (
    third_place_matches[
        third_place_matches["score1"]
        != third_place_matches["score2"]
    ]
    .copy()
)

third_place_outcome_validation["match_winner"] = np.where(
    third_place_outcome_validation["score1"]
    >
    third_place_outcome_validation["score2"],
    third_place_outcome_validation["team1"],
    third_place_outcome_validation["team2"]
)

third_place_outcome_validation[
    "winner_matches_third_place"
] = (
    third_place_outcome_validation["match_winner"]
    ==
    third_place_outcome_validation["third_place"]
)

print(
    "Third-place matches decided by recorded score:",
    len(third_place_outcome_validation)
)

print(
    "Winners matching recorded third place:",
    third_place_outcome_validation[
        "winner_matches_third_place"
    ].sum()
)

print(
    "Winner discrepancies:",
    (
        ~third_place_outcome_validation[
            "winner_matches_third_place"
        ]
    ).sum()
)

display(
    third_place_outcome_validation[
        ~third_place_outcome_validation[
            "winner_matches_third_place"
        ]
    ]
)
```

### Validating World Cup Editions Against Top Scorers

The cleaned World Cup top-scorer data is now formally validated against the corresponding edition-level information in `wc_editions`.

Both datasets contain one record per World Cup edition, but their representation of joint top scorers is not always identical. In particular, some editions with tied leading scorers are represented by a single selected player or an aggregated label in `wc_top_scorers`.

Validation therefore distinguishes between genuine value discrepancies and known differences in representation.

The comparison will examine:

- edition and year coverage;
- host;
- top-scorer goals;
- scorer/player representation;
- scorer country representation;
- team-result information where it can be compared with edition finishing positions.

```python
# Validating edition/year coverage 
edition_years = set(
    editions_clean["year"]
    .dropna()
    .astype(int)
)

top_scorer_years = set(
    top_scorers_clean["year"]
    .dropna()
    .astype(int)
)

print(
    "World Cup edition years:",
    len(edition_years)
)

print(
    "Top-scorer years:",
    len(top_scorer_years)
)

print(
    "\nEdition years missing from top scorers:"
)

missing_top_scorer_years = sorted(
    edition_years - top_scorer_years
)

if len(missing_top_scorer_years) == 0:
    print("None")
else:
    for year in missing_top_scorer_years:
        print(year)

print(
    "\nTop-scorer years without a corresponding edition:"
)

unexpected_top_scorer_years = sorted(
    top_scorer_years - edition_years
)

if len(unexpected_top_scorer_years) == 0:
    print("None")
else:
    for year in unexpected_top_scorer_years:
        print(year)
```

```python
# Creating the comparison table 
edition_scorer_validation = (
    editions_clean[
        [
            "edition",
            "year",
            "host",
            "top_scorer",
            "top_scorer_country",
            "top_scorer_goals"
        ]
    ]
    .merge(
        top_scorers_clean[
            [
                "edition",
                "year",
                "player",
                "country",
                "goals",
                "host",
                "team_result"
            ]
        ],
        on=["edition", "year"],
        how="outer",
        suffixes=(
            "_editions",
            "_top_scorers"
        ),
        validate="one_to_one",
        indicator=True
    )
)

print(
    "Comparison records:",
    len(edition_scorer_validation)
)

print("\nMerge status:")
print(
    edition_scorer_validation["_merge"]
    .value_counts()
)

display(edition_scorer_validation)
```

```python
# Validating team_Result against Edition finishing positions as these contain different categories with only 4 categories (champion, runner_up, third_place, fourth_place) being in editions_clean. 
def expected_top_four_result(row):
    if row["country"] == row["champion"]:
        return "Champions"
    elif row["country"] == row["runner_up"]:
        return "Runner-up"
    elif row["country"] == row["third_place"]:
        return "Third place"
    elif row["country"] == row["fourth_place"]:
        return "Fourth place"
    else:
        return np.nan


team_result_validation = (
    top_scorers_clean
    .merge(
        editions_clean[
            [
                "year",
                "champion",
                "runner_up",
                "third_place",
                "fourth_place"
            ]
        ],
        on="year",
        how="left",
        validate="one_to_one"
    )
)

team_result_validation["expected_top_four_result"] = (
    team_result_validation.apply(
        expected_top_four_result,
        axis=1
    )
)

directly_validatable_results = (
    team_result_validation[
        team_result_validation[
            "expected_top_four_result"
        ].notna()
    ]
    .copy()
)

directly_validatable_results["team_result_match"] = (
    directly_validatable_results["team_result"]
    ==
    directly_validatable_results[
        "expected_top_four_result"
    ]
)

print(
    "Top-scorer team results directly "
    "validatable from editions:",
    len(directly_validatable_results)
)

print(
    "Matching team results:",
    directly_validatable_results[
        "team_result_match"
    ].sum()
)

print(
    "Team-result discrepancies:",
    (
        ~directly_validatable_results[
            "team_result_match"
        ]
    ).sum()
)
```

# Correcting the 1994 Team Result 

#### Correcting the 1994 Top-Scorer Team Result

Formal cross-dataset validation identified one inconsistency in the `team_result` field of `wc_top_scorers`.

The 1994 record for Hristo Stoichkov currently classifies Bulgaria's result as `Semi-final`.

Cross-dataset evidence shows that:

- `wc_editions` records Bulgaria as fourth place in 1994;
- `wc_matches` records Sweden defeating Bulgaria 4–0 in the third-place match;
- `team_result` elsewhere distinguishes exact finishing positions such as `Third place` and `Fourth place`.

The value `Semi-final` is therefore less precise and inconsistent with the final tournament outcome represented elsewhere in the project data.

The 1994 value is corrected from `Semi-final` to `Fourth place`.

```python
top_scorers_clean.loc[
    (top_scorers_clean["year"] == 1994)
    &
    (top_scorers_clean["player"] == "Hristo Stoichkov"),
    "team_result"
] = "Fourth place"

display(
    top_scorers_clean[
        top_scorers_clean["year"] == 1994
    ][
        [
            "year",
            "player",
            "country",
            "team_result"
        ]
    ]
)
```

#### Investigating the Unmatched 1974 Germany Record

The initial formal comparison matched 183 of the 184 selected World Cup records.

The only unmatched record is the 1974 group-stage match between East Germany and West Germany.

The temporary historical matching convention used for the comparison mapped both `East Germany` and `West Germany` to `Germany`. For this particular fixture, that collapses two distinct historical teams into the same comparison label (`Germany` vs `Germany`) and prevents a valid match.

This does not demonstrate an error in the cleaned match data. Instead, it identifies a limitation in the temporary matching method.

The corresponding record in `results_clean` was inspected before refining the validation logic. No dataset values are changed.

#### Refining Historical Team-Name Matching

Inspection of the unmatched 1974 record confirmed that the discrepancy is caused by different historical naming conventions between the datasets.

The selected World Cup match dataset represents the teams as:

- `East Germany`
- `West Germany`

The wider international results dataset represents the same teams in this match as:

- `German DR`
- `Germany`

The initial temporary validation mapping incorrectly mapped both East Germany and West Germany to `Germany`, which collapsed two historically distinct teams into the same comparison label.

The temporary matching convention is therefore refined so that:

- `East Germany` maps to `German DR`;
- `West Germany` maps to `Germany`;
- `Soviet Union` continues to map to `Russia`.

These mappings are used only to support cross-dataset validation. The historical team names stored in the cleaned datasets are not changed.

```python
# Recreating the temp validation data
historical_match_name_map = {
    "West Germany": "Germany",
    "East Germany": "German DR",
    "Soviet Union": "Russia"
}

matches_validation = matches_clean.copy()
results_validation = world_cup_results.copy()

matches_validation["team1_match_name"] = (
    matches_validation["team1"]
    .replace(historical_match_name_map)
)

matches_validation["team2_match_name"] = (
    matches_validation["team2"]
    .replace(historical_match_name_map)
)

results_validation["home_match_name"] = (
    results_validation["home_team"]
    .replace(historical_match_name_map)
)

results_validation["away_match_name"] = (
    results_validation["away_team"]
    .replace(historical_match_name_map)
)
```

```python
# recreating the pair columns as we recreated the temporary dataframes 
matches_validation["team_pair"] = (
    matches_validation.apply(
        lambda row: tuple(
            sorted(
                [
                    row["team1_match_name"],
                    row["team2_match_name"]
                ]
            )
        ),
        axis=1
    )
)

results_validation["team_pair"] = (
    results_validation.apply(
        lambda row: tuple(
            sorted(
                [
                    row["home_match_name"],
                    row["away_match_name"]
                ]
            )
        ),
        axis=1
    )
)
```

```python
# Re running the formal match 
formal_match_validation = (
    matches_validation
    .merge(
        results_validation[
            [
                "date",
                "year",
                "team_pair",
                "home_team",
                "away_team",
                "home_match_name",
                "away_match_name",
                "home_score",
                "away_score"
            ]
        ],
        on=[
            "date",
            "year",
            "team_pair"
        ],
        how="left",
        indicator=True
    )
)

print(
    "Selected World Cup match records:",
    len(matches_validation)
)

print(
    "Matched to results_clean:",
    (
        formal_match_validation["_merge"]
        == "both"
    ).sum()
)

print(
    "Unmatched:",
    (
        formal_match_validation["_merge"]
        != "both"
    ).sum()
)
```

### Validating World Cup Match Scores Against International Results

All 184 selected World Cup match records have been successfully matched to corresponding records in `results_clean`.

The scores are now formally validated.

Because the two datasets do not always use the same team orientation, `score1` cannot automatically be compared with `home_score`. A selected match may list the same two teams in the reverse order from the wider results dataset.

The validation therefore first determines whether each matched record uses the same or reversed team orientation. Scores are then compared against the corresponding team rather than solely by column position.

The historical name mappings used for record matching remain temporary validation conventions and do not modify the cleaned datasets.

```python
formal_match_validation["same_orientation"] = (
    (
        formal_match_validation["team1_match_name"]
        ==
        formal_match_validation["home_match_name"]
    )
    &
    (
        formal_match_validation["team2_match_name"]
        ==
        formal_match_validation["away_match_name"]
    )
)

formal_match_validation["reversed_orientation"] = (
    (
        formal_match_validation["team1_match_name"]
        ==
        formal_match_validation["away_match_name"]
    )
    &
    (
        formal_match_validation["team2_match_name"]
        ==
        formal_match_validation["home_match_name"]
    )
)

print(
    "Same orientation:",
    formal_match_validation["same_orientation"].sum()
)

print(
    "Reversed orientation:",
    formal_match_validation["reversed_orientation"].sum()
)

print(
    "Neither orientation:",
    (
        ~formal_match_validation["same_orientation"]
        &
        ~formal_match_validation["reversed_orientation"]
    ).sum()
)
```

```python
# Aligning the reference scores 
formal_match_validation["expected_score1"] = np.where(
    formal_match_validation["same_orientation"],
    formal_match_validation["home_score"],
    formal_match_validation["away_score"]
)

formal_match_validation["expected_score2"] = np.where(
    formal_match_validation["same_orientation"],
    formal_match_validation["away_score"],
    formal_match_validation["home_score"]
)

formal_match_validation["score_match"] = (
    (
        formal_match_validation["score1"]
        ==
        formal_match_validation["expected_score1"]
    )
    &
    (
        formal_match_validation["score2"]
        ==
        formal_match_validation["expected_score2"]
    )
)

print(
    "Score matches:",
    formal_match_validation["score_match"].sum(),
    "/",
    len(formal_match_validation)
)

print(
    "Score discrepancies:",
    (~formal_match_validation["score_match"]).sum()
)
```

```python
# Extra check for discrepancies 
score_discrepancies = (
    formal_match_validation[
        ~formal_match_validation["score_match"]
    ][
        [
            "year",
            "date",
            "stage",
            "team1",
            "score1",
            "score2",
            "team2",
            "home_team",
            "home_score",
            "away_score",
            "away_team",
            "same_orientation",
            "reversed_orientation"
        ]
    ]
)

display(score_discrepancies)
```

### Validating Goalscorer Records Against Match Results

The cleaned goalscorer dataset is now formally validated against the wider international match results dataset.

The two datasets have different grains:

- `goalscorers` contains one row per recorded goal;
- `results` contains one row per match.

The raw datasets should therefore not be compared using a one-to-one merge.

Validation is performed in stages. First, each goalscorer record is checked for a corresponding match using the match date and team pairing. Goal records will then be aggregated to match level before their relationship with recorded match scores is assessed.

Missing scorer names or goal minutes are retained because these fields can be unknown even when the goal itself is valid.

```python
# Checking columns 
print("Goalscorer columns:")
print(goalscorers_clean.columns.tolist())

print("\nResults columns:")
print(results_clean.columns.tolist())
```

```python
# Before checking date + home team + away team as the match key, I'll confirm the conbination uniquely identifies matches in results_clean dataset
results_match_key_counts = (
    results_clean
    .groupby(
        [
            "date",
            "home_team",
            "away_team"
        ],
        dropna=False
    )
    .size()
    .reset_index(name="record_count")
)

duplicate_results_match_keys = (
    results_match_key_counts[
        results_match_key_counts["record_count"] > 1
    ]
)

print(
    "Unique match keys in results:",
    len(results_match_key_counts)
)

print(
    "Match keys appearing more than once:",
    len(duplicate_results_match_keys)
)

display(duplicate_results_match_keys.head(20))
```

```python
# Investigating the non=unique match keys 
non_unique_key_records = (
    results_clean.merge(
        duplicate_results_match_keys[
            [
                "date",
                "home_team",
                "away_team"
            ]
        ],
        on=[
            "date",
            "home_team",
            "away_team"
        ],
        how="inner"
    )
    .sort_values(
        [
            "date",
            "home_team",
            "away_team"
        ]
    )
)

display(
    non_unique_key_records[
        [
            "date",
            "home_team",
            "away_team",
            "home_score",
            "away_score",
            "tournament",
            "city",
            "country",
            "neutral",
            "match_completed"
        ]
    ]
)
```

```python
# Checking to see if these appear in goalscorers_clean dataset 
goalscorer_non_unique_key_check = (
    goalscorers_clean.merge(
        duplicate_results_match_keys[
            [
                "date",
                "home_team",
                "away_team"
            ]
        ],
        on=[
            "date",
            "home_team",
            "away_team"
        ],
        how="inner"
    )
)

print(
    "Goalscorer records associated with "
    "non-unique results match keys:",
    len(goalscorer_non_unique_key_check)
)

display(
    goalscorer_non_unique_key_check[
        [
            "date",
            "home_team",
            "away_team",
            "team",
            "scorer",
            "minute",
            "own_goal",
            "penalty"
        ]
    ]
)
```

#### Preparing a Unique Match Reference for Goalscorer Validation

The proposed match key of `date + home_team + away_team` produces 49,503 unique match-key combinations across the 49,505 records in `results`, with two match keys occurring more than once.

Inspection confirmed that the four records associated with these two repeated keys are not exact duplicates.

For Tahiti versus New Caledonia, the two records dated 17 February 1974 contain different scores: 2–1 and 1–2. External historical records confirm that Tahiti and New Caledonia played multiple friendlies in February 1974, including matches ending 2–1 and 1–2. However, the external sources checked do not provide sufficiently precise dates to independently confirm that both matches occurred on 17 February. The dates are therefore retained as recorded in the source dataset rather than changed without sufficient evidence.

For Gibraltar versus Cayman Islands, the two records dated 6 June 2026 have the same 4–1 score but different city values (`Gibraltar` and `Europa Point`). This repeated key is retained at this stage because resolving it is not necessary for the goalscorer validation.

Importantly, neither of the two repeated match keys appears in `goalscorers`. They therefore do not affect the goalscorer-to-results relationship being validated here.

Rather than altering or removing records from `results`, all four records belonging to the two non-unique keys are excluded only from the temporary reference table used for this specific validation.

This produces a temporary reference in which `date + home_team + away_team` is unique for the goalscorer records being assessed.

No values or records in the cleaned datasets are changed.

```python
ambiguous_match_keys = (
    duplicate_results_match_keys[
        [
            "date",
            "home_team",
            "away_team"
        ]
    ]
    .copy()
)

results_for_goalscorer_validation = (
    results_clean
    .merge(
        ambiguous_match_keys.assign(
            ambiguous_key=True
        ),
        on=[
            "date",
            "home_team",
            "away_team"
        ],
        how="left"
    )
)

results_for_goalscorer_validation = (
    results_for_goalscorer_validation[
        results_for_goalscorer_validation[
            "ambiguous_key"
        ].isna()
    ]
    .drop(columns="ambiguous_key")
    .copy()
)

print(
    "Records in temporary results reference:",
    len(results_for_goalscorer_validation)
)

print(
    "Duplicate match keys remaining:",
    results_for_goalscorer_validation.duplicated(
        subset=[
            "date",
            "home_team",
            "away_team"
        ]
    ).sum()
)
```

```python
# Validating goalscorer match coverage. I will perform the many-to-one merge
goalscorer_match_validation = (
    goalscorers_clean
    .merge(
        results_for_goalscorer_validation[
            [
                "date",
                "home_team",
                "away_team",
                "home_score",
                "away_score",
                "tournament"
            ]
        ],
        on=[
            "date",
            "home_team",
            "away_team"
        ],
        how="left",
        validate="many_to_one",
        indicator=True
    )
)

print(
    "Goalscorer records:",
    len(goalscorers_clean)
)

print(
    "Goalscorer records matched to a result:",
    (
        goalscorer_match_validation["_merge"]
        == "both"
    ).sum()
)

print(
    "Unmatched goalscorer records:",
    (
        goalscorer_match_validation["_merge"]
        != "both"
    ).sum()
)
```

```python
# Double checking unmatched goalscorer records 
unmatched_goalscorer_records = (
    goalscorer_match_validation[
        goalscorer_match_validation["_merge"]
        != "both"
    ]
)

print(
    "Unmatched goalscorer records:",
    len(unmatched_goalscorer_records)
)

display(
    unmatched_goalscorer_records[
        [
            "date",
            "home_team",
            "away_team",
            "team",
            "scorer",
            "minute",
            "own_goal",
            "penalty"
        ]
    ].head(30)
)
```

### Validating Goal-Event Counts Against Match Scores

All 47,886 individual goalscorer records have been successfully linked to corresponding records in `results`.

The next stage assesses whether the recorded goal events are consistent with the final match scores.

This comparison requires additional care because `goalscorers` contains individual goal events while `results` contains final match scores. Own goals must also be handled correctly before goal events can be attributed to the scoring side.

The representation of own goals is therefore inspected before aggregating goal events to match level. No transformations are applied during this initial inspection.

```python
print(
    "Total goal records:",
    len(goalscorers_clean)
)

print(
    "Own-goal records:",
    goalscorers_clean["own_goal"].sum()
)

print(
    "Penalty records:",
    goalscorers_clean["penalty"].sum()
)

print("\nOwn-goal values:")
print(
    goalscorers_clean["own_goal"]
    .value_counts(dropna=False)
)
```

```python
# Inspecting own goals 
own_goal_records = (
    goalscorers_clean[
        goalscorers_clean["own_goal"] == True
    ]
)

display(
    own_goal_records[
        [
            "date",
            "home_team",
            "away_team",
            "team",
            "scorer",
            "minute",
            "own_goal"
        ]
    ].head(30)
)
```

```python
# Checking whether 'team' is one of the match participants 
goalscorer_team_validation = goalscorers_clean.copy()

goalscorer_team_validation["team_is_home"] = (
    goalscorer_team_validation["team"]
    ==
    goalscorer_team_validation["home_team"]
)

goalscorer_team_validation["team_is_away"] = (
    goalscorer_team_validation["team"]
    ==
    goalscorer_team_validation["away_team"]
)

goalscorer_team_validation["team_is_match_participant"] = (
    goalscorer_team_validation["team_is_home"]
    |
    goalscorer_team_validation["team_is_away"]
)

print(
    "Goal records where team is a match participant:",
    goalscorer_team_validation[
        "team_is_match_participant"
    ].sum(),
    "/",
    len(goalscorer_team_validation)
)

print(
    "Goal records where team is not a match participant:",
    (
        ~goalscorer_team_validation[
            "team_is_match_participant"
        ]
    ).sum()
)
```

```python
# Double checking discrepancies 
goalscorer_team_discrepancies = (
    goalscorer_team_validation[
        ~goalscorer_team_validation[
            "team_is_match_participant"
        ]
    ]
)

display(
    goalscorer_team_discrepancies[
        [
            "date",
            "home_team",
            "away_team",
            "team",
            "scorer",
            "minute",
            "own_goal",
            "penalty"
        ]
    ].head(30)
)
```

#### Aggregating Goal Events to Match Level

Inspection of the own-goal records indicates that the `team` field represents the team credited with the goal rather than necessarily the scorer's own team.

For example, own-goal records can contain a scorer associated with the opposing side while the `team` value identifies the side receiving the goal.

The existing `team` field is therefore used to attribute goal events. Own-goal records are not manually reversed.

All goal events are now aggregated by match and credited team so that the resulting totals can be compared with the final scores recorded in `results`.

```python
goal_counts_by_team = (
    goalscorers_clean
    .groupby(
        [
            "date",
            "home_team",
            "away_team",
            "team"
        ],
        dropna=False
    )
    .size()
    .reset_index(name="recorded_goals")
)

print(
    "Match-team goal combinations:",
    len(goal_counts_by_team)
)

display(goal_counts_by_team.head(20))
```

```python
match_goal_counts = (
    goalscorers_clean[
        [
            "date",
            "home_team",
            "away_team"
        ]
    ]
    .drop_duplicates()
    .copy()
)

home_goal_counts = (
    goal_counts_by_team[
        goal_counts_by_team["team"]
        ==
        goal_counts_by_team["home_team"]
    ][
        [
            "date",
            "home_team",
            "away_team",
            "recorded_goals"
        ]
    ]
    .rename(
        columns={
            "recorded_goals": "recorded_home_goals"
        }
    )
)

away_goal_counts = (
    goal_counts_by_team[
        goal_counts_by_team["team"]
        ==
        goal_counts_by_team["away_team"]
    ][
        [
            "date",
            "home_team",
            "away_team",
            "recorded_goals"
        ]
    ]
    .rename(
        columns={
            "recorded_goals": "recorded_away_goals"
        }
    )
)

match_goal_counts = (
    match_goal_counts
    .merge(
        home_goal_counts,
        on=[
            "date",
            "home_team",
            "away_team"
        ],
        how="left",
        validate="one_to_one"
    )
    .merge(
        away_goal_counts,
        on=[
            "date",
            "home_team",
            "away_team"
        ],
        how="left",
        validate="one_to_one"
    )
)

match_goal_counts[
    [
        "recorded_home_goals",
        "recorded_away_goals"
    ]
] = (
    match_goal_counts[
        [
            "recorded_home_goals",
            "recorded_away_goals"
        ]
    ]
    .fillna(0)
    .astype(int)
)

print(
    "Matches represented in goalscorers:",
    len(match_goal_counts)
)

display(match_goal_counts.head(20))
```

```python
goal_score_validation = (
    match_goal_counts
    .merge(
        results_for_goalscorer_validation[
            [
                "date",
                "home_team",
                "away_team",
                "home_score",
                "away_score",
                "match_completed"
            ]
        ],
        on=[
            "date",
            "home_team",
            "away_team"
        ],
        how="left",
        validate="one_to_one"
    )
)

print(
    "Matches available for goal-count validation:",
    len(goal_score_validation)
)

print(
    "Matches missing a corresponding result:",
    goal_score_validation[
        "home_score"
    ].isna().sum()
)
```

#### Comparing Recorded Goal Events With Final Match Scores

The 47,886 individual goal records represent 15,532 distinct matches, and all 15,532 matches have been successfully linked to corresponding records in `results`.

The aggregated number of goal events credited to each side can therefore be compared with the final home and away scores.

A match is considered an exact goal-count match when both the recorded home-team goal-event count and recorded away-team goal-event count equal the corresponding final scores.

Any discrepancies are investigated before determining whether they represent data errors, incomplete goal-event coverage, or another structural difference between the datasets.

```python
goal_score_validation["home_goals_match"] = (
    goal_score_validation["recorded_home_goals"]
    ==
    goal_score_validation["home_score"]
)

goal_score_validation["away_goals_match"] = (
    goal_score_validation["recorded_away_goals"]
    ==
    goal_score_validation["away_score"]
)

goal_score_validation["both_scores_match"] = (
    goal_score_validation["home_goals_match"]
    &
    goal_score_validation["away_goals_match"]
)

print(
    "Matches with exact home and away goal counts:",
    goal_score_validation["both_scores_match"].sum(),
    "/",
    len(goal_score_validation)
)

print(
    "Matches with a goal-count discrepancy:",
    (
        ~goal_score_validation["both_scores_match"]
    ).sum()
)

print(
    "\nHome-side discrepancies:",
    (
        ~goal_score_validation["home_goals_match"]
    ).sum()
)

print(
    "Away-side discrepancies:",
    (
        ~goal_score_validation["away_goals_match"]
    ).sum()
)
```

```python
goal_score_validation["home_goal_difference"] = (
    goal_score_validation["recorded_home_goals"]
    -
    goal_score_validation["home_score"]
)

goal_score_validation["away_goal_difference"] = (
    goal_score_validation["recorded_away_goals"]
    -
    goal_score_validation["away_score"]
)

goal_count_discrepancies = (
    goal_score_validation[
        ~goal_score_validation["both_scores_match"]
    ]
    .copy()
)

print(
    "Goal-count discrepancy records:",
    len(goal_count_discrepancies)
)

print("\nHome-side difference distribution:")
print(
    goal_count_discrepancies[
        "home_goal_difference"
    ].value_counts().sort_index()
)

print("\nAway-side difference distribution:")
print(
    goal_count_discrepancies[
        "away_goal_difference"
    ].value_counts().sort_index()
)
```

```python
# Displaying actual matches 
display(
    goal_count_discrepancies[
        [
            "date",
            "home_team",
            "away_team",
            "recorded_home_goals",
            "home_score",
            "home_goal_difference",
            "recorded_away_goals",
            "away_score",
            "away_goal_difference"
        ]
    ].head(50)
)
```

#### Goalscorer and Match-Result Validation Conclusion

Formal cross-dataset validation between `goalscorers` and `results` is complete.

The initial match-key assessment identified two `date + home_team + away_team` combinations that were not unique within `results`. Investigation showed that neither of these match keys appears in `goalscorers`, so all four records associated with those two non-unique keys were excluded only from the temporary validation reference. No records were removed from the cleaned datasets.

The 47,886 goal-event records represent:

This interpretation was subsequently confirmed by the match-level validation: using the existing `team` field without manually reversing own goals reproduced the final match scores exactly.

### Validating Penalty Shoot-Out Records Against Match Results

The cleaned penalty shoot-out dataset is now formally validated against the wider international match results dataset.

The datasets contain different information. `results` records the match score, while `shootouts` identifies matches decided by a penalty shoot-out and records the shoot-out winner.

Validation therefore focuses on:

- whether each shoot-out record can be linked to a corresponding match;
- whether the participating teams are consistent;
- whether the associated match score is level where expected;
- whether the recorded shoot-out winner is one of the participating teams.

The `first_shooter` field is not required for match-level validation because it is missing for many historical records and does not determine whether the shoot-out record itself is valid.

```python
# Inspecting the Shootout Structures 
print("Shootout columns:")
print(shootouts_clean.columns.tolist())

print(
    "\nShootout records:",
    len(shootouts_clean)
)

print(
    "Unique shootout match keys:",
    shootouts_clean[
        [
            "date",
            "home_team",
            "away_team"
        ]
    ].drop_duplicates().shape[0]
)

print(
    "Duplicate shootout match keys:",
    shootouts_clean.duplicated(
        subset=[
            "date",
            "home_team",
            "away_team"
        ],
        keep=False
    ).sum()
)
```

```python
shootout_ambiguous_key_check = (
    shootouts_clean
    .merge(
        ambiguous_match_keys,
        on=[
            "date",
            "home_team",
            "away_team"
        ],
        how="inner"
    )
)

print(
    "Shootout records associated with "
    "non-unique results match keys:",
    len(shootout_ambiguous_key_check)
)

display(shootout_ambiguous_key_check)
```

```python
shootout_winner_validation = shootouts_clean.copy()

shootout_winner_validation[
    "winner_is_participant"
] = (
    (
        shootout_winner_validation["winner"]
        ==
        shootout_winner_validation["home_team"]
    )
    |
    (
        shootout_winner_validation["winner"]
        ==
        shootout_winner_validation["away_team"]
    )
)

print(
    "Shootout winners that are match participants:",
    shootout_winner_validation[
        "winner_is_participant"
    ].sum(),
    "/",
    len(shootout_winner_validation)
)

print(
    "Invalid shootout winner records:",
    (
        ~shootout_winner_validation[
            "winner_is_participant"
        ]
    ).sum()
)
```

```python
# Double checking invalid winner records 
invalid_shootout_winners = (
    shootout_winner_validation[
        ~shootout_winner_validation[
            "winner_is_participant"
        ]
    ]
)

display(
    invalid_shootout_winners[
        [
            "date",
            "home_team",
            "away_team",
            "winner",
            "first_shooter"
        ]
    ]
)
```

#### Link Shoot-Out Records to Match Results

The 682 shoot-out records contain 682 unique `date + home_team + away_team` combinations, with no duplicate shoot-out match keys.

None of the shoot-out records correspond to the two non-unique match keys previously identified in `results`. The unique temporary results reference can therefore be used safely for this validation.

All 682 recorded shoot-out winners are also one of the two teams participating in their respective matches.

The next step links each shoot-out record to its corresponding match in `results` to confirm match coverage before examining the associated scores.

```python
shootout_match_validation = (
    shootouts_clean
    .merge(
        results_for_goalscorer_validation[
            [
                "date",
                "home_team",
                "away_team",
                "home_score",
                "away_score",
                "tournament",
                "match_completed"
            ]
        ],
        on=[
            "date",
            "home_team",
            "away_team"
        ],
        how="left",
        validate="one_to_one",
        indicator=True
    )
)

print(
    "Shootout records:",
    len(shootouts_clean)
)

print(
    "Shootout records matched to a result:",
    (
        shootout_match_validation["_merge"] == "both"
    ).sum()
)

print(
    "Unmatched shootout records:",
    (
        shootout_match_validation["_merge"] != "both"
    ).sum()
)
```

```python
# Inspecting the unmatched shootouts 
unmatched_shootout_records = (
    shootout_match_validation[
        shootout_match_validation["_merge"] != "both"
    ]
)

print(
    "Unmatched shootout records:",
    len(unmatched_shootout_records)
)

display(
    unmatched_shootout_records[
        [
            "date",
            "home_team",
            "away_team",
            "winner",
            "first_shooter"
        ]
    ].head(30)
)
```

```python
shootout_match_validation["score_is_level"] = (
    shootout_match_validation["home_score"]
    ==
    shootout_match_validation["away_score"]
)

print(
    "Shootout matches with level recorded scores:",
    shootout_match_validation[
        "score_is_level"
    ].sum(),
    "/",
    len(shootout_match_validation)
)

print(
    "Shootout matches with non-level recorded scores:",
    (
        ~shootout_match_validation[
            "score_is_level"
        ]
    ).sum()
)
```

```python
non_level_shootout_matches = (
    shootout_match_validation[
        ~shootout_match_validation[
            "score_is_level"
        ]
    ]
)

display(
    non_level_shootout_matches[
        [
            "date",
            "home_team",
            "away_team",
            "home_score",
            "away_score",
            "winner",
            "tournament"
        ]
    ].head(30)
)
```

#### Refining the Shoot-Out Score Validation

The initial score-level comparison identified 38 records as having non-level scores. However, this calculation included the one shoot-out record that could not be matched to `results`.

Because its match scores are missing after the merge, it cannot meaningfully be classified as either level or non-level. Score-level validation is therefore restricted to the 681 successfully matched shoot-out records.

In addition, a non-level individual match score does not automatically indicate an error. Some competitions use penalty shoot-outs after two-legged ties when the aggregate score across both legs is level. In these cases, the second-leg score itself may legitimately be non-level.

The matched records are therefore separated and investigated before any data-quality conclusion is made.

```python
matched_shootout_records = (
    shootout_match_validation[
        shootout_match_validation["_merge"] == "both"
    ]
    .copy()
)

matched_shootout_records["score_is_level"] = (
    matched_shootout_records["home_score"]
    ==
    matched_shootout_records["away_score"]
)

print(
    "Successfully matched shootout records:",
    len(matched_shootout_records)
)

print(
    "Matched shootouts with level scores:",
    matched_shootout_records["score_is_level"].sum()
)

print(
    "Matched shootouts with non-level scores:",
    (~matched_shootout_records["score_is_level"]).sum()
)
```

```python
non_level_matched_shootouts = (
    matched_shootout_records[
        ~matched_shootout_records["score_is_level"]
    ]
    .copy()
)

print(
    "Non-level matched shootouts:",
    len(non_level_matched_shootouts)
)

print("\nBy tournament:")
print(
    non_level_matched_shootouts["tournament"]
    .value_counts()
)

print("\nBy decade:")
print(
    (
        non_level_matched_shootouts["date"].dt.year
        // 10 * 10
    )
    .value_counts()
    .sort_index()
)
```

```python
reverse_leg_check = (
    non_level_matched_shootouts[
        [
            "date",
            "home_team",
            "away_team",
            "home_score",
            "away_score",
            "winner",
            "tournament"
        ]
    ]
    .merge(
        results_clean[
            [
                "date",
                "home_team",
                "away_team",
                "home_score",
                "away_score",
                "tournament"
            ]
        ],
        left_on=[
            "away_team",
            "home_team"
        ],
        right_on=[
            "home_team",
            "away_team"
        ],
        how="left",
        suffixes=("_shootout_match", "_possible_first_leg")
    )
)

reverse_leg_check = (
    reverse_leg_check[
        reverse_leg_check["date_possible_first_leg"]
        <
        reverse_leg_check["date_shootout_match"]
    ]
    .copy()
)

reverse_leg_check["days_between"] = (
    reverse_leg_check["date_shootout_match"]
    -
    reverse_leg_check["date_possible_first_leg"]
).dt.days

reverse_leg_check = (
    reverse_leg_check[
        reverse_leg_check["days_between"].between(1, 365)
    ]
    .sort_values(
        [
            "date_shootout_match",
            "days_between"
        ]
    )
)

print(
    "Possible earlier reverse-leg records found:",
    len(reverse_leg_check)
)

display(
    reverse_leg_check[
        [
            "date_shootout_match",
            "home_team_shootout_match",
            "away_team_shootout_match",
            "home_score_shootout_match",
            "away_score_shootout_match",
            "date_possible_first_leg",
            "home_score_possible_first_leg",
            "away_score_possible_first_leg",
            "days_between",
            "winner",
            "tournament_shootout_match"
        ]
    ].head(50)
)
```

#### Testing Possible Two-Legged Ties Using Aggregate Scores

The search for earlier reverse fixtures returned 39 candidate records for the 37 non-level shoot-out matches.

The number of candidate records cannot be interpreted directly because some teams played more than one earlier reverse fixture within the search window. A single shoot-out match can therefore have multiple possible earlier matches.

To identify genuine two-legged explanations more precisely, the aggregate score across each candidate pair is calculated.

For a penalty shoot-out to be explained by a two-legged tie, the two teams should be level on aggregate after combining the scores from the earlier reverse fixture and the later shoot-out match.

This provides a stronger validation criterion than simply finding an earlier match between the same teams.

```python
reverse_leg_check["shootout_home_team_aggregate"] = (
    reverse_leg_check["home_score_shootout_match"]
    +
    reverse_leg_check["away_score_possible_first_leg"]
)

reverse_leg_check["shootout_away_team_aggregate"] = (
    reverse_leg_check["away_score_shootout_match"]
    +
    reverse_leg_check["home_score_possible_first_leg"]
)

reverse_leg_check["aggregate_is_level"] = (
    reverse_leg_check["shootout_home_team_aggregate"]
    ==
    reverse_leg_check["shootout_away_team_aggregate"]
)

print(
    "Candidate reverse-leg pairs:",
    len(reverse_leg_check)
)

print(
    "Candidate pairs level on aggregate:",
    reverse_leg_check["aggregate_is_level"].sum()
)

print(
    "Candidate pairs not level on aggregate:",
    (~reverse_leg_check["aggregate_is_level"]).sum()
)
```

```python
aggregate_level_candidates = (
    reverse_leg_check[
        reverse_leg_check["aggregate_is_level"]
    ]
    .copy()
)

print(
    "Aggregate-level candidate pairs:",
    len(aggregate_level_candidates)
)

print(
    "Unique non-level shootout matches explained "
    "by at least one aggregate-level candidate:",
    aggregate_level_candidates[
        [
            "date_shootout_match",
            "home_team_shootout_match",
            "away_team_shootout_match"
        ]
    ].drop_duplicates().shape[0]
)

display(
    aggregate_level_candidates[
        [
            "date_shootout_match",
            "home_team_shootout_match",
            "away_team_shootout_match",
            "home_score_shootout_match",
            "away_score_shootout_match",
            "date_possible_first_leg",
            "home_score_possible_first_leg",
            "away_score_possible_first_leg",
            "shootout_home_team_aggregate",
            "shootout_away_team_aggregate",
            "days_between",
            "winner",
            "tournament_shootout_match"
        ]
    ].sort_values(
        [
            "date_shootout_match",
            "days_between"
        ]
    )
)
```

```python
explained_non_level_keys = (
    aggregate_level_candidates[
        [
            "date_shootout_match",
            "home_team_shootout_match",
            "away_team_shootout_match"
        ]
    ]
    .drop_duplicates()
    .rename(
        columns={
            "date_shootout_match": "date",
            "home_team_shootout_match": "home_team",
            "away_team_shootout_match": "away_team"
        }
    )
)

unexplained_non_level_shootouts = (
    non_level_matched_shootouts
    .merge(
        explained_non_level_keys.assign(
            aggregate_explanation=True
        ),
        on=[
            "date",
            "home_team",
            "away_team"
        ],
        how="left"
    )
)

unexplained_non_level_shootouts = (
    unexplained_non_level_shootouts[
        unexplained_non_level_shootouts[
            "aggregate_explanation"
        ].isna()
    ]
    .copy()
)

explained_count = (
    len(non_level_matched_shootouts)
    -
    len(unexplained_non_level_shootouts)
)

print(
    "Total non-level shootouts:",
    len(non_level_matched_shootouts)
)

print(
    "Explained by an aggregate-level candidate:",
    explained_count
)

print(
    "Still unexplained:",
    len(unexplained_non_level_shootouts)
)

display(
    unexplained_non_level_shootouts[
        [
            "date",
            "home_team",
            "away_team",
            "home_score",
            "away_score",
            "winner",
            "tournament"
        ]
    ]
)
```

#### Investigating Remaining Shoot-Out Exceptions

After accounting for matches whose individual scores were level and non-level matches that could be explained by an earlier reverse fixture producing a level aggregate score, two non-level shoot-out records remained for individual investigation.

External historical evidence was used to investigate these records because their competition-specific tie-breaking rules cannot be determined from the available dataset fields alone.

**Libya 1–0 Syria — Palestine Cup, 23 December 1975**

Historical records confirm the recorded 1–0 match result and also record Syria as winning the subsequent penalty shoot-out. The combination of a non-level match score and a shoot-out is therefore historically supported and is retained without alteration.

**Paraguay 2–0 Argentina — Copa Félix Bogado, 31 August 1977**

Historical records confirm that Paraguay won the match 2–0 and subsequently won the penalty shoot-out 3–1. The competition was decided using points rather than aggregate goal difference. Argentina had won the earlier match 2–1, leaving the teams level on points after one victory each. A penalty shoot-out was therefore used to determine the competition winner despite the second match itself having a non-level score.

The record is therefore valid and is retained without alteration.

The single shoot-out record that did not match `results` was also investigated:

**Saare County vs Åland Islands — 29 June 2011**

Historical records confirm that this was a standalone penalty shoot-out used as a tie-break after the teams finished with identical group records at the 2011 Island Games. It was therefore not a conventional match requiring a corresponding match-result record in `results`.

Åland Islands won the shoot-out 4–3. The record is retained in `shootouts`.

These exceptions demonstrate that a non-level individual match score, or the absence of a conventional match record, does not necessarily indicate invalid shoot-out data. Competition-specific tie-breaking rules must be considered before classifying such records as errors.

No transformations are required.

```python
# Final Shooutout validation summary 
level_shootouts = matched_shootout_records[
    matched_shootout_records["score_is_level"]
]

aggregate_explained_shootouts = (
    aggregate_level_candidates[
        [
            "date_shootout_match",
            "home_team_shootout_match",
            "away_team_shootout_match"
        ]
    ]
    .drop_duplicates()
)

print("Total shootout records:", len(shootouts_clean))
print()
print(
    "Matched to results:",
    len(matched_shootout_records)
)
print(
    "Standalone / unmatched shootout:",
    len(unmatched_shootout_records)
)
print()
print(
    "Matched records with level individual score:",
    len(level_shootouts)
)
print(
    "Non-level records explained by level aggregate candidate:",
    len(aggregate_explained_shootouts)
)
print(
    "Non-level records explained by other competition rules:",
    len(unexplained_non_level_shootouts)
)

accounted_for = (
    len(level_shootouts)
    + len(aggregate_explained_shootouts)
    + len(unexplained_non_level_shootouts)
    + len(unmatched_shootout_records)
)

print()
print(
    "Shootout records accounted for:",
    accounted_for,
    "/",
    len(shootouts_clean)
)
```

#### Shoot-Out Validation Conclusion

Formal cross-dataset validation between `shootouts` and `results` is complete.

The `shootouts` dataset contains 682 records, all with unique `date + home_team + away_team` match keys. All 682 recorded shoot-out winners are one of the two participating teams.

Of the 682 shoot-out records:

- 681 could be linked directly to corresponding records in `results`;
- 1 was not represented as a conventional match in `results`.

Among the 681 matched records:

- 644 have level individual match scores;
- 37 have non-level individual match scores.

The 37 non-level records were investigated rather than automatically classified as errors.

For 35 of these records, an earlier reverse fixture was identified whose scores, when combined with the later shoot-out match, produce a level aggregate score. These records are therefore internally consistent with an aggregate-tie explanation, although the available datasets alone do not establish the competition rules for every individual case.

The remaining two non-level records were investigated individually. Historical evidence supports both records and shows that competition-specific tie-breaking rules explain why a penalty shoot-out occurred despite the individual match score being non-level.

The single record that did not match `results`, Saare County versus Åland Islands on 29 June 2011, was also investigated separately. Historical evidence indicates that this was a standalone tie-breaking penalty shoot-out rather than a conventional match, explaining why no corresponding normal match record exists in `results`.

All 682 shoot-out records are therefore accounted for:

- 644 level-score matched records;
- 35 non-level records with a level-aggregate candidate;
- 2 non-level competition-specific exceptions;
- 1 standalone tie-breaking shoot-out.

The 422 missing `first_shooter` values identified during the original cleaning remain valid missing information and do not affect the match-level or winner validation.

No transformations to `shootouts` or `results` were required during this formal cross-dataset validation.

### Validating Historical Former-Name References

The `former_names` dataset differs from the match-level datasets because it provides historical naming reference information rather than individual match records.

Historical team names have deliberately not been globally replaced during project-wide standardisation. Names such as `Soviet Union` and `Czechoslovakia` can contain meaningful historical context and should not automatically be converted to their modern equivalents.

Formal validation therefore assesses whether former names appearing in the historical match data are broadly consistent with the date ranges recorded in `former_names`.

```python
# Inspecting the structure 
print("Former-name records:", len(former_names_clean))
print()

print("Columns:")
print(former_names_clean.columns.tolist())

print("\nMissing values:")
print(former_names_clean.isna().sum())

print("\nDuplicate rows:")
print(former_names_clean.duplicated().sum())

display(former_names_clean.head(20))
```

```python
# Checking date-range 
former_names_clean["date_range_valid"] = (
    former_names_clean["start_date"]
    <=
    former_names_clean["end_date"]
)

print(
    "Former-name records with valid date ranges:",
    former_names_clean["date_range_valid"].sum(),
    "/",
    len(former_names_clean)
)

print(
    "Invalid date ranges:",
    (~former_names_clean["date_range_valid"]).sum()
)

display(
    former_names_clean[
        ~former_names_clean["date_range_valid"]
    ]
)
```

```python
results_team_names = set(
    results_clean["home_team"].dropna()
).union(
    set(results_clean["away_team"].dropna())
)

former_names_clean["former_appears_in_results"] = (
    former_names_clean["former"].isin(
        results_team_names
    )
)

print(
    "Former names appearing in results:",
    former_names_clean[
        "former_appears_in_results"
    ].sum(),
    "/",
    len(former_names_clean)
)

print(
    "Former names not appearing in results:",
    (
        ~former_names_clean[
            "former_appears_in_results"
        ]
    ).sum()
)

display(
    former_names_clean[
        ~former_names_clean[
            "former_appears_in_results"
        ]
    ][
        [
            "current",
            "former",
            "start_date",
            "end_date"
        ]
    ]
)
```

#### Former-Name Coverage Interpretation

None of the 36 literal `former` labels in `former_names` appears as a home or away team label in `results`.

This does not indicate 36 missing or erroneous historical entities. Instead, it indicates that the wider `results` dataset uses standardised team labels rather than preserving the historical name recorded in the `former` field.

For example, `former_names` records `Soviet Union` as a former name associated with `Russia`, but the historical results dataset does not use the literal `Soviet Union` label. The historical relationship is instead represented through the standardised current-name convention used by that dataset.

Therefore, literal former-name matching cannot be used to validate the date ranges in `former_names`.

The validation is revised to determine whether the corresponding `current` entities occur in `results` during the historical periods represented by the former-name records.

```python
# First checking whether the corresponding current names are in the result dataset. 
former_names_clean["current_appears_in_results"] = (
    former_names_clean["current"].isin(
        results_team_names
    )
)

print(
    "Current names appearing in results:",
    former_names_clean[
        "current_appears_in_results"
    ].sum(),
    "/",
    len(former_names_clean)
)

print(
    "Current names not appearing in results:",
    (
        ~former_names_clean[
            "current_appears_in_results"
        ]
    ).sum()
)

display(
    former_names_clean[
        ~former_names_clean[
            "current_appears_in_results"
        ]
    ][
        [
            "current",
            "former",
            "start_date",
            "end_date"
        ]
    ]
)
```

```python
results_home_team_history = (
    results_clean[
        [
            "date",
            "home_team"
        ]
    ]
    .rename(
        columns={
            "home_team": "team"
        }
    )
)

results_away_team_history = (
    results_clean[
        [
            "date",
            "away_team"
        ]
    ]
    .rename(
        columns={
            "away_team": "team"
        }
    )
)

results_team_history = (
    pd.concat(
        [
            results_home_team_history,
            results_away_team_history
        ],
        ignore_index=True
    )
    .drop_duplicates()
)

print(
    "Team-date combinations in results:",
    len(results_team_history)
)

print(
    "Unique teams in results history:",
    results_team_history["team"].nunique()
)

display(results_team_history.head())
```

```python
# Checking whether current names occur during forner-name periods 
former_period_validation = (
    former_names_clean
    .merge(
        results_team_history,
        left_on="current",
        right_on="team",
        how="left"
    )
)

former_period_validation["within_former_period"] = (
    (
        former_period_validation["date"]
        >=
        former_period_validation["start_date"]
    )
    &
    (
        former_period_validation["date"]
        <=
        former_period_validation["end_date"]
    )
)

former_period_summary = (
    former_period_validation
    .groupby(
        [
            "current",
            "former",
            "start_date",
            "end_date"
        ],
        as_index=False
    )
    .agg(
        matches_within_period=(
            "within_former_period",
            "sum"
        )
    )
)

former_period_summary[
    "historical_period_represented"
] = (
    former_period_summary[
        "matches_within_period"
    ] > 0
)

print(
    "Former-name periods with corresponding "
    "current-team match history:",
    former_period_summary[
        "historical_period_represented"
    ].sum(),
    "/",
    len(former_period_summary)
)

print(
    "Periods with no corresponding match history:",
    (
        ~former_period_summary[
            "historical_period_represented"
        ]
    ).sum()
)
```

```python
# There are no periods with no match coverage but will do a secondary check
former_period_no_coverage = (
    former_period_summary[
        ~former_period_summary[
            "historical_period_represented"
        ]
    ]
    .copy()
)

display(
    former_period_no_coverage[
        [
            "current",
            "former",
            "start_date",
            "end_date",
            "matches_within_period"
        ]
    ]
)
```

#### Historical Former-Name Validation Conclusion

Formal cross-dataset validation of `former_names` against the wider historical match data is complete.

The `former_names` dataset contains 36 historical naming records. All 36 records:

- contain complete values for `current`, `former`, `start_date` and `end_date`;
- have valid chronological date ranges;
- are unique, with no duplicate rows.

None of the 36 literal `former` labels appears as a home or away team name in `results`. However, all 36 corresponding `current` names are represented in the historical results data.

A date-aware comparison was therefore performed using the standardised `current` names. The home and away team records from `results` were combined into a temporary team-date history containing 98,866 unique team-date combinations across 336 team labels.

All 36 historical former-name periods contain at least one match involving the corresponding `current` team during the stated historical date range.

This demonstrates that `results` and `former_names` use different but internally compatible naming conventions: `results` generally uses standardised team names, while `former_names` preserves the historical naming relationships and their associated date ranges.

This comparison establishes cross-dataset coverage and consistency. It does not independently prove that every historical start and end date is externally correct.

Historical labels preserved elsewhere in the project are not globally replaced on the basis of this reference table because they may contain meaningful historical context.

No transformations to `former_names` or `results` were required during this formal validation.

```python
# Checking the totals, date ranges, and goals_per_match calculation for wc_editions dataset 
print("World Cup editions:", len(editions_clean))

print(
    "Invalid tournament date ranges:",
    (
        editions_clean["start_date"]
        >
        editions_clean["end_date"]
    ).sum()
)

print(
    "Teams <= 0:",
    (editions_clean["teams"] <= 0).sum()
)

print(
    "Matches <= 0:",
    (editions_clean["matches"] <= 0).sum()
)

print(
    "Goals < 0:",
    (editions_clean["goals"] < 0).sum()
)

print(
    "Attendance < 0:",
    (editions_clean["attendance"] < 0).sum()
)

calculated_goals_per_match = (
    editions_clean["goals"]
    /
    editions_clean["matches"]
)

goals_per_match_mismatch = (
    ~np.isclose(
        editions_clean["goals_per_match"],
        calculated_goals_per_match,
        atol=0.01
    )
)

print(
    "Goals-per-match inconsistencies:",
    goals_per_match_mismatch.sum()
)
```

```python
finishing_position_checks = []

for _, row in editions_clean.iterrows():

    finishing_teams = [
        row["champion"],
        row["runner_up"],
        row["third_place"],
        row["fourth_place"]
    ]

    finishing_position_checks.append(
        {
            "year": row["year"],
            "unique_finishing_teams":
                len(set(finishing_teams)),
            "positions_valid":
                len(set(finishing_teams)) == 4
        }
    )

finishing_position_checks = pd.DataFrame(
    finishing_position_checks
)

print(
    "Editions with four distinct finishing teams:",
    finishing_position_checks[
        "positions_valid"
    ].sum(),
    "/",
    len(finishing_position_checks)
)

display(
    finishing_position_checks[
        ~finishing_position_checks["positions_valid"]
    ]
)
```

```python
# Checking wc_matches 
print("Selected World Cup matches:", len(matches_clean))

print(
    "Negative score1 values:",
    (matches_clean["score1"] < 0).sum()
)

print(
    "Negative score2 values:",
    (matches_clean["score2"] < 0).sum()
)

print(
    "Matches with identical team1 and team2:",
    (
        matches_clean["team1"]
        ==
        matches_clean["team2"]
    ).sum()
)

print(
    "Missing match dates:",
    matches_clean["date"].isna().sum()
)

print(
    "Duplicate complete rows:",
    matches_clean.duplicated().sum()
)
```

```python
# checking wc_top_scorers 
print("Top-scorer records:", len(top_scorers_clean))

print(
    "Goals < 0:",
    (top_scorers_clean["goals"] < 0).sum()
)

print(
    "Assists < 0:",
    (top_scorers_clean["assists"] < 0).sum()
)

print(
    "Penalties < 0:",
    (top_scorers_clean["penalties"] < 0).sum()
)

print(
    "Matches played <= 0:",
    (top_scorers_clean["matches_played"] <= 0).sum()
)

print(
    "Duplicate edition-year combinations:",
    top_scorers_clean.duplicated(
        subset=["edition", "year"]
    ).sum()
)

print(
    "Unique editions:",
    top_scorers_clean["edition"].nunique()
)

print(
    "Unique years:",
    top_scorers_clean["year"].nunique()
)
```

```python
# Ensuring the 2 edition indentifiers still agree between edition and top-scorer datasets: 
edition_year_reference = (
    editions_clean[
        ["edition", "year"]
    ]
    .drop_duplicates()
)

top_scorer_edition_check = (
    top_scorers_clean[
        ["edition", "year"]
    ]
    .merge(
        edition_year_reference.assign(
            edition_year_valid=True
        ),
        on=["edition", "year"],
        how="left"
    )
)

print(
    "Top-scorer edition/year records checked:",
    len(top_scorer_edition_check)
)

print(
    "Invalid edition/year combinations:",
    top_scorer_edition_check[
        "edition_year_valid"
    ].isna().sum()
)
```

#### World Cup Dataset Validity Conclusion

Final validity checks across the World Cup-specific historical datasets identified no new structural or logical issues.

All 22 World Cup edition records have valid tournament date ranges, positive team and match counts, non-negative goals and attendance, and internally consistent goals-per-match calculations. Each edition also contains four distinct recorded finishing teams.

The selected `wc_matches` dataset contains 184 records with no negative scores, no matches in which a team plays itself, no missing match dates and no exact duplicate rows.

The `wc_top_scorers` dataset contains one record for each of the 22 editions and years, with no invalid negative values or non-positive match counts. All 22 edition/year combinations align with the corresponding World Cup edition records.

The structural validity of the `assists` and `penalties` fields does not override the data-quality limitations identified during independent cleaning. These fields remain potentially incomplete and should therefore be treated cautiously in later analysis.

No further transformations are required as a result of these checks.

```python
# Checking final schemas 
expected_schemas = {
    "results": [
        "date", "home_team", "away_team", "home_score", "away_score",
        "tournament", "city", "country", "neutral", "match_completed",
        "total_goals", "goal_difference", "result", "year", "month"
    ],

    "goalscorers": [
        "date", "home_team", "away_team", "team", "scorer",
        "minute", "own_goal", "penalty", "minute_known"
    ],

    "shootouts": [
        "date", "home_team", "away_team", "winner",
        "first_shooter", "first_shooter_known"
    ],

    "former_names": [
        "current", "former", "start_date", "end_date"
    ],

    "2026_fixtures": [
        "group", "stage", "team1", "team2", "venue", "city",
        "country", "date", "kickoff_et",
        "team1_confederation", "team1_fifa_rank", "team1_coach",
        "team2_confederation", "team2_fifa_rank", "team2_coach"
    ],

    "2026_teams": [
        "team", "group", "confederation", "fifa_rank",
        "coach", "best_wc_result", "debut_2026"
    ],

    "wc_editions": [
        "edition", "year", "host", "champion", "runner_up",
        "third_place", "fourth_place", "teams", "matches", "goals",
        "goals_per_match", "attendance", "top_scorer",
        "top_scorer_country", "top_scorer_goals",
        "start_date", "end_date", "final_city", "host_won", "format"
    ],

    "wc_matches": [
        "year", "stage", "team1", "score1", "score2", "team2",
        "venue", "city", "country", "date", "notes"
    ],

    "wc_top_scorers": [
        "edition", "year", "player", "country", "position",
        "goals", "assists", "penalties", "matches_played",
        "host", "team_result"
    ]
}

schema_check = []

for name, df in project_datasets.items():

    expected = expected_schemas[name]
    actual = df.columns.tolist()

    schema_check.append(
        {
            "dataset": name,
            "expected_columns": len(expected),
            "actual_columns": len(actual),
            "schema_matches": actual == expected
        }
    )

schema_check = pd.DataFrame(schema_check)

display(schema_check)

print(
    "\nAll final schemas correct:",
    schema_check["schema_matches"].all()
)
```

```python
# Final key check - shouldn't expect every dataset to have one unique row per match 
key_checks = []

# Shootouts: one shootout record per stored match key
key_checks.append(
    {
        "dataset": "shootouts",
        "key": "date + home_team + away_team",
        "duplicate_keys":
            shootouts_clean.duplicated(
                subset=[
                    "date",
                    "home_team",
                    "away_team"
                ]
            ).sum()
    }
)

# 2026 teams: one row per team
key_checks.append(
    {
        "dataset": "2026_teams",
        "key": "team",
        "duplicate_keys":
            teams_clean.duplicated(
                subset=["team"]
            ).sum()
    }
)

# World Cup editions: one row per edition/year
key_checks.append(
    {
        "dataset": "wc_editions",
        "key": "edition + year",
        "duplicate_keys":
            editions_clean.duplicated(
                subset=["edition", "year"]
            ).sum()
    }
)

# Top scorers: intentionally one stored row per edition/year
key_checks.append(
    {
        "dataset": "wc_top_scorers",
        "key": "edition + year",
        "duplicate_keys":
            top_scorers_clean.duplicated(
                subset=["edition", "year"]
            ).sum()
    }
)

key_checks = pd.DataFrame(key_checks)

display(key_checks)
```

```python
# Confirming final row counts 
expected_row_counts = {
    "results": 49505,
    "goalscorers": 47886,
    "shootouts": 682,
    "former_names": 36,
    "2026_fixtures": 104,
    "2026_teams": 48,
    "wc_editions": 22,
    "wc_matches": 184,
    "wc_top_scorers": 22
}

row_count_check = []

for name, df in project_datasets.items():

    expected_rows = expected_row_counts[name]
    actual_rows = len(df)

    row_count_check.append(
        {
            "dataset": name,
            "expected_rows": expected_rows,
            "actual_rows": actual_rows,
            "matches_expected": (
                actual_rows == expected_rows
            )
        }
    )

row_count_check = pd.DataFrame(
    row_count_check
)

display(row_count_check)

print(
    "\nAll row counts correct:",
    row_count_check[
        "matches_expected"
    ].all()
)
```

```python
all_schemas_valid = (
    schema_check["schema_matches"].all()
)

all_checked_keys_valid = (
    key_checks["duplicate_keys"].eq(0).all()
)

all_row_counts_valid = (
    row_count_check["matches_expected"].all()
)

print("FINAL PRE-EXPORT AUDIT")
print("-" * 40)

print(
    "All expected schemas correct:",
    all_schemas_valid
)

print(
    "All checked unique keys valid:",
    all_checked_keys_valid
)

print(
    "All expected row counts correct:",
    all_row_counts_valid
)

print()

if (
    all_schemas_valid
    and all_checked_keys_valid
    and all_row_counts_valid
):
    print(
        "STATUS: PASSED — datasets are ready "
        "for final export."
    )
else:
    print(
        "STATUS: REVIEW REQUIRED — investigate "
        "before export."
    )
```

#### Conclusion

The final pre-export audit passed successfully across all nine cleaned datasets.

All datasets contain their intended final schemas, with no temporary validation columns remaining. All expected row counts have been preserved, confirming that no records were unintentionally removed during project-wide standardisation or formal cross-dataset validation.

All fields expected to act as unique identifiers within the checked datasets also remain unique.

Combined with the previous missingness, data-type, numerical, logical and cross-dataset validation checks, the final working datasets are considered ready for export.

The remaining duplicate and missing-value patterns identified during quality control are previously investigated and intentionally retained rather than unresolved cleaning errors.

```python
final_output_paths = {
    "results": processed_dir / "results_clean.csv",
    "goalscorers": processed_dir / "goalscorers_clean.csv",
    "shootouts": processed_dir / "shootouts_clean.csv",
    "former_names": processed_dir / "former_names_clean.csv",

    "2026_fixtures": processed_dir / "wc_2026_fixtures_clean.csv",
    "2026_teams": processed_dir / "wc_2026_teams_clean.csv",
    "wc_editions": processed_dir / "wc_all_editions_clean.csv",
    "wc_matches": processed_dir / "wc_all_matches_clean.csv",
    "wc_top_scorers": processed_dir / "wc_top_scorers_clean.csv"
}

for name, path in final_output_paths.items():
    print(f"{name:<20} -> {path}")
```

```python
processed_dir.mkdir(
    parents=True,
    exist_ok=True
)

for name, df in project_datasets.items():

    output_path = final_output_paths[name]

    df.to_csv(
        output_path,
        index=False
    )

    print(
        f"Saved {name:<20} "
        f"-> {output_path.name}"
    )
```

```python
print("FINAL EXPORT FILE CHECK")
print("-" * 50)

for name, path in final_output_paths.items():

    print(
        f"{name:<20} "
        f"exists={path.exists()}  "
        f"file={path.name}"
    )
```

```python
reloaded_datasets = {
    name: pd.read_csv(path)
    for name, path in final_output_paths.items()
}

print("Reloaded final processed datasets:")
print()

for name, df in reloaded_datasets.items():

    print(
        f"{name:<20} "
        f"rows={len(df):>6}  "
        f"columns={df.shape[1]:>2}"
    )
```

```python
reload_verification = []

for name in project_datasets:

    original_df = project_datasets[name]
    reloaded_df = reloaded_datasets[name]

    reload_verification.append(
        {
            "dataset": name,

            "rows_match":
                len(original_df)
                ==
                len(reloaded_df),

            "columns_match":
                original_df.shape[1]
                ==
                reloaded_df.shape[1],

            "column_names_match":
                original_df.columns.tolist()
                ==
                reloaded_df.columns.tolist()
        }
    )

reload_verification = pd.DataFrame(
    reload_verification
)

display(reload_verification)

print(
    "\nAll exported structures verified:",
    (
        reload_verification[
            [
                "rows_match",
                "columns_match",
                "column_names_match"
            ]
        ]
        .all()
        .all()
    )
)
```

```python
export_value_checks = []

for name in project_datasets:

    original_compare = (
        project_datasets[name]
        .copy()
        .reset_index(drop=True)
    )

    reloaded_compare = (
        reloaded_datasets[name]
        .copy()
        .reset_index(drop=True)
    )

    # Convert datetime columns to the same textual
    # representation used in the CSV files
    for column in original_compare.columns:

        if pd.api.types.is_datetime64_any_dtype(
            original_compare[column]
        ):
            original_compare[column] = (
                original_compare[column]
                .dt.strftime("%Y-%m-%d")
            )

    # Convert both sides to comparable string values,
    # while preserving missing values consistently
    original_compare = (
        original_compare
        .astype("string")
        .fillna("<MISSING>")
    )

    reloaded_compare = (
        reloaded_compare
        .astype("string")
        .fillna("<MISSING>")
    )

    values_match = (
        original_compare.equals(
            reloaded_compare
        )
    )

    export_value_checks.append(
        {
            "dataset": name,
            "exported_values_match": values_match
        }
    )

export_value_checks = pd.DataFrame(
    export_value_checks
)

display(export_value_checks)

print(
    "\nAll exported values verified:",
    export_value_checks[
        "exported_values_match"
    ].all()
)
```

## Final Cleaning Conclusion

This notebook completed the cleaning, validation and project-wide standardisation of the additional FIFA World Cup datasets and integrated them with the datasets prepared in the initial cleaning notebook.

From the 2 folders given for this project, we have nine cleaned datasets:

- `results_clean`
- `goalscorers_clean`
- `shootouts_clean`
- `former_names_clean`
- `wc_2026_fixtures_clean`
- `wc_2026_teams_clean`
- `wc_all_editions_clean`
- `wc_all_matches_clean`
- `wc_top_scorers_clean`

### Cleaning approach

Each dataset was first assessed independently before any cross-dataset corrections were made. This included checking dimensions, missing values, duplicates, data types, categorical values, date ranges, numerical validity and dataset-specific structural rules.

Missing values were not automatically removed or imputed. Instead, their meaning was investigated and structurally valid missingness was retained. This includes incomplete historical match scores, unknown scorer names and goal minutes, unavailable first-shooter information, unresolved 2026 knockout-stage team metadata and optional World Cup match notes.

Historical and categorical information was also preserved where it carried meaningful context rather than being globally standardised without justification.

### Corrections and standardisation

A number of source-data inconsistencies were identified through internal and cross-dataset validation.

The World Cup editions data required corrections to historical host-winner information and the 1950 top-scorer goal total.

The selected historical World Cup match dataset contained a number of incorrect scores, dates, teams and locations. These were investigated against the wider international results data, resulting in 70 corrected cells across 40 records. Following correction, all 184 selected World Cup matches could be matched to corresponding FIFA World Cup records in the wider results dataset and all 184 scores were consistent.

The World Cup top-scorer dataset required four corrections across four records, including the 1950 goal total and incorrect recorded team results.

Project-wide naming standardisation was then applied only where differences represented formatting or modern naming conventions rather than meaningful historical distinctions. This resulted in 112 standardised cells across the project, including city, location-country, host and team-name fields.

Historical identities such as `West Germany`, `East Germany`, `Soviet Union` and `Czechoslovakia` were deliberately preserved where they provide historical context.

### Cross-dataset validation

Formal cross-dataset validation was performed only between datasets with meaningful relationships rather than assuming every dataset should directly correspond with every other dataset.

The following relationships were validated:

- 2026 fixtures against 2026 team metadata;
- World Cup editions against selected World Cup matches;
- World Cup editions against World Cup top scorers;
- selected World Cup matches against the wider international results data;
- goal events against match results;
- penalty shoot-outs against match results;
- historical former-name references against the wider results data.

All 48 named 2026 teams were represented in the fixture data, while progression labels used for future knockout matches were retained as structural placeholders. FIFA ranking, confederation and coach metadata were consistent between the named fixture teams and the team dataset.

All 47,886 goal-event records linked to corresponding match results. Across the 15,532 represented matches, aggregated goal events reproduced the recorded home and away scores exactly, including the treatment of own goals.

All 682 penalty shoot-out records were accounted for. Most were associated with level individual match scores, while non-level cases were investigated for aggregate or competition-specific tie-breaking explanations. One standalone tie-breaking shoot-out was retained despite having no corresponding conventional match record.

The historical former-name table was also consistent with the wider match history. All 36 current entities were represented in the results data and all 36 historical naming periods had corresponding match coverage. The reference table was retained without globally replacing historically meaningful labels elsewhere in the project.

### Known limitations retained

Some fields remain unsuitable for unrestricted analysis despite being structurally valid.

The `assists` and `penalties` fields in the World Cup top-scorer dataset appear historically incomplete and should be treated cautiously in later analysis.

The selected `wc_all_matches` dataset contains 184 selected historical World Cup matches and should not be interpreted as a complete match-level history of every World Cup.

The 2026 fixture dataset contains progression placeholders and structurally missing team metadata for knockout matches whose participants are not yet determined.

Historical naming conventions also differ between datasets. These differences have been retained where they represent meaningful historical context rather than data-quality errors.

### Final quality assurance

A final project-wide quality-control process confirmed:

- expected dataset dimensions and schemas;
- appropriate data types;
- valid date fields;
- expected and justified missingness;
- valid numerical ranges;
- internally consistent derived match fields;
- valid shoot-out winners;
- valid FIFA rankings;
- internally consistent World Cup edition statistics;
- preserved expected row counts;
- and no unintended temporary validation columns.

All nine final datasets were exported to the processed-data directory and reloaded successfully. Their row counts, schemas, column names and exported values were verified against the final in-memory cleaned datasets.

The cleaning and validation phase is therefore complete.
