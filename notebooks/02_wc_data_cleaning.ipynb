### Dataset Quality Overview

A function is created to provide an initial data-quality summary for each dataset.

For each dataset, the function will reports:

- No. of rows and columns;
- The data type of each column;
- The number of missing values;
- The percentage of values that are missing;
- The number of unique values; and
- The number of exact duplicate rows.

This provides a consistent initial assessment before any cleaning decisions are made.

```python
def dataset_overview(df, name):
    print(f"DATASET: {name}")
    print("=" * 60)

    print(f"Rows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")

    overview = pd.DataFrame({
        "data_type": df.dtypes,
        "missing_values": df.isna().sum(),
        "missing_percentage": (df.isna().mean() * 100).round(2),
        "unique_values": df.nunique(dropna=True)
    })

    display(overview)

    print("Exact duplicate rows:", df.duplicated().sum())
```

## Investigation into the Missing Values in 2026 Fixtures

Several fixture-related columns contain exactly 32 missing values. This will first be inspected to determine the appropriate decision.

```python
fixtures_clean[
    fixtures_clean["team1_fifa_rank"].isna()
][
    [
        "group",
        "stage",
        "team1",
        "team2",
        "date",
        "team1_confederation",
        "team1_fifa_rank",
        "team1_coach",
        "team2_confederation",
        "team2_fifa_rank",
        "team2_coach"
    ]
]
```

#### Conclusion 
Inspection of the 32 records containing missing team information showed that the missing values are structural rather than data-quality errors.

All affected records correspond to knockout-stage fixtures. The `team1` and `team2` columns contain tournament progression placeholders such as `1A`, `2B`, `Best 3rd #1`, `R32 W1`, `QF1`, `SF1` and `Finalist 1`, rather than confirmed team names.

Because the actual participating teams depend on the results of earlier tournament matches, attributes such as confederation, FIFA ranking and coach cannot be assigned to these placeholders.

The `group` field is also missing for these records because knockout-stage matches do not belong to a World Cup group.

Therefore:

- the 32 fixture records will be retained;
- the missing `group` values will be retained as not applicable to knockout-stage fixtures;
- missing confederation, FIFA ranking and coach values will not be imputed;
- tournament progression placeholders in `team1` and `team2` will be preserved;
- no rows will be removed because of these missing values.

This prevents artificial information from being introduced into the dataset and preserves the original tournament structure.

### Counting missing information by tournament stage

```python
fixtures_clean.groupby("stage").agg(
    total_fixtures=("stage", "size"),
    missing_group=("group", lambda x: x.isna().sum()),
    missing_team1_rank=("team1_fifa_rank", lambda x: x.isna().sum()),
    missing_team2_rank=("team2_fifa_rank", lambda x: x.isna().sum())
)
```

#### Conclusion 

Grouping the fixtures by tournament stage confirmed that the missing values follow a clear structural pattern.

All 72 Group Stage fixtures contain complete `group`, `team1_fifa_rank` and `team2_fifa_rank` information.

In contrast, all 32 knockout-stage fixtures contain missing values in these fields:

- 16 Round of 32 fixtures;
- 8 Round of 16 fixtures;
- 4 Quarter-final fixtures;
- 2 Semi-final fixtures;
- 1 Third Place Match; and
- 1 Final.

These 32 fixtures account for all of the missing values identified during the initial data-quality assessment.

This confirms that the missing values are associated with knockout-stage tournament placeholders rather than being randomly missing data. The values will therefore be retained as missing and will not be imputed or used as a reason to remove the fixture records.

### Checking text columns for Leading and Trailing Whitespace

Text columns are checked for leading or trailing whitespace.

Unnecessary whitespace can cause identical values to be treated as different categories and may create problems when datasets are joined using team names or other categorical variables.

The check is performed before modifying the values so that any required cleaning can be justified.

```python
fixture_text_columns = [
    "group",
    "stage",
    "team1",
    "team2",
    "venue",
    "city",
    "country",
    "kickoff_et",
    "team1_confederation",
    "team1_coach",
    "team2_confederation",
    "team2_coach"
]

for col in fixture_text_columns:
    non_missing = fixtures_clean[col].dropna()

    whitespace_count = (
        non_missing.astype(str)
        != non_missing.astype(str).str.strip()
    ).sum()

    print(f"{col}: {whitespace_count}")
```

#### Conclusion

No leading or trailing whitespace was identified in any of the inspected text columns.

This includes the group, stage, team, venue, city, country, kick-off time, confederation and coach fields.

Therefore, no whitespace cleaning is required for these columns and the existing text values will be retained.

### Validate Fixture Dates

The `date` column is currently stored as an object rather than a datetime value.

Before converting the column, the dates are parsed into a temporary series using `errors="coerce"`.

Any value that cannot be interpreted as a valid date will become a missing value (`NaT`) in the temporary result. This allows invalid dates to be identified before the original column is changed.

```python
fixture_dates_test = pd.to_datetime(
    fixtures_clean["date"],
    errors="coerce"
)

print("Invalid dates:", fixture_dates_test.isna().sum())
print("Earliest date:", fixture_dates_test.min())
print("Latest date:", fixture_dates_test.max())
```

#### Conclusion

All fixture dates were successfully parsed as valid datetime values, with zero invalid dates identified.

The fixture dates range from 11 June 2026 to 19 July 2026.

No invalid or unparseable date values were identified. Therefore, the `date` column can safely be converted from an object data type to a datetime data type.
