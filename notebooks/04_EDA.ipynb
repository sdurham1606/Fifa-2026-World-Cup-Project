##### Who has competed in the most World Cup tournaments?

```python
results_clean_df.columns
```

```python
# mask to filter out only FIFA World Cup matches
world_cup_mask = results_clean_df['tournament'] == 'FIFA World Cup'
```

```python
# create a dataframe of all teams that have participated in the FIFA World Cup, along with the year of participation
teams = pd.concat([
    results_clean_df[world_cup_mask][['year', 'home_team']].rename(columns={'home_team': 'team'}),
    results_clean_df[world_cup_mask][['year', 'away_team']].rename(columns={'away_team': 'team'})
])

# print(teams['year'].max())

# create a dataframe of the number of appearances for each team in the FIFA World Cup
world_cup_appearances = (
    teams.groupby('team')['year']
         .nunique()
         .reset_index(name='appearances')
         .sort_values(by='appearances', ascending=False)
         .reset_index(drop=True)
)

world_cup_appearances_plot = px.bar(world_cup_appearances.head(20), x='team', y='appearances', title='Top 20 World Cup Appearances by Country (1930 - 2026)', labels={'team': 'Country', 'appearances': 'Number of Appearances'}, color='appearances')

# print("Total World Cup Teams:", teams['team'].nunique())

# print(world_cup_appearances.sort_values(by='team').to_string())
# print("West Germany appearances:",world_cup_appearances[world_cup_appearances['team'] == 'West Germany']['appearances'].count())

world_cup_appearances_plot.show()
world_cup_appearances_plot.write_html(f"{FINAL_PLOT_DIR}/world_cup_appearances.html")
# Wikipedia says 84, but could be because they have combined some country names despite historical changes in circumstances. e.g. Germany instead of West Germany
```

From the above chart, it can be seen that Brazil has had the most historical appearances at the World Cup (23) closely followed by Germany and Argentina (21 & 19 respectively). Paraguay, Hungary, Scotland and Chile have had the fewest (out of the top 20), with 9 appearances each.

##### Does the appearances align with the number of tournaments won, and therefore contribute to the team's success in the FIFA World Cup?
A ratio of `wins_per_appearance` will be calculated, since some teams have had more opportunity to win depending on how many appearances they have had. We want to purely investigate whether having more World Cup experience informs whether a team is likely to win the World Cup or not.

```python
wins_vs_appearances = pd.merge(world_cup_wins.rename(columns={'champion':'team'}), world_cup_appearances, left_on='team', right_on='team', how='inner')

wins_vs_appearances['wins_per_appearance'] = round((wins_vs_appearances['Wins'] / wins_vs_appearances['appearances']),3)

# display(wins_vs_appearances)

wins_vs_appearances_plot = px.scatter(wins_vs_appearances[wins_vs_appearances['Wins'] > 0], x='appearances', y='wins_per_appearance', text='team', title='World Cup Wins per Appearance vs Appearances by Country', labels={'appearances': 'Number of Appearances', 'wins_per_appearance': 'Number of Wins per Appearance'}, color='wins_per_appearance', size='wins_per_appearance')

wins_vs_appearances_plot.update_traces(textposition='top center')

wins_vs_appearances_matrix = px.imshow(wins_vs_appearances[['wins_per_appearance', 'appearances']].rename(columns={'wins_per_appearance': 'Number of Wins per Appearance', 'appearances':'Number of Appearances'}).corr(), text_auto=True, zmin=-1, zmax=1)

wins_vs_appearances_plot.show()
wins_vs_appearances_matrix.show()

wins_vs_appearances_plot.write_html(f"{FINAL_PLOT_DIR}/wins_vs_appearances.html")
wins_vs_appearances_matrix.write_html(f"{FINAL_PLOT_DIR}/wins_vs_appearances_matrix.html")
# West Germany doesn't show any appearances due to being named Germany in the results dataset, despite having 3 wins

#TODO Wins covers 1930 - 2022, appearances covers 1930 - 2026. Need to correct the values based on this
```

The scatter plot above shows a general positive correlation between nunber of appearances vs number of wins per appearance. Brazil, Argentina, Italy and Germany all feature in the plot, which is interesting as they are the countries with the highest number of historical appearances. Not only this, but the other countries in the plot also feature in the Top 10 number of appearances.

This is backed by the correlation of 0.65, which is closer to 1 than 0, showing a relatively strong positive correlation.

##### Does the average number of goals scored per match have any bearing on historical wins?

```python
# number of goals scored by each team during World Cup
home_matches = results_clean_df[world_cup_mask][['date','home_team', 'home_score']]
# print(home_matches[home_matches['home_team']=='France'].groupby('home_team')['home_score'].sum())
away_matches = results_clean_df[world_cup_mask][['date', 'away_team', 'away_score']]
# print(away_matches[away_matches['away_team']=='France'].groupby('away_team')['away_score'].sum())

team_goals = pd.concat([home_matches.rename(columns={'home_team':'team', 'home_score': 'score'}),
                        away_matches.rename(columns={'away_team':'team', 'away_score':'score'})]
                       ).groupby('team')['score'].agg(goals_scored='sum')   

# print(team_goals.head())           
# print(team_goals[team_goals['team']=='France'].groupby('team')['score'].agg('sum'))                                   

# number of matches played by each team during World Cup
team_matches = pd.concat([home_matches.rename(columns={'home_team':'team'}),
                         away_matches.rename(columns={'away_team':'team'})]
                         ).groupby('team')['date'].agg(matches_played='count')

# print(team_matches.head())

# combine dataframes

team_goals_per_match = team_goals.merge(team_matches, how='inner', on='team').merge(wins_vs_appearances, how='inner', left_on='team', right_on='team')

# display(team_goals_per_match)

# check number of matches is 1064 as expected
# print(team_goals_per_match['matches_played'].sum()/2)

# add new column for 'average goals per match'
team_goals_per_match['average_goals_per_match'] = round(team_goals_per_match['goals_scored'] / team_goals_per_match['matches_played'], 3)

# display(team_goals_per_match.sort_values(by='Wins', ascending=False))

# create scatter plot to show average goals per match vs number of wins for each team

team_goals_per_match_plot = px.scatter(team_goals_per_match, x='wins_per_appearance', y='average_goals_per_match', text='team', title='How Does Average Goals per Match Affect Number of Wins per Appearance? 1930 - 2026', color='wins_per_appearance', size='wins_per_appearance', labels={'average_goals_per_match':'Average Goals per Match', 'wins_per_appearance':'Number of Wins per Appearance'})

team_goals_per_match_plot.update_traces(textposition='top center')

team_goals_per_match_plot.show()

team_goals_per_match_matrix = px.imshow(team_goals_per_match[['average_goals_per_match', 'wins_per_appearance']].rename(columns={'average_goals_per_match':'Average Goals per Match', 'wins_per_appearance':'Number of Wins per Appearance'}).corr(), text_auto=True, zmin=-1, zmax=1)

team_goals_per_match_matrix.show()

team_goals_per_match_plot.write_html(f"{FINAL_PLOT_DIR}/team_goals_per_match.html")
team_goals_per_match_matrix.write_html(f"{FINAL_PLOT_DIR}/team_goals_per_match_matrix.html")

# Should we compare vs wins/appearances instead of just wins - so that it's normalised? Since we've seen above that number of appearances does influence number of wins. Unless that comparison should also be wins/appearances vs appearances?
```

The correlation matrix shows a correlation of 0.52, which is closer to 1 than it is to 0, suggesting that there is some linearity between the average number of goals scored per match and the number of historical wins a team has had.
