##### Does the number of team appearances align with the number of tournaments won, and therefore contribute to the team's success in the FIFA World Cup?
A ratio of `wins_per_appearance` will be calculated, since some teams have had more opportunity to win depending on how many appearances they have had. We want to purely investigate whether having more World Cup experience informs whether a team is likely to win the World Cup or not. Since the `world_cup_wins` only goes up to 2022, a new dataframe to count world cup appearances will be created that only spans 1930-2022 for consistency.

```python
# create a dataframe of the number of appearances for each team in the FIFA World Cup exclusing 2026
world_cup_appearances_exc_26 = (
    teams[teams['year'] < 2026].groupby('team')['year']
         .nunique()
         .reset_index(name='appearances')
         .sort_values(by='appearances', ascending=False)
         .reset_index(drop=True)
)

wins_vs_appearances_exc_26 = pd.merge(world_cup_wins.rename(columns={'champion':'team'}), world_cup_appearances_exc_26, left_on='team', right_on='team', how='inner')

wins_vs_appearances_exc_26['wins_per_appearance'] = round((wins_vs_appearances_exc_26['Wins'] / wins_vs_appearances_exc_26['appearances']),3)

# display(wins_vs_appearances)

wins_vs_appearances_plot = px.scatter(wins_vs_appearances_exc_26[wins_vs_appearances_exc_26['Wins'] > 0], x='appearances', y='wins_per_appearance', text='team', title='World Cup Wins per Appearance vs Appearances by Country (1930 - 2022)', labels={'appearances': 'Number of Appearances', 'wins_per_appearance': 'Number of Wins per Appearance'}, color='wins_per_appearance', size='wins_per_appearance')

wins_vs_appearances_plot.update_traces(textposition='top center')

wins_vs_appearances_matrix = px.imshow(wins_vs_appearances_exc_26[['wins_per_appearance', 'appearances']].rename(columns={'wins_per_appearance': 'Number of Wins per Appearance', 'appearances':'Number of Appearances'}).corr(), text_auto=True, zmin=-1, zmax=1)

wins_vs_appearances_plot.show(config=config)
wins_vs_appearances_matrix.show(config=config)

wins_vs_appearances_plot.write_html(f"{FINAL_PLOT_DIR}/wins_vs_appearances.html")
wins_vs_appearances_matrix.write_html(f"{FINAL_PLOT_DIR}/wins_vs_appearances_matrix.html")
```

The scatter plot above shows a general positive correlation between nunber of appearances vs number of wins per appearance. Brazil, Argentina, Italy and Germany all feature in the plot, which is interesting as they are the countries with the highest number of historical appearances. Not only this, but the other countries in the plot also feature in the Top 10 number of appearances.

This is backed by the correlation of 0.72, which is closer to 1 than 0, showing a relatively strong positive correlation.

##### Does the average number of goals scored per match have any bearing on historical wins?

```python
# wins_vs_appearances including 2026
# wins_vs_appearances = pd.merge(world_cup_wins.rename(columns={'champion':'team'}), world_cup_appearances, left_on='team', right_on='team', how='inner')

# wins_vs_appearances['wins_per_appearance'] = round((wins_vs_appearances['Wins'] / wins_vs_appearances['appearances']),3)

# number of goals scored by each team during World Cup
home_matches = results_clean_df[world_cup_mask & (results_clean_df['year'] < 2026)][['date','home_team', 'home_score']]
# print(home_matches[home_matches['home_team']=='France'].groupby('home_team')['home_score'].sum())
away_matches = results_clean_df[world_cup_mask & (results_clean_df['year'] < 2026)][['date', 'away_team', 'away_score']]
# print(away_matches[away_matches['away_team']=='France'].groupby('away_team')['away_score'].sum())

team_goals = pd.concat([home_matches.rename(columns={'home_team':'team', 'home_score': 'score'}),
                        away_matches.rename(columns={'away_team':'team', 'away_score':'score'})]
                       ).groupby('team')['score'].agg(goals_scored='sum')   

# print(team_goals.head())           
# print(team_goals[team_goals['team']=='France'].groupby('team')['score'].agg('sum'))                                   

# number of matches played by each team during World Cup
team_matches = pd.concat([home_matches.rename(columns={'home_team':'team'}),
                         away_matches.rename(columns={'away_team':'team'})]
                         ).groupby('team')['date'].agg(matches_played='count')

# print(team_matches.head())

# combine dataframes
team_goals_per_match = team_goals.merge(team_matches, how='inner', on='team').merge(wins_vs_appearances_exc_26, how='inner', left_on='team', right_on='team')

# display(team_goals_per_match)

# check number of matches is 1064 as expected
# print(team_goals_per_match['matches_played'].sum()/2)

# add new column for 'average goals per match'
team_goals_per_match['average_goals_per_match'] = round(team_goals_per_match['goals_scored'] / team_goals_per_match['matches_played'], 3)

# display(team_goals_per_match.sort_values(by='Wins', ascending=False))

# create scatter plot to show average goals per match vs number of wins for each team

team_goals_per_match_plot = px.scatter(team_goals_per_match, x='wins_per_appearance', y='average_goals_per_match', text='team', title='Average Goals per Match vs. Number of Wins per Appearance? (1930 - 2022)', color='wins_per_appearance', size='wins_per_appearance', labels={'average_goals_per_match':'Average Goals per Match', 'wins_per_appearance':'Number of Wins per Appearance'})

team_goals_per_match_plot.update_traces(textposition='top center')

team_goals_per_match_plot.show(config=config)

team_goals_per_match_matrix = px.imshow(team_goals_per_match[['average_goals_per_match', 'wins_per_appearance']].rename(columns={'average_goals_per_match':'Average Goals per Match', 'wins_per_appearance':'Number of Wins per Appearance'}).corr(), text_auto=True, zmin=-1, zmax=1)

team_goals_per_match_matrix.show(config=config)

team_goals_per_match_plot.write_html(f"{FINAL_PLOT_DIR}/team_goals_per_match.html")
team_goals_per_match_matrix.write_html(f"{FINAL_PLOT_DIR}/team_goals_per_match_matrix.html")
```

The correlation matrix shows a correlation of 0.47, which is closer to 0 than it is to 1, suggesting that there is some, but not a lot, of positive linearity between the average number of goals scored per match and the number of historical wins a team has had.

##### Initialise `MinMaxScaler`

```python
scaler = MinMaxScaler()
```

##### Create team attribute dataframe

```python
# create a dataframe to hold all the teams attribute scores

# team  AttackingScore  PhysicalScore
# England   90              30
# France    80              27
```

##### Define `PhysicalScore` for each team
Each team is given a `PhysicalScore` between 0 - 100, based on the attributes in the `physical_clean_df` dataframe

```python
physical_attributes = physical_clean_df.columns[1:]

# Create z-scores
physical_z_scores = StandardScaler().fit_transform(physical_clean_df[physical_attributes])

# Average z-scores
physical_clean_df['PhysicalZScore'] = physical_z_scores.mean(axis=1)

# Scale 0-100
physical_clean_df['PhysicalScore'] = scaler.fit_transform(physical_clean_df[['PhysicalZScore']]) * 100

# display(physical_clean_df.sort_values(by='PhysicalScore', ascending=False).head(10))
```

##### Define `AttackingScore` for each team
Each team is given an `AttackingScore` between 0 - 100, based on the attributes in the `attacking_clean_df` dataframe

```python
attacking_attributes = ['Goals', 'Assists', 'AttemptsOnTarget', 'AttemptsAtGoalConversionRatePC', 'AttemptsInsideThePenaltyArea', 'xG', 'xGEfficiency', 'PossessionControlPC'] # need to confirm which attributes we want to include here

# Create z-scores
attacking_z_scores = StandardScaler().fit_transform(attacking_clean_df[attacking_attributes])

# Average z-scores
attacking_clean_df['AttackingZScore'] = attacking_z_scores.mean(axis=1)

# Scale 0-100
attacking_clean_df['AttackingScore'] = scaler.fit_transform(attacking_clean_df[['AttackingZScore']]) * 100

# display(attacking_clean_df.sort_values(by='AttackingScore', ascending=False).head(10))
```

##### Define `DefendingScore` for each team
Each team is given an `DefendingScore` between 0 - 100, based on the attributes in the `defending_clean_df` dataframe

```python
defending_attributes = ['GoalsConceded', 'ForcedTurnovers','BallRecoveryTime', 'DefensivePressuresApplied', 'DefensivePressuresDirectlyApplied', 'DirectDefensivePressurePC'] # need to confirm which attributes we want to include here

# Create z-scores
defending_z_scores = pd.DataFrame(StandardScaler().fit_transform(defending_clean_df[defending_attributes]), columns=defending_attributes, index=defending_clean_df.index)

# inverse scores where lower value should contribute to higher score
defending_z_scores['BallRecoveryTime'] *= -1
defending_z_scores['GoalsConceded'] *= -1

# Average z-scores
defending_clean_df['DefendingZScore'] = defending_z_scores.mean(axis=1)

# Scale 0-100
defending_clean_df['DefendingScore'] = scaler.fit_transform(defending_clean_df[['DefendingZScore']]) * 100

# display(defending_clean_df.sort_values(by='DefendingScore', ascending=False).head(10))
```

##### Define `DisciplineScore` for each team
Each team is given an `DisciplineScore` between 0 - 100, based on the attributes in the `discipline_clean_df` dataframe

```python
discipline_attributes = ['FoulsAgainst', 'YellowCards', 'RedCards', 'IndirectRedCards', 'Offsides'] # need to confirm which attributes we want to include here

# Create z-scores
discipline_z_scores = pd.DataFrame(StandardScaler().fit_transform(discipline_clean_df[discipline_attributes]), columns=discipline_attributes, index=discipline_clean_df.index)

# inverse scores where lower value should contribute to higher score
discipline_z_scores *= -1

# Average z-scores
discipline_clean_df['DisciplineZScore'] = discipline_z_scores.mean(axis=1)

# Scale 0-100
discipline_clean_df['DisciplineScore'] = scaler.fit_transform(discipline_clean_df[['DisciplineZScore']]) * 100

# display(discipline_clean_df.sort_values(by='DisciplineScore', ascending=False).head(10))
```

##### Define `DistrubitionScore` for each team
Each team is given an `DistributionScore` between 0 - 100, based on the attributes in the `distribution_clean_df` dataframe

```python
distribution_attributes = ['PassesCompleted', 'PassingAccuracyPC', 'CrossingAccuracyPC', 'TakeOnsCompleted', 'DefensiveLineBreaksAccPC', 'SwitchesOfPlayAccPC'] # need to confirm which attributes we want to include here

# Create z-scores
distribution_z_scores = pd.DataFrame(StandardScaler().fit_transform(distribution_clean_df[distribution_attributes]), columns=distribution_attributes, index=distribution_clean_df.index)

# inverse scores where lower value should contribute to higher score

# Average z-scores
distribution_clean_df['DistributionZScore'] = distribution_z_scores.mean(axis=1)

# Scale 0-100
distribution_clean_df['DistributionScore'] = scaler.fit_transform(distribution_clean_df[['DistributionZScore']]) * 100

# display(distribution_clean_df.sort_values(by='DistributionScore', ascending=False).head(10))
```

##### Define `GoalkeepingScore` for each team
Each team is given an `GoalkeepingScore` between 0 - 100, based on the attributes in the `goalkeeping_clean_df` dataframe

```python
goalkeeping_attributes = ['CleanSheets','GoalsConceded','OutsidePenaltyAreaPC'] # need to confirm which attributes we want to include here

# Create z-scores
goalkeeping_z_scores = pd.DataFrame(StandardScaler().fit_transform(goalkeeping_clean_df[goalkeeping_attributes]), columns=goalkeeping_attributes, index=goalkeeping_clean_df.index)

# inverse scores where lower value should contribute to higher score
goalkeeping_z_scores['GoalsConceded'] *= -1

# Average z-scores
goalkeeping_clean_df['GoalkeepingZScore'] = goalkeeping_z_scores.mean(axis=1)

# Scale 0-100
goalkeeping_clean_df['GoalkeepingScore'] = scaler.fit_transform(goalkeeping_clean_df[['GoalkeepingZScore']]) * 100

# display(goalkeeping_clean_df.sort_values(by='GoalkeepingScore', ascending=False).head(10))
```

```python
movement_attributes = ['OffersToReceive', 'OffersInBehind', 'OffersInBetween', 'OffersOutsideTeamShape', 'ReceptionsInBehind', 'ReceptionsBetweenMidfieldAndDefensiveLine', 'ReceptionsUnderPressure'] # need to confirm which attributes we want to include here

# Create z-scores
movement_z_scores = pd.DataFrame(StandardScaler().fit_transform(movement_clean_df[movement_attributes]), columns=movement_attributes, index=movement_clean_df.index)

# inverse scores where lower value should contribute to higher score

# Average z-scores
movement_clean_df['MovementZScore'] = movement_z_scores.mean(axis=1)

# Scale 0-100
movement_clean_df['MovementScore'] = scaler.fit_transform(movement_clean_df[['MovementZScore']]) * 100

# display(movement_clean_df.sort_values(by='MovementScore', ascending=False).head(10))
```
