## Creating a Reusable Data-Quality Overview

A reusable function is created to provide the same initial quality checks for each dataset.

For each DataFrame, the function reports:

- number of rows and columns;
- data types;
- number of missing values;
- percentage of missing values; and
- number of exact duplicate rows.

```python
def dataset_overview(df, name):
    print(f"===== {name} =====")
    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")
    print()
    
    print("Data types:")
    print(df.dtypes)
    print()
    
    print("Missing values:")
    print(df.isna().sum())
    print()
    
    print("Missing percentages:")
    print((df.isna().mean() * 100).round(2))
    print()
    
    print("Exact duplicate rows:")
    print(df.duplicated().sum())
```

## Create Match Result Category

A categorical `result` variable is created from the home and away scores.

Completed matches are classified as:

- `Home Win` when the home score is greater than the away score;
- `Away Win` when the away score is greater than the home score;
- `Draw` when the scores are equal.

Only records where `match_completed` is `True` are classified. Records without known scores remain missing rather than being assigned an unsupported outcome.

This provides a clear match-outcome variable that can later be used for exploratory analysis and modelling.

```python
results_clean["result"] = pd.NA

completed = results_clean["match_completed"]

results_clean.loc[
    completed
    & (results_clean["home_score"] > results_clean["away_score"]),
    "result"
] = "Home Win"

results_clean.loc[
    completed
    & (results_clean["home_score"] < results_clean["away_score"]),
    "result"
] = "Away Win"

results_clean.loc[
    completed
    & (results_clean["home_score"] == results_clean["away_score"]),
    "result"
] = "Draw"
```

```python
# Check how many Home Wins, Away Wins, Draws and missing results there are.
results_clean["result"].value_counts(dropna=False)
```

## Investigating Missing Goal Minutes

The `minute` column is inspected for missing values.

A missing minute means that the exact timing of the goal is unavailable; it does not necessarily mean that the goal itself did not occur.

Deleting these records could remove genuine goals. Similarly, replacing missing minutes with zero, a mean or a median would fabricate goal timing and could bias later analysis of early, late or first goals.

The records are therefore retained with their original missing minute values.

```python
# Inspecting missing goal minutes
goalscorers_clean[goalscorers_clean["minute"].isna()].head(20)
```

- The missing goal minutes have not been imputed because the timing of these goals cannot be reliably inferred from the available data. Imputing an arbitrary, mean or median minute would introduce fabricated information and could bias later analysis of goal timing.

**Decision:** Retain these goal records and do not impute the missing minutes. The records remain useful for analyses that do not require exact goal timing.

# Cross-Dataset Validation

The datasets describe related aspects of the same football matches.

A match is identified using the combination:

`date + home_team + away_team`

Unique match keys are created from both the results and goalscorers datasets.

```python
result_match_keys = (
    results_clean[["date", "home_team", "away_team"]]
    .drop_duplicates()
)

goal_match_keys = (
    goalscorers_clean[["date", "home_team", "away_team"]]
    .drop_duplicates()
)
```

## Validate Goalscorer Match References

The unique matches represented in `goalscorers.csv` are compared with the matches in `results.csv`.

A left merge with an indicator is used to identify whether each goalscorer match key can be found in the results dataset.

Records classified as `left_only` would represent goal-event matches for which a corresponding results record was not found and would require investigation.

```python
goal_match_validation = goal_match_keys.merge(
    result_match_keys,
    on=["date", "home_team", "away_team"],
    how="left",
    indicator=True
)

goal_match_validation["_merge"].value_counts()
```

```python
goal_match_validation[
    goal_match_validation["_merge"] == "left_only"
]
```

## Reconciling Goal Events with Final Match Scores (Comparing total goal records with final scores)

The number of goal-event records is calculated for each match and compared with the total number of goals implied by the final score in `results.csv`.

This tests whether the event-level goalscorer data agrees with the match-level results data.

It also provides evidence for the earlier decision not to automatically remove duplicate-looking goal records. If the existing goal-event counts reconcile with the final scores, deleting apparently duplicated rows could incorrectly remove genuine goals.

```python
goals_per_match = (
    goalscorers_clean
    .groupby(["date", "home_team", "away_team"])
    .size()
    .reset_index(name="recorded_goals")
)

goal_score_validation = goals_per_match.merge(
    results_clean[
        [
            "date",
            "home_team",
            "away_team",
            "home_score",
            "away_score"
        ]
    ],
    on=["date", "home_team", "away_team"],
    how="left"
)

goal_score_validation["result_total_goals"] = (
    goal_score_validation["home_score"]
    + goal_score_validation["away_score"]
)

goal_score_validation["goals_match_score"] = (
    goal_score_validation["recorded_goals"]
    == goal_score_validation["result_total_goals"]
)

goal_score_validation[
    "goals_match_score"
].value_counts(dropna=False)
```

```python
goal_score_validation[
    goal_score_validation["goals_match_score"] == False
]
```

## Validating Home and Away Goals Separately

Matching the total number of goals is useful, but it does not prove that those goals have been attributed to the correct teams.

Goal events are therefore separated according to whether the scoring team is the home or away team. These counts are then compared independently with `home_score` and `away_score`.

```python
home_goal_counts = (
    goalscorers_clean[
        goalscorers_clean["team"]
        == goalscorers_clean["home_team"]
    ]
    .groupby(["date", "home_team", "away_team"])
    .size()
    .reset_index(name="recorded_home_goals")
)

away_goal_counts = (
    goalscorers_clean[
        goalscorers_clean["team"]
        == goalscorers_clean["away_team"]
    ]
    .groupby(["date", "home_team", "away_team"])
    .size()
    .reset_index(name="recorded_away_goals")
)
```

```python
team_goal_validation = goals_per_match[
    ["date", "home_team", "away_team"]
].copy()

team_goal_validation = team_goal_validation.merge(
    home_goal_counts,
    on=["date", "home_team", "away_team"],
    how="left"
)

team_goal_validation = team_goal_validation.merge(
    away_goal_counts,
    on=["date", "home_team", "away_team"],
    how="left"
)

team_goal_validation[
    ["recorded_home_goals", "recorded_away_goals"]
] = (
    team_goal_validation[
        ["recorded_home_goals", "recorded_away_goals"]
    ]
    .fillna(0)
    .astype(int)
)
```

```python
team_goal_validation = team_goal_validation.merge(
    results_clean[
        [
            "date",
            "home_team",
            "away_team",
            "home_score",
            "away_score"
        ]
    ],
    on=["date", "home_team", "away_team"],
    how="left"
)

team_goal_validation["home_goals_match"] = (
    team_goal_validation["recorded_home_goals"]
    == team_goal_validation["home_score"]
)

team_goal_validation["away_goals_match"] = (
    team_goal_validation["recorded_away_goals"]
    == team_goal_validation["away_score"]
)

print(
    team_goal_validation[
        "home_goals_match"
    ].value_counts(dropna=False)
)

print(
    team_goal_validation[
        "away_goals_match"
    ].value_counts(dropna=False)
)
```

## Validating Shootout Match References

Shootout records are matched against the results dataset using the same combination of date, home team and away team.

This checks whether every recorded shootout can be linked to a corresponding match in the main results dataset.

```python
shootout_match_keys = (
    shootouts_clean[
        ["date", "home_team", "away_team"]
    ]
    .drop_duplicates()
)

shootout_match_validation = shootout_match_keys.merge(
    result_match_keys,
    on=["date", "home_team", "away_team"],
    how="left",
    indicator=True
)

shootout_match_validation["_merge"].value_counts()
```

```python
shootout_match_validation[
    shootout_match_validation["_merge"] == "left_only"
]
```

## Comparing Shootouts with Recorded Match Scores

Shootout records are merged with their corresponding match scores to investigate whether the underlying match was recorded as level before the shootout outcome.

This check must be interpreted carefully because historical football datasets may use different conventions for recording scores around extra time and penalty shootouts.

Any unexpected records should therefore be investigated rather than automatically removed.

```python
shootout_score_validation = shootouts_clean.merge(
    results_clean[
        [
            "date",
            "home_team",
            "away_team",
            "home_score",
            "away_score"
        ]
    ],
    on=["date", "home_team", "away_team"],
    how="left"
)

shootout_score_validation["match_was_drawn"] = (
    shootout_score_validation["home_score"]
    == shootout_score_validation["away_score"]
)

shootout_score_validation[
    "match_was_drawn"
].value_counts(dropna=False)
```

```python
shootout_score_validation[
    shootout_score_validation["match_was_drawn"] == False
]
```

A final set of programmatic assertions is used to verify important logical rules after cleaning.

The checks confirm that:

- a team is never recorded as playing against itself;
- known scores are non-negative;
- every scoring team participated in its match;
- every shootout winner participated in its match;

# Cleaning Summary

## Results Dataset

The results dataset contains 49,509 international football match records.

The dataset was assessed for missing values, duplicates, invalid dates, impossible scores and inconsistent participating teams. Four records contain missing match scores. These scores were replaced by the legitimate values recorded on the Official FIFA Website. The records were updated and a `match_completed` indicator was created. 

It was also noticed that the 4 most recent FIFA World Cup Matches were not present in the dataset (Semi-finals, Bronze Final, and Final), therefore the values for these rows were found on the official FIFA Website, saved in a separate dataframe `missing_results_rows`, and appended to the results dataset. 

Dates were converted to datetime format and scores were converted to nullable integer values. Text fields were stripped of surrounding whitespace.

International city names containing accented and non-ASCII characters were investigated. These were largely legitimate local-language place names rather than encoding errors and were therefore preserved.

Additional analytical variables were created, including `total_goals`, `goal_difference`, `result`, `year` and `month`.

## Goalscorers Dataset

The goalscorers dataset contains 47,886 goal-event records.

There are 44 missing scorer names. These records were retained because the absence of the player's name does not invalidate the goal event, and other information about the goal remains useful.

There are also 254 missing goal-minute values. These were retained without imputation because inventing goal times could bias later timing-based analysis. A `minute_known` indicator was created to distinguish records with available timing information.

A total of 79 duplicate-looking records were identified. These were not automatically removed because separate genuine goals can appear identical when information such as goal minute is unavailable. Cross-dataset validation was therefore used to assess consistency with final match scores.

Goal dates and minutes were converted to appropriate data types, scoring teams were validated against participating teams, and penalty and own-goal indicators were inspected.

## Shootouts Dataset

The shootouts dataset contains 682 penalty shootout records.

The `first_shooter` field is missing for 422 records, representing approximately 61.9% of shootouts. These values were not imputed because there is no reliable method of determining which team took the first penalty from the available data.

The shootout records were retained because the participating teams and winner remain useful even when first-shooter information is unavailable. A `first_shooter_known` availability indicator was created.

All recorded winners and known first-shooter teams were validated to ensure that they correspond to one of the participating teams.

## Former Names Dataset

The former-names dataset contains 36 historical team-name mappings.

No missing values or exact duplicates were identified. Historical start and end dates were validated and converted to datetime format.

Repeated current team names are not treated as duplicates because one team can legitimately have multiple historical names.

Historical names have not been globally replaced with modern names. The mapping has instead been preserved for potential date-aware standardisation later in the project.

## Cross-Dataset Validation

The related datasets were validated against one another using the combination of `date`, `home_team` and `away_team` as a match identifier.

Goal-event records were compared with final match scores to assess whether recorded goals reconcile with match-level outcomes. Home and away goal totals were also checked separately.

Shootout records were compared with the results dataset to confirm that they correspond to known matches.

These checks provide stronger evidence of data quality than relying solely on independent checks within each dataset.

## Overall Cleaning Approach

The datasets are generally of high quality.

The cleaning process deliberately avoided unnecessary deletion and imputation. Missing or unusual values were preserved where the underlying record remained valid, and information was only changed when there was a clear and defensible reason to do so.

This approach reduces the risk of introducing fabricated information or accidentally removing genuine historical football events.

The cleaned datasets are now suitable for the next stage of the project: Exploratory data analysis and the identification of factors associated with FIFA World Cup success and match outcomes.
