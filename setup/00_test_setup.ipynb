## Test 1: Python Version

Let's verify you're running Python 3.10 or higher.

> **Important:** If you just installed UV, make sure you have **restarted your IDE (e.g., VS Code)** before running these tests. UV updates your system PATH, but your IDE won't see it until it restarts. Without this step, UV commands will not be available inside the IDE.

```python
import sys

print(f"Python version: {sys.version}")
print(f"Version info: {sys.version_info.major}.{sys.version_info.minor}.{sys.version_info.micro}")

if sys.version_info >= (3, 10):
    print("Python version is 3.10 or higher - Good to go!")
else:
    print("Python version is too old. Please upgrade to Python 3.10 or higher.")
```

## Test 2: Import Required Packages

Let's import all the packages we'll be using during the training.

```python
try:
    import pandas as pd
    print(f"pandas {pd.__version__} - imported successfully")
except ImportError as e:
    print(f"pandas - not found. Error: {e}")

try:
    import numpy as np
    print(f"numpy {np.__version__} - imported successfully")
except ImportError as e:
    print(f"numpy - not found. Error: {e}")

try:
    import plotly
    import plotly.express as px
    print(f"plotly {plotly.__version__} - imported successfully")
except ImportError as e:
    print(f"plotly - not found. Error: {e}")

try:
    import seaborn as sns
    print(f"seaborn {sns.__version__} - imported successfully")
except ImportError as e:
    print(f"seaborn - not found. Error: {e}")

try:
    import sklearn
    print(f"scikit-learn {sklearn.__version__} - imported successfully")
except ImportError as e:
    print(f"scikit-learn - not found. Error: {e}")

print("\nAll packages imported successfully!")
```

## Test 3: Create and Manipulate Data

Let's test basic pandas and numpy operations.

```python
# Create a simple DataFrame
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age': [25, 30, 35, 28],
    'score': [85, 92, 78, 95]
}

df = pd.DataFrame(data)
print("DataFrame created successfully!")
print("\nDataFrame contents:")
print(df)

# Test basic operations
average_age = df['age'].mean()
print(f"\nAverage age calculated: {average_age}")

# Test numpy operations
arr = np.array([1, 2, 3, 4, 5])
arr_sum = arr.sum()
print(f"NumPy array sum: {arr_sum}")
```

## Test 4: Plotting Capability

Let's verify that plotly can create plots.

```python
# Create a simple bar plot
import plotly.express as px

categories = ['A', 'B', 'C', 'D']
values = [23, 45, 37, 52]

fig = px.bar(
    x=categories,
    y=values,
    title='Test Plot',
    labels={'x': 'Category', 'y': 'Value'}
)
fig.update_layout(template='plotly_white')
fig.show()

print("Plot generated successfully!")
```
